```python
import os
import re
import pandas as pd
import hashlib
import random

class KeywordValidationSampler:
    def __init__(self, data_directory="./"):
        self.data_directory = data_directory
        self.all_articles = []
        self.articles = []  # 仅包含有关键词的文献
        
        # === CICES V5.2分类词典 ===
        self.cices_categories = {
            'Provisioning': {
                'terrestrial_crops': ['crops', 'cereals', 'root crops', 'soft fruit', 'fodder', 'livestock feed', 'fibres', 'materials', 'cultivated plants', 'fungi', 'algae', 'bacteria', 'plantation', 'energy source', 'bioenergy'],
                'aquatic_crops': ['aquaculture', 'in-situ aquaculture', 'cultivated aquatic plants'],
                'reared_animals': ['reared animals', 'beef', 'dairy', 'animal products', 'fibres from animals', 'mechanical energy', 'biogas'],
                'aquatic_animals': ['reared aquatic animals', 'in-situ aquaculture animals'],
                'wild_plants': ['wild plants', 'terrestrial plants', 'aquatic plants', 'wild fungi', 'wild algae', 'wild plant materials', 'wild plant energy'],
                'wild_animals': ['wild animals', 'terrestrial animals', 'aquatic animals', 'wild animal nutrition', 'wild animal materials'],
                'genetic_material': ['genetic material', 'seeds', 'spores', 'plant materials', 'breeding', 'new strains', 'varieties', 'individual genes', 'biological entities', 'animal genetic material', 'population establishment']
            },
            'Regulating': {
                'biochemical_transformation': ['bio-remediation', 'micro-organisms', 'pollution degradation', 'filtration', 'sequestration', 'storage', 'accumulation', 'waste mediation', 'toxic substances', 'anthropogenic origin', 'smell reduction', 'noise attenuation', 'visual screening'],
                'flow_extreme_events': ['erosion control', 'water erosion', 'wind erosion', 'hydrological cycle', 'water flow regulation', 'runoff', 'base flows', 'peak flows', 'hazard mitigation', 'mass movement', 'landslides', 'flood mitigation', 'surge protection', 'wind protection', 'fire protection'],
                'lifecycle_maintenance': ['pollination', 'gamete dispersal', 'seed dispersal', 'nursery populations', 'breeding grounds', 'gene pool protection', 'refuge habitats', 'feeding grounds'],
                'pest_disease_control': ['pest control', 'invasive species', 'disease control'],
                'soil_quality': ['soil quality', 'weathering processes', 'decomposition', 'fixing processes', 'soil structure', 'biological agents', 'soil porosity'],
                'water_conditions': ['water conditions', 'chemical condition', 'macronutrients', 'freshwater regulation', 'saltwater regulation'],
                'atmospheric_conditions': ['atmospheric composition', 'oceanic circulation', 'climate change', 'temperature regulation', 'humidity', 'ventilation', 'transpiration', 'microclimate']
            },
            'Cultural': {
                'physical_experiential': ['recreational activities', 'health promotion', 'recuperation', 'enjoyment', 'active interactions', 'immersive interactions', 'passive interactions', 'observational interactions'],
                'intellectual_representative': ['scientific investigation', 'traditional ecological knowledge', 'genetic diversity', 'education', 'training', 'culture', 'heritage', 'cultural resonance', 'aesthetic experiences'],
                'indirect_interactions': ['entertainment', 'representation outside setting'],
                'spiritual_symbolic': ['symbolic meaning', 'sense of place', 'distinctiveness', 'spiritual meaning', 'religious meaning', 'contemporary existence', 'conservation importance', 'inter-generational existence', 'genetic diversity conservation']
            }
        }
        
        # === 纯数据科学与AI关键词词典 (不含传统统计词汇) ===
        self.ai_keywords = [
            "machine learning", "supervised learning", "unsupervised learning", "semi supervised learning", "self training", "active learning", "reinforcement learning", "transfer learning", "continual learning", "ensemble learning", "ensemble methods", "meta learning",
            "deep learning", "neural network", "artificial neural network", "ANN", "deep neural network", "DNN", "convolutional neural network", "CNN", "recurrent neural network", "RNN", "LSTM", "Long Short-Term Memory", "GRU", "gated recurrent unit", "transformer", "transformers", "attention mechanism", "BERT", "GPT", "generative adversarial network", "GAN", "VAE", "variational autoencoder",
            "decision tree", "random forest", "support vector machine", "SVM", "k-means", "k-nearest neighbor", "KNN", "naive bayes", "bayesian network", "hierarchical clustering", "DBSCAN", "gaussian mixture model", "GMM", "principal component analysis", "PCA", "independent component analysis", "ICA", "linear discriminant analysis", "LDA", "factor analysis",
            "data mining","big data", "predictive modeling", "statistical modeling", "pattern recognition", "feature selection", "feature extraction", "feature engineering", "dimensionality reduction", "anomaly detection", "outlier detection", "change detection",
            "computer vision", "image processing", "object detection", "image classification", "semantic segmentation", "instance segmentation", "image recognition", "facial recognition", "object recognition", "pattern matching", "edge detection", "feature matching", "optical character recognition", "OCR",
            "natural language processing", "NLP", "text mining", "text analysis", "sentiment analysis", "topic modeling", "information extraction", "named entity recognition", "NER", "text classification", "text clustering", "machine translation", "question answering", "text generation",
            "genetic algorithm", "evolutionary computation", "particle swarm optimization", "PSO", "ant colony optimization", "ACO", "simulated annealing", "gradient boosting",
            "time series analysis", "time series forecasting", "ARIMA", "SARIMA", "predictive model", "trend analysis", "Kalman filter",
            "expert system", "knowledge base", "inference engine", "rule based system", "fuzzy logic", "fuzzy system", "decision support system", "DSS", "semantic network", "knowledge graph",
            "explainable AI", "XAI", "interpretable machine learning", "LIME", "SHAP", "federated learning", "adversarial training", "domain adaptation", "few-shot learning", "zero-shot learning", "multi-task learning", "automated machine learning", "AutoML", "neural architecture search", "NAS",
            "ML", "DL", "RL", "AI", "ANN", "CNN", "RNN", "SVM", "KNN", "PCA", "ICA", "LDA", "GAN", "VAE", "NLP", "OCR", "PSO", "ACO", "SGD", "XAI"
        ]

    def generate_article_hash(self, article):
        """修复点：完整继承主分析程序的 Hash 逻辑，包含 Journal 字段"""
        identifier_parts = []
        if article.get('title'):
            title_cleaned = re.sub(r'\s+', ' ', article['title'].lower().strip())
            identifier_parts.append(title_cleaned)
        if article.get('year'):
            identifier_parts.append(article['year'])
        if article.get('authors') and len(article['authors']) > 0:
            first_author_cleaned = re.sub(r'\s+', ' ', article['authors'][0].lower().strip())
            identifier_parts.append(first_author_cleaned)
        if article.get('journal'):
            journal_cleaned = re.sub(r'\s+', ' ', article['journal'].lower().strip())
            identifier_parts.append(journal_cleaned)
            
        if len(identifier_parts) < 2:
            identifier_parts.append(str(hash(str(article))))
            
        identifier_string = '||'.join(identifier_parts)
        return hashlib.md5(identifier_string.encode('utf-8')).hexdigest()

    def read_literature_files(self):
        files = [f for f in os.listdir(self.data_directory) if f.endswith('.txt') and 'download' in f and any(year in f for year in ['2015', '2016', '2017', '2018', '2019', '2020', '2021', '2022', '2023', '2024', '2025'])]
        files.sort()
        print(f"📚 找到 {len(files)} 个文献文件，正在读取...")
        for file in files:
            file_path = os.path.join(self.data_directory, file)
            try:
                with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
                    self.parse_articles(f.read(), file)
            except Exception as e:
                pass

    def parse_articles(self, content, filename):
        """修复点：与主分析程序对齐字典结构，加入 filename, abstract, journal 等字段以稳定哈希值"""
        articles = re.split(r'\nER\s*\n', content)
        for article_text in articles:
            if not article_text.strip() or not re.search(r'(^|\n)PT\s+', article_text):
                continue
            
            article_data = {
                'filename': filename,
                'keywords': set(),
                'keywords_plus': set(),
                'title': '',
                'year': '',
                'abstract': '',
                'journal': '',
                'authors': []
            }
            
            ti_match = re.search(r'\nTI\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if ti_match: article_data['title'] = ti_match.group(1).replace('\n', ' ').strip()
                
            py_match = re.search(r'\nPY\s+(\d{4})', article_text)
            if py_match: article_data['year'] = py_match.group(1)
                
            au_match = re.search(r'\nAU\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if au_match: article_data['authors'] = [author.strip() for author in au_match.group(1).split('\n') if author.strip()]
            
            ab_match = re.search(r'\nAB\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if ab_match: article_data['abstract'] = ab_match.group(1).replace('\n', ' ').strip()

            so_match = re.search(r'\nSO\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if so_match: article_data['journal'] = so_match.group(1).replace('\n', ' ').strip()
            
            id_match = re.search(r'\nID\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if id_match: article_data['keywords_plus'].update(self.extract_keywords(id_match.group(1)))
                
            de_match = re.search(r'\nDE\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if de_match: article_data['keywords'].update(self.extract_keywords(de_match.group(1)))

            if article_data['keywords'] or article_data['keywords_plus']:
                article_data['all_keywords'] = article_data['keywords'] | article_data['keywords_plus']
                self.articles.append(article_data)

    def extract_keywords(self, field_content):
        field_content = field_content.replace('\n', ' ')
        return [re.sub(r'\s+', ' ', kw.strip().lower()).strip('.,;()[]{}"\'-') for kw in re.split(r'[;\n]', field_content) if kw.strip() and len(kw.strip()) > 2]

    def is_complete_word_match(self, article_keyword, cices_keyword):
        article_kw = article_keyword.lower().strip()
        cices_kw = cices_keyword.lower().strip()
        if article_kw == cices_kw: return True
        
        pattern = r'(?:^|\s)' + re.escape(cices_kw) + r'(?:\s|$)' if ' ' in cices_kw else r'\b' + re.escape(cices_kw) + r'\b'
        if bool(re.search(pattern, article_kw)): return True
            
        reverse_pattern = r'(?:^|\s)' + re.escape(article_kw) + r'(?:\s|$)' if ' ' in article_kw else r'\b' + re.escape(article_kw) + r'\b'
        return bool(re.search(reverse_pattern, cices_kw))

    def is_ai_keyword_match(self, search_text, ai_keyword):
        search_text = search_text.lower().strip()
        ai_kw = ai_keyword.lower().strip()
        if search_text == ai_kw: return True
        
        strict_match_keywords = ['ml', 'dl', 'rl', 'rnn', 'cnn', 'lstm']
        if ai_kw in strict_match_keywords or len(ai_kw) <= 3:
            return bool(re.search(r'(?:^|[^a-zA-Z])' + re.escape(ai_kw) + r'(?:[^a-zA-Z]|$)', search_text))
        if ' ' in ai_kw:
            return bool(re.search(r'(?:^|\s)' + re.escape(ai_kw) + r'(?:\s|$)', search_text))
        return bool(re.search(r'\b' + re.escape(ai_kw) + r'\b', search_text))

    def perform_matching_and_sampling(self):
        print("\n🔍 正在使用最新规则进行精准匹配...")
        
        pools = {
            'Regulating': [],
            'Cultural': [],
            'Provisioning': [],
            'DataScience_AI': []
        }
        
        added_hashes = {k: set() for k in pools.keys()}
        priority_order = ['Regulating', 'Cultural', 'Provisioning']
        
        for article in self.articles:
            article_hash = self.generate_article_hash(article)
            
            # 1. CICES 匹配
            matched_cices = None
            matched_cices_keywords = []
            
            for main_category in priority_order:
                for subcategory, keywords in self.cices_categories[main_category].items():
                    for article_keyword in article['all_keywords']:
                        for cices_keyword in keywords:
                            if self.is_complete_word_match(article_keyword, cices_keyword):
                                matched_cices = main_category
                                matched_cices_keywords.append(f"[{cices_keyword}] (hit: {article_keyword})")
                if matched_cices:
                    break
            
            if matched_cices and article_hash not in added_hashes[matched_cices]:
                pools[matched_cices].append({
                    'Title': article['title'],
                    'Category': matched_cices,
                    'Matched_Keywords': " | ".join(set(matched_cices_keywords))
                })
                added_hashes[matched_cices].add(article_hash)
            
            # 2. AI / 数据科学匹配 (🚨 仅使用严格的 ai_keywords)
            keywords_text = ' '.join(article['all_keywords']).lower()
            matched_ai_kws = []
            
            for ai_kw in self.ai_keywords:
                if self.is_ai_keyword_match(keywords_text, ai_kw):
                    matched_ai_kws.append(ai_kw)
            
            if matched_ai_kws and article_hash not in added_hashes['DataScience_AI']:
                pools['DataScience_AI'].append({
                    'Title': article['title'],
                    'Category': 'DataScience_AI',
                    'Matched_Keywords': " | ".join(set(matched_ai_kws))
                })
                added_hashes['DataScience_AI'].add(article_hash)

        print("-" * 50)
        print("📊 匹配结果统计 (供核对一致性):")
        for cat in priority_order + ['DataScience_AI']:
            print(f"  - {cat:<15}: {len(pools[cat])} 篇")
        print("-" * 50)
        
        sampled_results = []
        for cat in priority_order + ['DataScience_AI']:
            sample_size = min(25, len(pools[cat]))
            sampled_items = random.sample(pools[cat], sample_size)
            sampled_results.extend(sampled_items)
            print(f"✅ 从 {cat} 中随机抽取了 {sample_size} 篇用于人工验证")

        if sampled_results:
            df = pd.DataFrame(sampled_results)
            output_file = 'keyword_matching_validation_sample_corrected.csv'
            df.to_csv(output_file, index=False, encoding='utf-8-sig')
            print(f"\n💾 抽样验证文件已保存至: {output_file}")

if __name__ == "__main__":
    sampler = KeywordValidationSampler()
    sampler.read_literature_files()
    sampler.perform_matching_and_sampling()
```

```python
import os
import re
import pandas as pd
from collections import defaultdict
import random

class EcosystemServicesAuthorInstAnalyzer:
    def __init__(self, data_directory="./"):
        self.data_directory = data_directory
        self.articles = []
        
        # === 纯净版：现代数据科学与先进AI关键词词典 (🚨 已剔除回归、分类、聚类等传统统计词) ===
        self.ai_keywords = [
            "machine learning", "supervised learning", "unsupervised learning", "semi supervised learning", "self training", "active learning", "reinforcement learning", "transfer learning", "continual learning", "ensemble learning", "ensemble methods", "meta learning",
            "deep learning", "neural network", "artificial neural network", "ANN", "deep neural network", "DNN", "convolutional neural network", "CNN", "recurrent neural network", "RNN", "LSTM", "Long Short-Term Memory", "GRU", "gated recurrent unit", "transformer", "transformers", "attention mechanism", "BERT", "GPT", "generative adversarial network", "GAN", "VAE", "variational autoencoder",
            "decision tree", "random forest", "support vector machine", "SVM", "k-means", "k-nearest neighbor", "KNN", "naive bayes", "bayesian network", "hierarchical clustering", "DBSCAN", "gaussian mixture model", "GMM", "principal component analysis", "PCA", "independent component analysis", "ICA", "linear discriminant analysis", "LDA", "factor analysis",
            "data mining","big data", "predictive modeling", "statistical modeling", "pattern recognition", "feature selection", "feature extraction", "feature engineering", "dimensionality reduction", "anomaly detection", "outlier detection", "change detection",
            "computer vision", "image processing", "object detection", "image classification", "semantic segmentation", "instance segmentation", "image recognition", "facial recognition", "object recognition", "pattern matching", "edge detection", "feature matching", "optical character recognition", "OCR",
            "natural language processing", "NLP", "text mining", "text analysis", "sentiment analysis", "topic modeling", "information extraction", "named entity recognition", "NER", "text classification", "text clustering", "machine translation", "question answering", "text generation",
            "genetic algorithm", "evolutionary computation", "particle swarm optimization", "PSO", "ant colony optimization", "ACO", "simulated annealing", "gradient boosting",
            "time series analysis", "time series forecasting", "ARIMA", "SARIMA", "predictive model", "trend analysis", "Kalman filter",
            "expert system", "knowledge base", "inference engine", "rule based system", "fuzzy logic", "fuzzy system", "decision support system", "DSS", "semantic network", "knowledge graph",
            "explainable AI", "XAI", "interpretable machine learning", "LIME", "SHAP", "federated learning", "adversarial training", "domain adaptation", "few-shot learning", "zero-shot learning", "multi-task learning", "automated machine learning", "AutoML", "neural architecture search", "NAS",
            "ML", "DL", "RL", "AI", "ANN", "CNN", "RNN", "SVM", "KNN", "PCA", "ICA", "LDA", "GAN", "VAE", "NLP", "OCR", "PSO", "ACO", "SGD", "XAI"
        ]

    def read_literature_files(self):
        files = [f for f in os.listdir(self.data_directory) if f.endswith('.txt') and 'download' in f]
        print(f"📚 找到 {len(files)} 个文献文件，正在解析...")
        for file in files:
            try:
                with open(os.path.join(self.data_directory, file), 'r', encoding='utf-8', errors='ignore') as f:
                    self.parse_articles(f.read())
            except Exception:
                pass
        print(f"✅ 解析完毕。具有关键词的有效文献：{len(self.articles)} 篇。")

    def parse_articles(self, content):
        for article_text in re.split(r'\nER\s*\n', content):
            if not article_text.strip() or not re.search(r'(^|\n)PT\s+', article_text): continue
            
            article = {'keywords': set(), 'keywords_plus': set(), 'authors': [], 'institutions': []}
            
            py = re.search(r'\nPY\s+(\d{4})', article_text)
            article['year'] = int(py.group(1)) if py else 9999
            
            # 提取作者
            au = re.search(r'\nAU\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if au:
                article['authors'] = [a.strip().upper() for a in au.group(1).split('\n') if a.strip()]
            
            # 提取并清洗机构 (C1字段)
            c1 = re.search(r'\nC1\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if c1:
                article['institutions'] = self.clean_institutions(c1.group(1))
            
            # 提取关键词
            de = re.search(r'\nDE\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if de: article['keywords'].update(self.extract_kws(de.group(1)))
                
            id_match = re.search(r'\nID\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if id_match: article['keywords_plus'].update(self.extract_kws(id_match.group(1)))

            if article['keywords'] or article['keywords_plus']:
                article['all_keywords'] = article['keywords'] | article['keywords_plus']
                self.articles.append(article)

    def extract_kws(self, text):
        return [re.sub(r'\s+', ' ', kw.strip().lower()).strip('.,;()[]{}"\'-') for kw in re.split(r'[;\n]', text.replace('\n', ' ')) if len(kw.strip()) > 2]

    def clean_institutions(self, c1_text):
        """清洗WoS机构名称：剥离作者括号，取机构主干（遇到第一个逗号截止）"""
        insts = set()
        addresses = re.split(r'[;\n]', c1_text)
        for addr in addresses:
            addr = addr.strip()
            if not addr: continue
            # 移除 [Author1, Author2] 结构
            if ']' in addr:
                addr = addr.split(']', 1)[1].strip()
            # 提取大学/研究机构主名（通常在第一个逗号之前）
            primary_inst = addr.split(',')[0].strip().upper()
            # 过滤掉一些明显的杂音
            primary_inst = re.sub(r'^\W+|\W+$', '', primary_inst)
            if len(primary_inst) > 3:
                insts.add(primary_inst)
        return list(insts)

    def is_ai_keyword_match(self, keywords_text, ai_keyword):
        """核心先进AI短语匹配逻辑 (与全项目保持绝对一致)"""
        ai_kw = ai_keyword.lower()
        strict_keys = ['ml', 'dl', 'rl', 'rnn', 'cnn', 'lstm', 'svm', 'knn', 'pca', 'ica', 'lda', 'gan', 'vae', 'nlp', 'ocr', 'pso', 'aco', 'sgd', 'xai']
        
        if ai_kw in strict_keys or len(ai_kw) <= 3:
            return bool(re.search(r'(?:^|[^a-zA-Z])' + re.escape(ai_kw) + r'(?:[^a-zA-Z]|$)', keywords_text))
        pattern = r'(?:^|\s)' + re.escape(ai_kw) + r'(?:\s|$)' if ' ' in ai_kw else r'\b' + re.escape(ai_kw) + r'\b'
        return bool(re.search(pattern, keywords_text))

    def analyze_entities(self):
        print("\n🔍 正在聚合作者与机构发文数据，并检测 先进AI 关键词...")
        
        # 结构: stats[name] = {'total': 0, 'ai_total': 0, 'first_ai_year': 9999}
        author_stats = defaultdict(lambda: {'total': 0, 'ai_total': 0, 'first_ai_year': 9999})
        inst_stats = defaultdict(lambda: {'total': 0, 'ai_total': 0, 'first_ai_year': 9999})
        
        for art in self.articles:
            year = art['year']
            kws_str = " ".join(art['all_keywords'])
            
            # 判断该文献是否为 先进AI 文献
            is_ai = False
            for ai_kw in self.ai_keywords:
                if self.is_ai_keyword_match(kws_str, ai_kw):
                    is_ai = True
                    break
            
            # 更新作者统计
            for author in art['authors']:
                author_stats[author]['total'] += 1
                if is_ai:
                    author_stats[author]['ai_total'] += 1
                    if year < author_stats[author]['first_ai_year']:
                        author_stats[author]['first_ai_year'] = year
            
            # 更新机构统计
            for inst in art['institutions']:
                inst_stats[inst]['total'] += 1
                if is_ai:
                    inst_stats[inst]['ai_total'] += 1
                    if year < inst_stats[inst]['first_ai_year']:
                        inst_stats[inst]['first_ai_year'] = year

        return author_stats, inst_stats

    def get_stratified_samples(self, stats_dict):
        """对实体进行阶层抽样：高产(Top20)、中产(4-15篇中抽20)、低产(1-3篇中抽20)"""
        records = []
        for name, stats in stats_dict.items():
            if name == '': continue
            ai_rate = (stats['ai_total'] / stats['total'] * 100) if stats['total'] > 0 else 0
            records.append({
                'Name': name,
                'Total_Articles': stats['total'],
                'AI_Articles': stats['ai_total'],
                'AI_Rate_%': round(ai_rate, 2),
                'First_AI_Year': stats['first_ai_year'] if stats['first_ai_year'] != 9999 else 'N/A'
            })
        
        # 按总发文量降序排序
        records.sort(key=lambda x: x['Total_Articles'], reverse=True)
        
        # 高产组：Top 20
        high_yield = records[:20]
        for r in high_yield: r['Tier'] = 'High-Yield (Top)'
        
        # 中产组：发文量在 4 到 15 之间 (随机抽取20个)
        mid_pool = [r for r in records if 4 <= r['Total_Articles'] <= 15]
        mid_yield = random.sample(mid_pool, min(20, len(mid_pool)))
        mid_yield.sort(key=lambda x: x['Total_Articles'], reverse=True)
        for r in mid_yield: r['Tier'] = 'Medium-Yield (4-15)'
        
        # 低产组：发文量在 1 到 3 之间 (随机抽取20个)
        low_pool = [r for r in records if 1 <= r['Total_Articles'] <= 3]
        low_yield = random.sample(low_pool, min(20, len(low_pool)))
        low_yield.sort(key=lambda x: x['Total_Articles'], reverse=True)
        for r in low_yield: r['Tier'] = 'Low-Yield (1-3)'
        
        return high_yield + mid_yield + low_yield

    def print_and_export_results(self, author_samples, inst_samples):
        print("\n" + "="*80)
        print("🏆 机构与作者技术采纳 (先进AI) 分析报告 (阶层抽样)")
        print("="*80)
        
        def print_table(samples, title):
            print(f"\n📌 {title}")
            print("-" * 85)
            print(f"{'Tier':<20} | {'Name':<35} | {'Total':<6} | {'AI':<4} | {'Rate(%)':<8} | {'1st Year':<8}")
            print("-" * 85)
            for r in samples:
                # 截断过长的名字以保证打印美观
                display_name = r['Name'][:32] + "..." if len(r['Name']) > 35 else r['Name']
                print(f"{r['Tier']:<20} | {display_name:<35} | {r['Total_Articles']:<6} | {r['AI_Articles']:<4} | {r['AI_Rate_%']:<8.2f} | {r['First_AI_Year']:<8}")
                
        print_table(inst_samples, "研究机构数据 (Institutions)")
        print_table(author_samples, "作者数据 (Authors)")

        # 导出到 Excel
        with pd.ExcelWriter('ecosystem_ai_adoption_strata_optimized.xlsx', engine='openpyxl') as writer:
            df_authors = pd.DataFrame(author_samples)
            df_inst = pd.DataFrame(inst_samples)
            
            # 调整列顺序
            cols = ['Tier', 'Name', 'Total_Articles', 'AI_Articles', 'AI_Rate_%', 'First_AI_Year']
            df_authors = df_authors[cols]
            df_inst = df_inst[cols]
            
            df_inst.to_excel(writer, sheet_name='Institutions_Adoption', index=False)
            df_authors.to_excel(writer, sheet_name='Authors_Adoption', index=False)

        print("\n💾 详细分层数据已导出至: ecosystem_ai_adoption_strata_optimized.xlsx")
        print("💡 提示: 您可以在 Excel 中对比 高产/中产/低产 组的平均 AI Rate，以验证您的猜想！")

if __name__ == "__main__":
    analyzer = EcosystemServicesAuthorInstAnalyzer()
    analyzer.read_literature_files()
    
    author_stats, inst_stats = analyzer.analyze_entities()
    
    # 提取分层样本
    author_samples = analyzer.get_stratified_samples(author_stats)
    inst_samples = analyzer.get_stratified_samples(inst_stats)
    
    # 打印并保存结果
    analyzer.print_and_export_results(author_samples, inst_samples)
```

```python
如何解读跑出来的数据来回答你的猜想？
当你运行代码并打开生成的 ecosystem_ai_adoption_strata.xlsx 后，你可以按照以下几个维度去验证你的猜想：

计算均值对比：
分别算出 High-Yield、Medium-Yield 和 Low-Yield 三个组别的 AI_Rate_% 的平均值。

如果 High > Medium > Low，说明“马太效应”显著，大组更喜欢/更有能力用AI。

如果 Low 或 Medium 的均值惊人地高，说明存在一批“新锐小团队”专攻 AI+生态。

观察 First_AI_Year（首篇发文年份）：
看看顶级大牛（High-Yield）第一次用AI是在哪一年（比如 2018 年）？中低产作者是不是近两年（2023-2024）才突然冒出来的？这能反映出技术的**扩散路径（Diffusion of Innovations）**是从上至下，还是边缘突破。
```
