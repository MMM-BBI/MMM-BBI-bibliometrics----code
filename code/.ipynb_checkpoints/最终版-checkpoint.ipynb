```python
import os
import re
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
from collections import Counter, defaultdict
import numpy as np

# 设置字体为Times New Roman
plt.rcParams["font.family"] = "Times New Roman"
plt.rcParams["axes.unicode_minus"] = False

class EcosystemServicesComprehensiveAnalyzer:
    def __init__(self, data_directory="./"):
        self.data_directory = data_directory
        self.all_articles = []  # 所有文献（包括没有关键词的）
        self.articles = []  # 有关键词的文献 - 这些就是我们要分析的文献
        
        # 严格按照CICES V5.2和您提供的关键词列表
        self.cices_categories = {
            'Provisioning': {
                'terrestrial_crops': [
                    'crops', 'cereals', 'root crops', 'soft fruit', 'fodder', 'livestock feed',
                    'fibres', 'materials', 'cultivated plants', 'fungi', 'algae', 'bacteria',
                    'plantation', 'energy source', 'bioenergy'
                ],
                'aquatic_crops': [
                    'aquaculture', 'in-situ aquaculture', 'cultivated aquatic plants'
                ],
                'reared_animals': [
                    'reared animals', 'beef', 'dairy', 'animal products',
                    'fibres from animals', 'mechanical energy', 'biogas'
                ],
                'aquatic_animals': [
                    'reared aquatic animals', 'in-situ aquaculture animals'
                ],
                'wild_plants': [
                    'wild plants', 'terrestrial plants', 'aquatic plants', 'wild fungi', 'wild algae',
                    'wild plant materials', 'wild plant energy'
                ],
                'wild_animals': [
                    'wild animals', 'terrestrial animals', 'aquatic animals',
                    'wild animal nutrition', 'wild animal materials'
                ],
                'genetic_material': [
                    'genetic material', 'seeds', 'spores', 'plant materials',
                    'breeding', 'new strains', 'varieties',
                    'individual genes', 'biological entities',
                    'animal genetic material', 'population establishment'
                ]
            },
            'Regulating': {
                'biochemical_transformation': [
                    'bio-remediation', 'micro-organisms', 'pollution degradation',
                    'filtration', 'sequestration', 'storage', 'accumulation',
                    'waste mediation', 'toxic substances', 'anthropogenic origin',
                    'smell reduction', 'noise attenuation', 'visual screening'
                ],
                'flow_extreme_events': [
                    'erosion control', 'water erosion', 'wind erosion',
                    'hydrological cycle', 'water flow regulation', 'runoff', 'base flows', 'peak flows',
                    'hazard mitigation', 'mass movement', 'landslides', 'flood mitigation', 'surge protection',
                    'wind protection', 'fire protection'
                ],
                'lifecycle_maintenance': [
                    'pollination', 'gamete dispersal', 'seed dispersal',
                    'nursery populations', 'breeding grounds', 'gene pool protection',
                    'refuge habitats', 'feeding grounds'
                ],
                'pest_disease_control': [
                    'pest control', 'invasive species', 'disease control'
                ],
                'soil_quality': [
                    'soil quality', 'weathering processes', 'decomposition', 'fixing processes',
                    'soil structure', 'biological agents', 'soil porosity'
                ],
                'water_conditions': [
                    'water conditions', 'chemical condition', 'macronutrients',
                    'freshwater regulation', 'saltwater regulation'
                ],
                'atmospheric_conditions': [
                    'atmospheric composition', 'oceanic circulation', 'climate change',
                    'temperature regulation', 'humidity', 'ventilation', 'transpiration', 'microclimate'
                ]
            },
            'Cultural': {
                'physical_experiential': [
                    'recreational activities', 'health promotion', 'recuperation', 'enjoyment',
                    'active interactions', 'immersive interactions',
                    'passive interactions', 'observational interactions'
                ],
                'intellectual_representative': [
                    'scientific investigation', 'traditional ecological knowledge', 'genetic diversity',
                    'education', 'training',
                    'culture', 'heritage', 'cultural resonance',
                    'aesthetic experiences'
                ],
                'indirect_interactions': [
                    'entertainment', 'representation outside setting'
                ],
                'spiritual_symbolic': [
                    'symbolic meaning', 'sense of place', 'distinctiveness',
                    'spiritual meaning', 'religious meaning',
                    'contemporary existence', 'conservation importance',
                    'inter-generational existence', 'genetic diversity conservation'
                ]
            }
        }
        
    def read_literature_files(self):
        """读取所有文献文件"""
        files = []
        expected_total = 0  # 根据文件名统计预期文献总数
        
        for f in os.listdir(self.data_directory):
            if f.endswith('.txt') and 'download' in f and any(year in f for year in 
                ['2015', '2016', '2017', '2018', '2019', '2020', '2021', '2022', '2023', '2024', '2025']):
                files.append(f)
                # 从文件名提取预期文献数
                match = re.search(r'u(\d+)\.txt', f)
                if match:
                    expected_total += int(match.group(1))
        
        files.sort()
        print(f"📚 找到 {len(files)} 个文献文件")
        print(f"📁 文件列表: {files}")
        print(f"📊 根据文件名预期文献总数: {expected_total}")
        print("-" * 80)
        
        for file in files:
            file_path = os.path.join(self.data_directory, file)
            try:
                with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
                    content = f.read()
                    count = self.parse_articles(content, file)
                    
                    # 获取预期数量
                    match = re.search(r'u(\d+)\.txt', file)
                    expected = int(match.group(1)) if match else 0
                    status = "✅" if count == expected else "⚠️"
                    print(f"{status} {file}: 预期 {expected} 篇, 实际解析 {count} 篇")
                    
            except Exception as e:
                print(f"❌ 读取文件 {file} 时出错: {e}")
        
        print("-" * 80)
        print(f"📊 统计结果:")
        print(f"  - 所有文献总数: {len(self.all_articles)} 篇")
        print(f"  - 有关键词的文献: {len(self.articles)} 篇")
        print(f"  - 预期文献总数: {expected_total} 篇")
        
        if len(self.all_articles) != expected_total:
            print(f"  ⚠️ 警告：实际解析总数与预期不符！差异: {len(self.all_articles) - expected_total}")
        
    def parse_articles(self, content, filename):
        """解析文章内容"""
        # WOS格式使用ER作为记录结束标记
        articles = re.split(r'\nER\s*\n', content)
        
        file_article_count = 0
        
        for article_text in articles:
            if not article_text.strip():
                continue
            
            # 检查是否包含PT字段（记录开始标记）
            if not re.search(r'(^|\n)PT\s+', article_text):
                continue
            
            file_article_count += 1
            
            article_data = {
                'filename': filename,
                'keywords': set(),  # DE字段关键词
                'keywords_plus': set(),  # ID字段关键词
                'title': '',
                'year': '',
                'abstract': '',
                'journal': '',
                'authors': [],
                'citations': 0,
                'references': []
            }
            
            # 提取ID字段 (Keywords Plus)
            id_match = re.search(r'\nID\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if id_match:
                id_content = id_match.group(1)
                keywords_plus = self.extract_keywords_from_field(id_content)
                article_data['keywords_plus'].update(keywords_plus)
            
            # 提取DE字段 (Author's keywords)
            de_match = re.search(r'\nDE\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if de_match:
                de_content = de_match.group(1)
                keywords = self.extract_keywords_from_field(de_content)
                article_data['keywords'].update(keywords)
            
            # 提取标题
            ti_match = re.search(r'\nTI\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if ti_match:
                article_data['title'] = ti_match.group(1).replace('\n', ' ').strip()
            
            # 提取年份
            py_match = re.search(r'\nPY\s+(\d{4})', article_text)
            if py_match:
                article_data['year'] = py_match.group(1)
            
            # 提取摘要
            ab_match = re.search(r'\nAB\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if ab_match:
                article_data['abstract'] = ab_match.group(1).replace('\n', ' ').strip()
            
            # 提取期刊名称
            so_match = re.search(r'\nSO\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if so_match:
                article_data['journal'] = so_match.group(1).replace('\n', ' ').strip()
            
            # 提取作者
            au_match = re.search(r'\nAU\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if au_match:
                authors_text = au_match.group(1)
                authors = [author.strip() for author in authors_text.split('\n') if author.strip()]
                article_data['authors'] = authors
            
            # 提取引用次数 (TC字段)
            tc_match = re.search(r'\nTC\s+(\d+)', article_text)
            if tc_match:
                article_data['citations'] = int(tc_match.group(1))
            
            # 添加到所有文献列表
            self.all_articles.append(article_data)
            
            # 只有当文章有关键词时才添加到articles列表
            if article_data['keywords'] or article_data['keywords_plus']:
                # 合并keywords和keywords_plus用于后续分析
                article_data['all_keywords'] = article_data['keywords'] | article_data['keywords_plus']
                self.articles.append(article_data)
        
        return file_article_count
                
    def extract_keywords_from_field(self, field_content):
        """从ID或DE字段内容中提取关键词"""
        if not field_content:
            return []
            
        field_content = field_content.replace('\n', ' ')
        keywords = re.split(r'[;\n]', field_content)
        
        cleaned_keywords = []
        for keyword in keywords:
            keyword = keyword.strip().lower()
            keyword = re.sub(r'\s+', ' ', keyword)
            keyword = keyword.strip('.,;()[]{}"\'-')
            
            if keyword and len(keyword) > 2:
                cleaned_keywords.append(keyword)
                
        return cleaned_keywords
    
    def is_complete_word_match(self, article_keyword, cices_keyword):
        """
        检查CICES关键词是否作为完整词或短语出现在文章关键词中
        确保避免部分匹配（如"crop"匹配"microscopy"）
        """
        article_kw = article_keyword.lower().strip()
        cices_kw = cices_keyword.lower().strip()
        
        # 如果是完全相同，直接匹配
        if article_kw == cices_kw:
            return True
        
        # 创建词边界匹配模式
        # 对于单词，使用\b词边界；对于短语，确保前后是空格或字符串边界
        if ' ' in cices_kw:
            # 多词短语：确保短语作为完整单位出现
            pattern = r'(?:^|\s)' + re.escape(cices_kw) + r'(?:\s|$)'
        else:
            # 单词：使用词边界确保完整单词匹配
            pattern = r'\b' + re.escape(cices_kw) + r'\b'
        
        # 双向匹配：检查cices_keyword是否在article_keyword中，或反之
        match_in_article = bool(re.search(pattern, article_kw))
        
        # 反向匹配：检查article_keyword是否作为完整词出现在cices_keyword中
        if ' ' in article_kw:
            reverse_pattern = r'(?:^|\s)' + re.escape(article_kw) + r'(?:\s|$)'
        else:
            reverse_pattern = r'\b' + re.escape(article_kw) + r'\b'
        match_in_cices = bool(re.search(reverse_pattern, cices_kw))
        
        return match_in_article or match_in_cices
        
    def classify_articles_by_cices(self):
        """根据CICES框架对文献进行分类 - 修正：直接对所有有关键词的文献进行分类"""
        classification_results = {
            'Provisioning': [],
            'Regulating': [],
            'Cultural': []
        }
        
        # 修正：直接对self.articles（所有有关键词的文献）进行分类，而不是es_articles
        for article in self.articles:
            article_categories = set()
            
            for main_category in ['Provisioning', 'Regulating', 'Cultural']:
                for subcategory, keywords in self.cices_categories[main_category].items():
                    for article_keyword in article['all_keywords']:
                        for cices_keyword in keywords:
                            # 使用改进的完整词匹配逻辑
                            if self.is_complete_word_match(article_keyword, cices_keyword):
                                article_categories.add(main_category)
                                classification_results[main_category].append(article)
                                break
                        if main_category in article_categories:
                            break
                    if main_category in article_categories:
                        break
            
            article['cices_categories'] = list(article_categories)
        
        # 去重
        for category in classification_results:
            seen = set()
            unique_articles = []
            for article in classification_results[category]:
                article_id = id(article)
                if article_id not in seen:
                    seen.add(article_id)
                    unique_articles.append(article)
            classification_results[category] = unique_articles
        
        return classification_results
    
    def analyze_yearly_distribution(self):
        """分析文献的年度分布 - 修正：分析所有有关键词的文献"""
        yearly_stats = defaultdict(lambda: {
            'total': 0,
            'cices_main_counts': defaultdict(int),
            'articles': []
        })
        
        # 修正：分析self.articles而不是es_articles
        for article in self.articles:
            year = article['year']
            if not year:
                year = 'Unknown'
            
            yearly_stats[year]['total'] += 1
            yearly_stats[year]['articles'].append(article)
            
            # 统计CICES分类
            for category in article.get('cices_categories', []):
                yearly_stats[year]['cices_main_counts'][category] += 1
        
        return yearly_stats
    
    def analyze_keyword_patterns(self):
        """分析文献中的关键词模式 - 修正：分析所有有关键词的文献"""
        # 统计所有关键词频次
        all_keywords = Counter()
        de_keywords = Counter()  # DE字段关键词
        id_keywords = Counter()  # ID字段关键词
        
        # 修正：分析self.articles而不是es_articles
        for article in self.articles:
            # 统计DE字段关键词
            for keyword in article['keywords']:
                all_keywords[keyword] += 1
                de_keywords[keyword] += 1
                
            # 统计ID字段关键词
            for keyword in article['keywords_plus']:
                all_keywords[keyword] += 1
                id_keywords[keyword] += 1
        
        return all_keywords, de_keywords, id_keywords
    
    def analyze_journal_distribution(self):
        """分析文献的期刊分布 - 修正：分析所有有关键词的文献"""
        journals = set()
        for article in self.articles:  # 修正：分析self.articles而不是es_articles
            if article['journal']:
                journals.add(article['journal'].lower().strip())
        
        journal_counts = Counter()
        for article in self.articles:  # 修正：分析self.articles而不是es_articles
            if article['journal']:
                normalized_journal = article['journal'].lower().strip()
                journal_counts[normalized_journal] += 1
        
        print(f"期刊总数: {len(journals)}")
        return journal_counts
    
    def create_line_chart(self, yearly_stats, main_categories_count):
        """创建生态系统服务研究趋势折线图 - 修正：只显示到2024年，使用指定颜色和字体"""
        # 准备数据 - 只包含2015-2024年
        valid_years = [str(year) for year in range(2015, 2025)]  # 2015-2024
        sorted_years = [year for year in valid_years if year in yearly_stats and yearly_stats[year]['total'] > 0]
        
        if not sorted_years:
            print("没有有效的年份数据用于可视化")
            return
        
        # 构建数据框架
        data = {
            'Year': [int(year) for year in sorted_years],
            'Total': [yearly_stats[year]['total'] for year in sorted_years],
            'Provisioning': [yearly_stats[year]['cices_main_counts'].get('Provisioning', 0) for year in sorted_years],
            'Regulating': [yearly_stats[year]['cices_main_counts'].get('Regulating', 0) for year in sorted_years],
            'Cultural': [yearly_stats[year]['cices_main_counts'].get('Cultural', 0) for year in sorted_years]
        }
        
        # 创建图形
        fig, ax = plt.subplots(figsize=(14, 10), dpi=150)
        
        # 使用指定的颜色
        categories = ['Provisioning', 'Regulating', 'Cultural', 'Total']
        colors = ['#A06593', '#6086C5', '#408868', '#2C3E50']  # 用户指定的颜色 + 总数用深色
        
        x = data['Year']
        
        for i, category in enumerate(categories):
            y = data[category]
            ax.plot(x, y, marker="o", label=category, linewidth=4, markersize=10, 
                   color=colors[i], alpha=0.9)
        
        # 添加高亮期间（快速增长期）
        if len(x) > 0:
            y_max = max(max(data[cat]) for cat in categories)
            
            # 找到增长最快的时期
            if len(x) >= 5:
                start_highlight = x[len(x)//3]
                end_highlight = x[2*len(x)//3]
                
                ax.fill_betweenx([0, y_max * 1.1], start_highlight, end_highlight, 
                               color="#c2768b", alpha=0.2)
                ax.plot([start_highlight, start_highlight], [0, y_max * 1.1], 
                       color="#c2768b", linestyle="--", alpha=0.8)
                ax.text(start_highlight - 0.1, y_max * 0.95, str(start_highlight), 
                       ha="right", va="bottom", color="#c2768b", fontweight='bold', fontsize=14)
                ax.plot([end_highlight, end_highlight], [0, y_max * 1.1], 
                       color="#c2768b", linestyle="--", alpha=0.8)
                ax.text(end_highlight + 0.1, y_max * 0.95, str(end_highlight), 
                       ha="left", va="bottom", color="#c2768b", fontweight='bold', fontsize=14)
                ax.text((start_highlight + end_highlight) / 2, y_max * 0.88, 
                       "Rapid Growth Period", ha="center", va="bottom", 
                       color="#c2768b", fontweight='bold', style='italic', fontsize=14)
        
        # 添加峰值标注
        total_values = data['Total']
        if total_values:
            peak_idx = total_values.index(max(total_values))
            peak_year = data['Year'][peak_idx]
            peak_value = total_values[peak_idx]
            ax.text(peak_year, peak_value * 1.05, f"Peak\n{peak_value}", 
                   ha="center", va="bottom", color="#2C3E50", fontweight='bold', 
                   fontsize=14, bbox=dict(boxstyle="round,pad=0.3", facecolor="yellow", alpha=0.7))
        
        # 设置标题和标签 - 增大字体
        ax.set_title("Ecosystem Services Research Trends (2015-2024)", 
                    fontsize=20, fontweight='bold', loc='left', x=0.02, y=0.95)
        ax.set_xlabel("Year", fontsize=16, fontweight='bold')
        ax.set_ylabel("Number of Publications", fontsize=16, fontweight='bold')
        
        # 设置范围
        if x:
            ax.set_xlim(min(x) - 0.5, max(x) + 0.5)
            if total_values:
                ax.set_ylim(0, max(total_values) * 1.15)
        
        # 创建图例 - 增大字体
        ax.legend(loc="upper left", bbox_to_anchor=(0.02, 0.85), frameon=True, 
                 fancybox=True, shadow=True, fontsize=14)
        
        # 设置刻度 - 增大字体
        ax.tick_params(axis='both', which='major', labelsize=14)
        ax.xaxis.set_major_locator(ticker.MultipleLocator(1))
        ax.xaxis.set_minor_locator(ticker.MultipleLocator(1))
        
        # 格式化y轴
        ax.yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, _: f"{int(x)}"))
        
        # 设置网格
        ax.grid(which="major", axis="y", linestyle="--", linewidth=0.5, color="#d3dce8", alpha=0.7)
        ax.set_facecolor('#fafafa')
        
        plt.tight_layout()
        plt.savefig('ecosystem_services_trend_chart.png', dpi=300, bbox_inches='tight',
                    facecolor='white', edgecolor='none')
        plt.show()
    
    def generate_comprehensive_report(self, yearly_stats, all_keywords, journal_counts,
                                    main_categories_count, de_keywords, id_keywords):
        """生成综合分析报告 - 修正：报告所有有关键词的文献"""
        print("\n" + "="*80)
        print("🌍 生态系统服务文献综合回顾报告")
        print("="*80)
        
        # 1. 总体统计
        print(f"\n📊 一、总体统计")
        print(f"  ├─ 所有文献总数: {len(self.all_articles)} 篇")
        print(f"  ├─ 有关键词文献数: {len(self.articles)} 篇") 
        print(f"  └─ 有关键词文献占比: {len(self.articles)/len(self.all_articles)*100:.2f}%")
        
        # 1.1 关键词字段统计
        print(f"\n📋 关键词字段统计:")
        de_count = len(de_keywords)
        id_count = len(id_keywords)
        print(f"  ├─ DE字段关键词数: {de_count}")
        print(f"  ├─ ID字段关键词数: {id_count}")
        print(f"  └─ 总关键词数(去重): {len(all_keywords)}")
        
        # 2. CICES分类统计
        print(f"\n🌳 二、CICES分类统计")
        print(f"{'主类别':<15} {'文献数':<10} {'占比(%)':<10}")
        print("-" * 40)
        for category, count in main_categories_count.items():
            percentage = (count / len(self.articles)) * 100  # 修正：基于self.articles计算
            print(f"{category:<15} {count:<10} {percentage:<10.1f}")
        
        # 3. 年度分布统计
        # 只显示2015-2024年的数据
        valid_years = [str(year) for year in range(2015, 2025)]
        sorted_years = [year for year in valid_years if year in yearly_stats]
        if 'Unknown' in yearly_stats:
            sorted_years.append('Unknown')
        
        print(f"\n📈 三、年度分布统计")
        print(f"{'年份':<8} {'文献数':<8} {'Prov':<8} {'Reg':<8} {'Cult':<8} {'占比(%)':<10}")
        print("-" * 55)
        
        for year in sorted_years:
            stats = yearly_stats[year]
            count = stats['total']
            prov_count = stats.get('cices_main_counts', {}).get('Provisioning', 0)
            reg_count = stats.get('cices_main_counts', {}).get('Regulating', 0)
            cult_count = stats.get('cices_main_counts', {}).get('Cultural', 0)
            percentage = (count / len(self.articles)) * 100  # 修正：基于self.articles计算
            print(f"{year:<8} {count:<8} {prov_count:<8} {reg_count:<8} {cult_count:<8} {percentage:<10.1f}")
        
        # 4. 高频关键词统计
        print(f"\n🏷️ 四、高频关键词分析（前70）")
        print(f"{'排名':<6} {'关键词':<40} {'频次':<8} {'来源':<8}")
        print("-" * 70)
        for i, (keyword, count) in enumerate(all_keywords.most_common(70), 1):
            # 判断关键词来源
            source = ""
            if keyword in de_keywords:
                source += "DE"
            if keyword in id_keywords:
                source += ("+" if source else "") + "ID"
            print(f"{i:<6} {keyword[:38]:<40} {count:<8} {source:<8}")
        
        # 5. 期刊分布
        print(f"\n📚 五、主要发表期刊（前10）")
        print(f"{'排名':<6} {'期刊名称':<50} {'文献数':<8}")
        print("-" * 70)
        for i, (journal, count) in enumerate(journal_counts.most_common(10), 1):
            print(f"{i:<6} {journal[:48]:<50} {count:<8}")
        
        return sorted_years
    
    def export_comprehensive_results(self, yearly_stats, all_keywords, 
                                    journal_counts, sorted_years,
                                    main_categories_count, de_keywords, id_keywords):
        """导出综合分析结果到Excel - 修正：基于所有有关键词的文献"""
        with pd.ExcelWriter('./ecosystem_services_comprehensive_review.xlsx', engine='openpyxl') as writer:
            
            # 1. 总体统计
            summary_data = {
                'Metric': ['所有文献总数', '有关键词文献数', '有关键词文献占比(%)', 
                          '涉及年份数', '平均年文献数',
                          'Provisioning文献数', 'Regulating文献数', 'Cultural文献数',
                          'DE字段关键词数', 'ID字段关键词数', '总关键词数(去重)'],
                'Value': [
                    len(self.all_articles),
                    len(self.articles),
                    round(len(self.articles)/len(self.all_articles)*100, 2),
                    len([y for y in sorted_years if y != 'Unknown']),
                    round(len(self.articles)/len([y for y in sorted_years if y != 'Unknown']), 1) if len([y for y in sorted_years if y != 'Unknown']) > 0 else 0,
                    main_categories_count.get('Provisioning', 0),
                    main_categories_count.get('Regulating', 0),
                    main_categories_count.get('Cultural', 0),
                    len(de_keywords),
                    len(id_keywords),
                    len(all_keywords)
                ]
            }
            summary_df = pd.DataFrame(summary_data)
            summary_df.to_excel(writer, sheet_name='总体统计', index=False)
            
            # 2. 年度分布
            yearly_data = []
            for year in sorted_years:
                yearly_data.append({
                    'Year': year,
                    'Articles': yearly_stats[year]['total'],
                    'Percentage': round((yearly_stats[year]['total']/len(self.articles))*100, 2),
                    'Provisioning': yearly_stats[year]['cices_main_counts'].get('Provisioning', 0),
                    'Regulating': yearly_stats[year]['cices_main_counts'].get('Regulating', 0),
                    'Cultural': yearly_stats[year]['cices_main_counts'].get('Cultural', 0)
                })
            yearly_df = pd.DataFrame(yearly_data)
            yearly_df.to_excel(writer, sheet_name='年度分布', index=False)
            
            # 3. 高频关键词
            keyword_data = []
            for keyword, count in all_keywords.most_common(200):
                source = ""
                if keyword in de_keywords:
                    source += "DE"
                if keyword in id_keywords:
                    source += ("+" if source else "") + "ID"
                
                keyword_data.append({
                    'Keyword': keyword,
                    'Frequency': count,
                    'Percentage': round((count/len(self.articles))*100, 2),
                    'Source': source,
                    'DE_Frequency': de_keywords.get(keyword, 0),
                    'ID_Frequency': id_keywords.get(keyword, 0)
                })
            keyword_df = pd.DataFrame(keyword_data)
            keyword_df.to_excel(writer, sheet_name='高频关键词', index=False)
            
            # 4. 期刊分布
            journal_data = []
            for journal, count in journal_counts.most_common():
                journal_data.append({
                    'Journal': journal,
                    'Articles': count,
                    'Percentage': round((count/len(self.articles))*100, 2)
                })
            journal_df = pd.DataFrame(journal_data)
            journal_df.to_excel(writer, sheet_name='期刊分布', index=False)
        
        print("\n✅ 综合分析结果已导出到 ecosystem_services_comprehensive_review.xlsx")
    
    def run_comprehensive_analysis(self):
        """运行完整的综合分析流程 - 修正：移除不必要的筛选步骤"""
        print("🚀 开始生态系统服务文献综合回顾分析...")
        print("="*80)
        
        # 步骤1: 读取文献
        self.read_literature_files()
        
        if not self.all_articles:
            print("❌ 未找到有效的文章数据！")
            return
        
        if not self.articles:
            print("❌ 未找到有关键词的文献！")
            return
        
        print(f"\n📊 WOS检索已确保所有文献都是生态系统服务相关，直接分析 {len(self.articles)} 篇有关键词的文献")
        
        # 步骤2: CICES分类 - 直接对所有有关键词的文献进行分类
        print("\n📊 正在进行CICES分类...")
        classification_results = self.classify_articles_by_cices()
        
        # 获取CICES主类别统计
        main_categories_count = {}
        for category in ['Provisioning', 'Regulating', 'Cultural']:
            main_categories_count[category] = len(classification_results[category])
        
        print(f"分类结果: Provisioning={main_categories_count.get('Provisioning', 0)}, "
              f"Regulating={main_categories_count.get('Regulating', 0)}, "
              f"Cultural={main_categories_count.get('Cultural', 0)}")
        
        # 步骤3: 年度分布分析
        print("📊 正在分析年度分布...")
        yearly_stats = self.analyze_yearly_distribution()
        
        # 步骤4: 关键词模式分析
        print("🔍 正在分析关键词模式...")
        all_keywords, de_keywords, id_keywords = self.analyze_keyword_patterns()
        
        # 步骤5: 期刊分布分析
        print("📚 正在分析期刊分布...")
        journal_counts = self.analyze_journal_distribution()
        
        # 步骤6: 生成综合报告
        sorted_years = self.generate_comprehensive_report(
            yearly_stats, all_keywords, journal_counts,
            main_categories_count, de_keywords, id_keywords
        )
        
        # 步骤7: 创建可视化
        print("\n📊 正在创建可视化图表...")
        self.create_line_chart(yearly_stats, main_categories_count)
        
        # 步骤8: 导出结果
        print("💾 正在导出分析结果...")
        self.export_comprehensive_results(
            yearly_stats, all_keywords, journal_counts, sorted_years,
            main_categories_count, de_keywords, id_keywords
        )
        
        print("\n" + "="*80)
        print("✅ 综合分析完成！生成的文件:")
        print("  📊 ecosystem_services_trend_chart.png")
        print("  📁 ecosystem_services_comprehensive_review.xlsx")
        print("="*80)

# 运行分析
if __name__ == "__main__":
    analyzer = EcosystemServicesComprehensiveAnalyzer()
    analyzer.run_comprehensive_analysis()
```

```python
import os
import re
import pandas as pd
import numpy as np
from collections import Counter, defaultdict
import matplotlib.pyplot as plt
import seaborn as sns
import hashlib

# 设置Times New Roman字体
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['axes.unicode_minus'] = False

class EcosystemServicesBibliometricAnalyzer:
    def __init__(self, data_directory="./"):
        self.data_directory = data_directory
        self.all_articles = []
        self.articles = []  # 所有有关键词的文献 - 这是我们要分析的数据集
        
        # CICES V5.2分类
        self.cices_categories = {
            'Provisioning': {
                'terrestrial_crops': [
                    'crops', 'cereals', 'root crops', 'soft fruit', 'fodder', 'livestock feed',
                    'fibres', 'materials', 'cultivated plants', 'fungi', 'algae', 'bacteria',
                    'plantation', 'energy source', 'bioenergy'
                ],
                'aquatic_crops': [
                    'aquaculture', 'in-situ aquaculture', 'cultivated aquatic plants'
                ],
                'reared_animals': [
                    'reared animals', 'beef', 'dairy', 'animal products',
                    'fibres from animals', 'mechanical energy', 'biogas'
                ],
                'aquatic_animals': [
                    'reared aquatic animals', 'in-situ aquaculture animals'
                ],
                'wild_plants': [
                    'wild plants', 'terrestrial plants', 'aquatic plants', 'wild fungi', 'wild algae',
                    'wild plant materials', 'wild plant energy'
                ],
                'wild_animals': [
                    'wild animals', 'terrestrial animals', 'aquatic animals',
                    'wild animal nutrition', 'wild animal materials'
                ],
                'genetic_material': [
                    'genetic material', 'seeds', 'spores', 'plant materials',
                    'breeding', 'new strains', 'varieties',
                    'individual genes', 'biological entities',
                    'animal genetic material', 'population establishment'
                ]
            },
            'Regulating': {
                'biochemical_transformation': [
                    'bio-remediation', 'micro-organisms', 'pollution degradation',
                    'filtration', 'sequestration', 'storage', 'accumulation',
                    'waste mediation', 'toxic substances', 'anthropogenic origin',
                    'smell reduction', 'noise attenuation', 'visual screening'
                ],
                'flow_extreme_events': [
                    'erosion control', 'water erosion', 'wind erosion',
                    'hydrological cycle', 'water flow regulation', 'runoff', 'base flows', 'peak flows',
                    'hazard mitigation', 'mass movement', 'landslides', 'flood mitigation', 'surge protection',
                    'wind protection', 'fire protection'
                ],
                'lifecycle_maintenance': [
                    'pollination', 'gamete dispersal', 'seed dispersal',
                    'nursery populations', 'breeding grounds', 'gene pool protection',
                    'refuge habitats', 'feeding grounds'
                ],
                'pest_disease_control': [
                    'pest control', 'invasive species', 'disease control'
                ],
                'soil_quality': [
                    'soil quality', 'weathering processes', 'decomposition', 'fixing processes',
                    'soil structure', 'biological agents', 'soil porosity'
                ],
                'water_conditions': [
                    'water conditions', 'chemical condition', 'macronutrients',
                    'freshwater regulation', 'saltwater regulation'
                ],
                'atmospheric_conditions': [
                    'atmospheric composition', 'oceanic circulation', 'climate change',
                    'temperature regulation', 'humidity', 'ventilation', 'transpiration', 'microclimate'
                ]
            },
            'Cultural': {
                'physical_experiential': [
                    'recreational activities', 'health promotion', 'recuperation', 'enjoyment',
                    'active interactions', 'immersive interactions',
                    'passive interactions', 'observational interactions'
                ],
                'intellectual_representative': [
                    'scientific investigation', 'traditional ecological knowledge', 'genetic diversity',
                    'education', 'training',
                    'culture', 'heritage', 'cultural resonance',
                    'aesthetic experiences'
                ],
                'indirect_interactions': [
                    'entertainment', 'representation outside setting'
                ],
                'spiritual_symbolic': [
                    'symbolic meaning', 'sense of place', 'distinctiveness',
                    'spiritual meaning', 'religious meaning',
                    'contemporary existence', 'conservation importance',
                    'inter-generational existence', 'genetic diversity conservation'
                ]
            }
        }
        
    def read_literature_files(self):
        """读取所有文献文件"""
        files = []
        for f in os.listdir(self.data_directory):
            if f.endswith('.txt') and 'download' in f and any(year in f for year in 
                ['2015', '2016', '2017', '2018', '2019', '2020', '2021', '2022', '2023', '2024', '2025']):
                files.append(f)
        
        files.sort()
        print(f"📚 找到 {len(files)} 个文献文件")
        
        for file in files:
            file_path = os.path.join(self.data_directory, file)
            try:
                with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
                    content = f.read()
                    self.parse_articles(content, file)
            except Exception as e:
                print(f"❌ 读取文件 {file} 时出错: {e}")
        
        print(f"✅ 总共解析了 {len(self.all_articles)} 篇文章")
        print(f"✅ 有关键词的文章: {len(self.articles)} 篇")
        
    def parse_articles(self, content, filename):
        """解析文章内容，提取完整信息"""
        articles = re.split(r'\nER\s*\n', content)
        
        for article_text in articles:
            if not article_text.strip():
                continue
            
            # 检查是否包含PT字段
            if not re.search(r'(^|\n)PT\s+', article_text):
                continue
            
            article_data = {
                'filename': filename,
                'keywords': set(),  # DE字段关键词
                'keywords_plus': set(),  # ID字段关键词
                'title': '',
                'year': '',
                'abstract': '',
                'journal': '',
                'authors': [],
                'author_count': 0,  # 作者数量
                'citations': 0,
                'reference_count': 0
            }
            
            # 提取ID字段 (Keywords Plus)
            id_match = re.search(r'\nID\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if id_match:
                id_content = id_match.group(1)
                keywords_plus = self.extract_keywords_from_field(id_content)
                article_data['keywords_plus'].update(keywords_plus)
            
            # 提取DE字段 (Author's keywords)
            de_match = re.search(r'\nDE\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if de_match:
                de_content = de_match.group(1)
                keywords = self.extract_keywords_from_field(de_content)
                article_data['keywords'].update(keywords)
            
            # 提取标题
            ti_match = re.search(r'\nTI\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if ti_match:
                article_data['title'] = ti_match.group(1).replace('\n', ' ').strip()
            
            # 提取年份
            py_match = re.search(r'\nPY\s+(\d{4})', article_text)
            if py_match:
                article_data['year'] = py_match.group(1)
            
            # 提取摘要
            ab_match = re.search(r'\nAB\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if ab_match:
                article_data['abstract'] = ab_match.group(1).replace('\n', ' ').strip()
            
            # 提取期刊名称
            so_match = re.search(r'\nSO\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if so_match:
                article_data['journal'] = so_match.group(1).replace('\n', ' ').strip()
            
            # 提取作者（统计数量）
            au_match = re.search(r'\nAU\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if au_match:
                authors_text = au_match.group(1)
                authors = [author.strip() for author in authors_text.split('\n') if author.strip()]
                article_data['authors'] = authors
                article_data['author_count'] = len(authors)
            
            # 提取引用次数 (TC字段)
            tc_match = re.search(r'\nTC\s+(\d+)', article_text)
            if tc_match:
                article_data['citations'] = int(tc_match.group(1))
            
            # 提取参考文献数量 (NR字段)
            nr_match = re.search(r'\nNR\s+(\d+)', article_text)
            if nr_match:
                article_data['reference_count'] = int(nr_match.group(1))
            
            # 添加到所有文献列表
            self.all_articles.append(article_data)
            
            # 只有当文章有关键词时才添加到articles列表
            if article_data['keywords'] or article_data['keywords_plus']:
                # 合并keywords和keywords_plus用于分类
                article_data['all_keywords'] = article_data['keywords'] | article_data['keywords_plus']
                self.articles.append(article_data)
                
    def extract_keywords_from_field(self, field_content):
        """从ID或DE字段内容中提取关键词"""
        if not field_content:
            return []
            
        field_content = field_content.replace('\n', ' ')
        keywords = re.split(r'[;\n]', field_content)
        
        cleaned_keywords = []
        for keyword in keywords:
            keyword = keyword.strip().lower()
            keyword = re.sub(r'\s+', ' ', keyword)
            keyword = keyword.strip('.,;()[]{}"\'-')
            
            if keyword and len(keyword) > 2:
                cleaned_keywords.append(keyword)
                
        return cleaned_keywords
    
    def generate_article_hash(self, article):
        """生成文章唯一标识符，用于去重"""
        # 使用标题、年份、第一作者生成hash
        identifier_string = f"{article.get('title', '')}{article.get('year', '')}"
        if article.get('authors'):
            identifier_string += article['authors'][0]
        return hashlib.md5(identifier_string.encode()).hexdigest()
    
    def is_complete_word_match(self, article_keyword, cices_keyword):
        """
        检查CICES关键词是否作为完整词或短语出现在文章关键词中
        确保避免部分匹配（如"crop"匹配"microscopy"）
        """
        article_kw = article_keyword.lower().strip()
        cices_kw = cices_keyword.lower().strip()
        
        # 如果是完全相同，直接匹配
        if article_kw == cices_kw:
            return True
        
        # 创建词边界匹配模式
        # 对于单词，使用\b词边界；对于短语，确保前后是空格或字符串边界
        if ' ' in cices_kw:
            # 多词短语：确保短语作为完整单位出现
            pattern = r'(?:^|\s)' + re.escape(cices_kw) + r'(?:\s|$)'
        else:
            # 单词：使用词边界确保完整单词匹配
            pattern = r'\b' + re.escape(cices_kw) + r'\b'
        
        # 双向匹配：检查cices_keyword是否在article_keyword中，或反之
        match_in_article = bool(re.search(pattern, article_kw))
        
        # 反向匹配：检查article_keyword是否作为完整词出现在cices_keyword中
        if ' ' in article_kw:
            reverse_pattern = r'(?:^|\s)' + re.escape(article_kw) + r'(?:\s|$)'
        else:
            reverse_pattern = r'\b' + re.escape(article_kw) + r'\b'
        match_in_cices = bool(re.search(reverse_pattern, cices_kw))
        
        return match_in_article or match_in_cices
        
    def classify_articles_by_cices(self):
        """根据CICES框架对文献进行分类，使用完整词匹配并确保去重"""
        classification_results = {
            'Provisioning': [],
            'Regulating': [],
            'Cultural': []
        }
        
        # 使用字典追踪已添加的文章（基于hash去重）
        added_articles = {
            'Provisioning': set(),
            'Regulating': set(), 
            'Cultural': set()
        }
        
        # 分类文章
        for article in self.articles:
            article_categories = set()
            article_hash = self.generate_article_hash(article)
            
            for main_category in ['Provisioning', 'Regulating', 'Cultural']:
                for subcategory, keywords in self.cices_categories[main_category].items():
                    for article_keyword in article['all_keywords']:
                        for cices_keyword in keywords:
                            # 使用改进的完整词匹配逻辑
                            if self.is_complete_word_match(article_keyword, cices_keyword):
                                article_categories.add(main_category)
                                
                                # 只有当文章hash未被添加过才添加到结果中
                                if article_hash not in added_articles[main_category]:
                                    classification_results[main_category].append(article)
                                    added_articles[main_category].add(article_hash)
                                break
                        if main_category in article_categories:
                            break
                    if main_category in article_categories:
                        break
            
            article['cices_categories'] = list(article_categories)
        
        return classification_results
    
    def calculate_total_cices_articles(self, classification_results):
        """计算包含CICES分类的文献总数（去重）"""
        all_cices_hashes = set()
        
        for category, articles in classification_results.items():
            for article in articles:
                article_hash = self.generate_article_hash(article)
                all_cices_hashes.add(article_hash)
        
        return len(all_cices_hashes)
    
    def calculate_bibliometric_indicators(self, articles, category_name="总体文献"):
        """计算文献计量学指标"""
        if not articles:
            return None
        
        indicators = {}
        
        # 1. Documents (文献总数)
        indicators['Documents'] = len(articles)
        
        # 2. Sources (来源期刊数)
        journals = set()
        for article in articles:
            if article['journal']:
                journals.add(article['journal'].lower().strip())
        indicators['Sources'] = len(journals)
        
        # 3. Author's keywords (作者关键词总数)
        author_keywords = set()
        for article in articles:
            author_keywords.update(article['keywords'])
        indicators['Authors_Keywords'] = len(author_keywords)
        
        # 4. Keywords Plus (ID关键词增益)
        keywords_plus = set()
        for article in articles:
            keywords_plus.update(article['keywords_plus'])
        indicators['Keywords_Plus_ID'] = len(keywords_plus)
        
        # 5. Period (时间段)
        years = []
        for article in articles:
            if article['year'] and article['year'].isdigit():
                years.append(int(article['year']))
        if years:
            indicators['Period'] = f"{min(years)}-{max(years)}"
            indicators['Years_Span'] = max(years) - min(years) + 1
        else:
            indicators['Period'] = "Unknown"
            indicators['Years_Span'] = 0
        
        # 6. Single-authored documents (单作者文献数)
        single_authored = 0
        multi_authored = 0
        for article in articles:
            if article['author_count'] == 1:
                single_authored += 1
            elif article['author_count'] > 1:
                multi_authored += 1
        indicators['Single_Authored_Docs'] = single_authored
        indicators['Multi_Authored_Docs'] = multi_authored
        
        # 7. Average citations per article (每篇文献平均引用次数)
        total_citations = sum(article['citations'] for article in articles)
        indicators['Total_Citations'] = total_citations
        if indicators['Documents'] > 0:
            indicators['Avg_Citations_per_Article'] = round(
                total_citations / indicators['Documents'], 2
            )
        else:
            indicators['Avg_Citations_per_Article'] = 0
        
        # 8. Average references per article (每篇文献平均参考文献数)
        total_references = sum(article['reference_count'] for article in articles)
        indicators['Total_References'] = total_references
        if indicators['Documents'] > 0:
            indicators['Avg_References_per_Article'] = round(
                total_references / indicators['Documents'], 2
            )
        else:
            indicators['Avg_References_per_Article'] = 0
        
        # 9. Citation statistics (引用统计)
        citations_list = [article['citations'] for article in articles]
        if citations_list:
            indicators['Max_Citations'] = max(citations_list)
            indicators['Min_Citations'] = min(citations_list)
            indicators['Median_Citations'] = np.median(citations_list)
        else:
            indicators['Max_Citations'] = 0
            indicators['Min_Citations'] = 0
            indicators['Median_Citations'] = 0
        
        # 10. Highly cited articles (高被引文献，引用次数>平均值+1个标准差)
        if citations_list and len(citations_list) > 1:
            mean_citations = np.mean(citations_list)
            std_citations = np.std(citations_list)
            highly_cited_threshold = mean_citations + std_citations
            indicators['Highly_Cited_Threshold'] = round(highly_cited_threshold, 2)
            indicators['Highly_Cited_Articles'] = sum(1 for c in citations_list if c > highly_cited_threshold)
        else:
            indicators['Highly_Cited_Threshold'] = 0
            indicators['Highly_Cited_Articles'] = 0
        
        # 11. Annual publication statistics (年度出版统计)
        if years:
            indicators['Avg_Articles_per_Year'] = round(len(articles) / indicators['Years_Span'], 2)
            year_distribution = Counter(years)
            indicators['Most_Productive_Year'] = year_distribution.most_common(1)[0][0]
            indicators['Max_Articles_per_Year'] = year_distribution.most_common(1)[0][1]
        else:
            indicators['Avg_Articles_per_Year'] = 0
            indicators['Most_Productive_Year'] = 'Unknown'
            indicators['Max_Articles_per_Year'] = 0
        
        # 12. Journal statistics (期刊统计)
        journal_distribution = Counter()
        for article in articles:
            if article['journal']:
                journal_distribution[article['journal'].lower().strip()] += 1
        
        if journal_distribution:
            indicators['Most_Productive_Journal'] = journal_distribution.most_common(1)[0][0]
            indicators['Max_Articles_per_Journal'] = journal_distribution.most_common(1)[0][1]
            # Gini系数计算期刊集中度
            journal_counts = list(journal_distribution.values())
            journal_counts.sort()
            n = len(journal_counts)
            if n > 1:
                index = np.arange(1, n + 1)
                gini = (2 * np.sum(index * journal_counts)) / (n * np.sum(journal_counts)) - (n + 1) / n
                indicators['Journal_Concentration_Gini'] = round(gini, 3)
            else:
                indicators['Journal_Concentration_Gini'] = 0
        else:
            indicators['Most_Productive_Journal'] = 'Unknown'
            indicators['Max_Articles_per_Journal'] = 0
            indicators['Journal_Concentration_Gini'] = 0
        
        return indicators
    
    def generate_bibliometric_report(self, overall_indicators, cices_indicators, total_cices_articles):
        """生成文献计量学报告"""
        print("\n" + "="*80)
        print("📊 Ecosystem Services Bibliometric Analysis Report")
        print("="*80)
        
        # 定义指标中英文对照
        indicator_names = {
            'Documents': 'Documents',
            'Sources': 'Sources (Journals)',
            'Authors_Keywords': "Author's Keywords",
            'Keywords_Plus_ID': 'Keywords Plus (ID)',
            'Period': 'Period',
            'Years_Span': 'Years Span',
            'Single_Authored_Docs': 'Single-Authored Documents',
            'Multi_Authored_Docs': 'Multi-Authored Documents',
            'Total_Citations': 'Total Citations',
            'Avg_Citations_per_Article': 'Average Citations per Article',
            'Total_References': 'Total References',
            'Avg_References_per_Article': 'Average References per Article',
            'Max_Citations': 'Maximum Citations',
            'Min_Citations': 'Minimum Citations',
            'Median_Citations': 'Median Citations',
            'Highly_Cited_Threshold': 'Highly Cited Threshold',
            'Highly_Cited_Articles': 'Highly Cited Articles',
            'Avg_Articles_per_Year': 'Average Articles per Year',
            'Most_Productive_Year': 'Most Productive Year',
            'Max_Articles_per_Year': 'Max Articles per Year',
            'Most_Productive_Journal': 'Most Productive Journal',
            'Max_Articles_per_Journal': 'Max Articles per Journal',
            'Journal_Concentration_Gini': 'Journal Concentration (Gini)'
        }
        
        # 打印CICES文献总数统计
        print(f"\n📊 CICES Classification Summary")
        print("-" * 60)
        print(f"Total Articles with Keywords: {len(self.articles)}")
        print(f"Articles with CICES Classification: {total_cices_articles}")
        print(f"CICES Coverage Rate: {(total_cices_articles/len(self.articles)*100):.2f}%")
        
        for category in ['Provisioning', 'Regulating', 'Cultural']:
            if category in cices_indicators and cices_indicators[category]:
                count = cices_indicators[category]['Documents']
                rate = (count/len(self.articles)*100)
                print(f"{category} Services: {count} articles ({rate:.2f}%)")
        
        # 打印总体指标
        print("\n📈 I. Overall Bibliometric Indicators")
        print("-" * 60)
        print(f"{'Indicator':<35} {'Value':<25}")
        print("-" * 60)
        for key, value in overall_indicators.items():
            name = indicator_names.get(key, key)
            print(f"{name:<35} {str(value):<25}")
        
        # 打印CICES分类指标
        for category in ['Provisioning', 'Regulating', 'Cultural']:
            if category in cices_indicators and cices_indicators[category]:
                print(f"\n📊 {category} Services Bibliometric Indicators")
                print("-" * 60)
                print(f"{'Indicator':<35} {'Value':<25}")
                print("-" * 60)
                for key, value in cices_indicators[category].items():
                    name = indicator_names.get(key, key)
                    print(f"{name:<35} {str(value):<25}")
        
        # 打印对比表
        print("\n📊 CICES Categories Comparison")
        print("-" * 100)
        print(f"{'Indicator':<30} {'Overall':<15} {'Provisioning':<15} {'Regulating':<15} {'Cultural':<15}")
        print("-" * 100)
        
        for key in overall_indicators.keys():
            name = indicator_names.get(key, key)[:29]
            values = [str(overall_indicators[key])[:14]]
            for category in ['Provisioning', 'Regulating', 'Cultural']:
                if category in cices_indicators and cices_indicators[category]:
                    values.append(str(cices_indicators[category].get(key, 'N/A'))[:14])
                else:
                    values.append('N/A')
            print(f"{name:<30} {values[0]:<15} {values[1]:<15} {values[2]:<15} {values[3]:<15}")
    
    def create_bibliometric_visualizations(self, overall_indicators, cices_indicators):
        """创建文献计量学可视化"""
        fig = plt.figure(figsize=(20, 15))
        
        # 1. 文献数量对比
        plt.subplot(3, 3, 1)
        categories = ['Overall', 'Provisioning', 'Regulating', 'Cultural']
        doc_counts = [overall_indicators['Documents']]
        for cat in ['Provisioning', 'Regulating', 'Cultural']:
            if cat in cices_indicators and cices_indicators[cat]:
                doc_counts.append(cices_indicators[cat]['Documents'])
            else:
                doc_counts.append(0)
        
        colors = ['#333333', '#4CAF50', '#2196F3', '#FF9800']
        plt.bar(categories, doc_counts, color=colors, alpha=0.7)
        plt.xlabel('Category', fontsize=12)
        plt.ylabel('Number of Documents', fontsize=12)
        plt.title('Documents by Category', fontsize=14, fontweight='bold')
        
        # 添加数值标签
        for i, v in enumerate(doc_counts):
            plt.text(i, v + max(doc_counts)*0.01, str(v), ha='center', va='bottom')
        
        # 2. 被引次数分析
        plt.subplot(3, 3, 2)
        categories = ['Overall', 'Provisioning', 'Regulating', 'Cultural']
        avg_citations = [overall_indicators['Avg_Citations_per_Article']]
        
        for cat in ['Provisioning', 'Regulating', 'Cultural']:
            if cat in cices_indicators and cices_indicators[cat]:
                avg_citations.append(cices_indicators[cat]['Avg_Citations_per_Article'])
            else:
                avg_citations.append(0)
        
        plt.bar(categories, avg_citations, color=['#607D8B', '#4CAF50', '#2196F3', '#FF9800'], alpha=0.7)
        plt.xlabel('Category', fontsize=12)
        plt.ylabel('Average Citations per Article', fontsize=12)
        plt.title('Average Citations by Category', fontsize=14, fontweight='bold')
        
        # 添加数值标签
        for i, v in enumerate(avg_citations):
            plt.text(i, v + max(avg_citations)*0.01, f'{v:.2f}', ha='center', va='bottom')
        
        # 3. 期刊与关键词多样性
        plt.subplot(3, 3, 3)
        metrics = ['Sources', "Author's Keywords", 'Keywords Plus']
        overall_values = [
            overall_indicators['Sources'],
            overall_indicators['Authors_Keywords'],
            overall_indicators['Keywords_Plus_ID']
        ]
        
        x = np.arange(len(metrics))
        width = 0.2
        
        plt.bar(x - 1.5*width, overall_values, width, label='Overall', color='#333333', alpha=0.7)
        
        for i, cat in enumerate(['Provisioning', 'Regulating', 'Cultural']):
            if cat in cices_indicators and cices_indicators[cat]:
                values = [
                    cices_indicators[cat]['Sources'],
                    cices_indicators[cat]['Authors_Keywords'],
                    cices_indicators[cat]['Keywords_Plus_ID']
                ]
                plt.bar(x + (i-0.5)*width, values, width, label=cat, color=colors[i+1], alpha=0.7)
        
        plt.xlabel('Indicators', fontsize=12)
        plt.ylabel('Count', fontsize=12)
        plt.title('Source and Keyword Diversity', fontsize=14, fontweight='bold')
        plt.xticks(x, metrics)
        plt.legend()
        
        # 4. 单作者vs多作者文献
        plt.subplot(3, 3, 4)
        categories = ['Overall', 'Provisioning', 'Regulating', 'Cultural']
        single_authored = [overall_indicators['Single_Authored_Docs']]
        multi_authored = [overall_indicators['Multi_Authored_Docs']]
        
        for cat in ['Provisioning', 'Regulating', 'Cultural']:
            if cat in cices_indicators and cices_indicators[cat]:
                single_authored.append(cices_indicators[cat]['Single_Authored_Docs'])
                multi_authored.append(cices_indicators[cat]['Multi_Authored_Docs'])
            else:
                single_authored.append(0)
                multi_authored.append(0)
        
        x = np.arange(len(categories))
        width = 0.35
        
        plt.bar(x - width/2, single_authored, width, label='Single-authored', color='#E91E63', alpha=0.7)
        plt.bar(x + width/2, multi_authored, width, label='Multi-authored', color='#9C27B0', alpha=0.7)
        
        plt.xlabel('Category', fontsize=12)
        plt.ylabel('Number of Documents', fontsize=12)
        plt.title('Single vs Multi-authored Documents', fontsize=14, fontweight='bold')
        plt.xticks(x, categories)
        plt.legend()
        
        # 5. 时间分布（如果有年份数据）
        plt.subplot(3, 3, 5)
        # 计算总体文献的年份分布
        years_data = []
        for article in self.articles:
            if article['year'] and article['year'].isdigit():
                years_data.append(int(article['year']))
        
        if years_data:
            year_counts = Counter(years_data)
            years = sorted(year_counts.keys())
            counts = [year_counts[year] for year in years]
            
            plt.plot(years, counts, marker='o', linewidth=2, markersize=6, color='#333333')
            plt.xlabel('Year', fontsize=12)
            plt.ylabel('Number of Documents', fontsize=12)
            plt.title('Annual Publication Trend', fontsize=14, fontweight='bold')
            plt.grid(True, alpha=0.3)
        else:
            plt.text(0.5, 0.5, 'No year data', ha='center', va='center', transform=plt.gca().transAxes)
            plt.title('Annual Publication Trend', fontsize=14, fontweight='bold')
        
        # 6. 参考文献分析
        plt.subplot(3, 3, 6)
        categories = ['Overall', 'Provisioning', 'Regulating', 'Cultural']
        avg_references = [overall_indicators['Avg_References_per_Article']]
        
        for cat in ['Provisioning', 'Regulating', 'Cultural']:
            if cat in cices_indicators and cices_indicators[cat]:
                avg_references.append(cices_indicators[cat]['Avg_References_per_Article'])
            else:
                avg_references.append(0)
        
        plt.bar(categories, avg_references, color=['#795548', '#4CAF50', '#2196F3', '#FF9800'], alpha=0.7)
        plt.xlabel('Category', fontsize=12)
        plt.ylabel('Average References per Article', fontsize=12)
        plt.title('Average References by Category', fontsize=14, fontweight='bold')
        
        # 添加数值标签
        for i, v in enumerate(avg_references):
            plt.text(i, v + max(avg_references)*0.01, f'{v:.1f}', ha='center', va='bottom')
        
        # 7. 高被引文献分析
        plt.subplot(3, 3, 7)
        categories = ['Overall', 'Provisioning', 'Regulating', 'Cultural']
        highly_cited = [overall_indicators['Highly_Cited_Articles']]
        
        for cat in ['Provisioning', 'Regulating', 'Cultural']:
            if cat in cices_indicators and cices_indicators[cat]:
                highly_cited.append(cices_indicators[cat]['Highly_Cited_Articles'])
            else:
                highly_cited.append(0)
        
        plt.bar(categories, highly_cited, color=['#9C27B0', '#4CAF50', '#2196F3', '#FF9800'], alpha=0.7)
        plt.xlabel('Category', fontsize=12)
        plt.ylabel('Highly Cited Articles', fontsize=12)
        plt.title('Highly Cited Articles by Category', fontsize=14, fontweight='bold')
        
        # 添加数值标签
        for i, v in enumerate(highly_cited):
            plt.text(i, v + max(highly_cited)*0.01 if max(highly_cited) > 0 else 0.1, str(v), ha='center', va='bottom')
        
        # 8. 期刊集中度分析
        plt.subplot(3, 3, 8)
        categories = ['Overall', 'Provisioning', 'Regulating', 'Cultural']
        gini_coeffs = [overall_indicators['Journal_Concentration_Gini']]
        
        for cat in ['Provisioning', 'Regulating', 'Cultural']:
            if cat in cices_indicators and cices_indicators[cat]:
                gini_coeffs.append(cices_indicators[cat]['Journal_Concentration_Gini'])
            else:
                gini_coeffs.append(0)
        
        plt.bar(categories, gini_coeffs, color=['#FF5722', '#4CAF50', '#2196F3', '#FF9800'], alpha=0.7)
        plt.xlabel('Category', fontsize=12)
        plt.ylabel('Gini Coefficient', fontsize=12)
        plt.title('Journal Concentration (Gini Coefficient)', fontsize=14, fontweight='bold')
        plt.ylim(0, 1)
        
        # 添加数值标签
        for i, v in enumerate(gini_coeffs):
            plt.text(i, v + 0.02, f'{v:.3f}', ha='center', va='bottom')
        
        # 9. 综合指标热力图
        plt.subplot(3, 3, 9)
        
        # 准备热力图数据
        heatmap_data = []
        metrics_list = ['Documents', 'Sources', 'Avg Citations', 'Avg References', 'Highly Cited']
        
        for cat in ['Overall', 'Provisioning', 'Regulating', 'Cultural']:
            if cat == 'Overall':
                row = [
                    overall_indicators['Documents'],
                    overall_indicators['Sources'],
                    overall_indicators['Avg_Citations_per_Article'],
                    overall_indicators['Avg_References_per_Article'],
                    overall_indicators['Highly_Cited_Articles']
                ]
            elif cat in cices_indicators and cices_indicators[cat]:
                row = [
                    cices_indicators[cat]['Documents'],
                    cices_indicators[cat]['Sources'],
                    cices_indicators[cat]['Avg_Citations_per_Article'],
                    cices_indicators[cat]['Avg_References_per_Article'],
                    cices_indicators[cat]['Highly_Cited_Articles']
                ]
            else:
                row = [0, 0, 0, 0, 0]
            heatmap_data.append(row)
        
        # 归一化处理
        heatmap_array = np.array(heatmap_data, dtype=float)
        for i in range(heatmap_array.shape[1]):
            max_val = heatmap_array[:, i].max()
            if max_val > 0:
                heatmap_array[:, i] = heatmap_array[:, i] / max_val
        
        sns.heatmap(heatmap_array, 
                   xticklabels=metrics_list, 
                   yticklabels=['Overall', 'Provisioning', 'Regulating', 'Cultural'],
                   annot=True, fmt='.2f', cmap='YlOrRd', cbar_kws={'label': 'Normalized Value'})
        plt.title('Comprehensive Indicators Heatmap', fontsize=14, fontweight='bold')
        
        plt.tight_layout()
        plt.savefig('./bibliometric_analysis.png', dpi=300, bbox_inches='tight')
        plt.show()
    
    def export_bibliometric_results(self, overall_indicators, cices_indicators, classification_results, total_cices_articles):
        """导出文献计量学结果到Excel"""
        with pd.ExcelWriter('./bibliometric_analysis.xlsx', engine='openpyxl') as writer:
            
            # 1. CICES总体统计（新增）
            cices_summary_data = {
                'Metric': [
                    'Total Articles with Keywords',
                    'Articles with CICES Classification', 
                    'CICES Coverage Rate (%)',
                    'Provisioning Services Articles',
                    'Regulating Services Articles', 
                    'Cultural Services Articles'
                ],
                'Value': [
                    len(self.articles),
                    total_cices_articles,
                    round((total_cices_articles/len(self.articles)*100), 2),
                    cices_indicators.get('Provisioning', {}).get('Documents', 0) if 'Provisioning' in cices_indicators else 0,
                    cices_indicators.get('Regulating', {}).get('Documents', 0) if 'Regulating' in cices_indicators else 0,
                    cices_indicators.get('Cultural', {}).get('Documents', 0) if 'Cultural' in cices_indicators else 0
                ]
            }
            cices_summary_df = pd.DataFrame(cices_summary_data)
            cices_summary_df.to_excel(writer, sheet_name='CICES_Summary', index=False)
            
            # 2. 总体指标
            overall_df = pd.DataFrame([overall_indicators]).T
            overall_df.columns = ['Value']
            overall_df.index.name = 'Indicator'
            overall_df.to_excel(writer, sheet_name='Overall_Indicators')
            
            # 3. CICES分类指标对比
            comparison_data = {'Indicator': list(overall_indicators.keys())}
            comparison_data['Overall'] = list(overall_indicators.values())
            
            for category in ['Provisioning', 'Regulating', 'Cultural']:
                if category in cices_indicators and cices_indicators[category]:
                    comparison_data[category] = [cices_indicators[category].get(key, 'N/A') 
                                                for key in overall_indicators.keys()]
                else:
                    comparison_data[category] = ['N/A'] * len(overall_indicators)
            
            comparison_df = pd.DataFrame(comparison_data)
            comparison_df.to_excel(writer, sheet_name='CICES_Comparison', index=False)
            
            # 4. 各类别详细指标
            for category in ['Provisioning', 'Regulating', 'Cultural']:
                if category in cices_indicators and cices_indicators[category]:
                    cat_df = pd.DataFrame([cices_indicators[category]]).T
                    cat_df.columns = ['Value']
                    cat_df.index.name = 'Indicator'
                    cat_df.to_excel(writer, sheet_name=f'{category}_Indicators')
            
            # 5. 文献列表统计
            for category, articles in classification_results.items():
                if articles:
                    article_data = []
                    for article in articles[:200]:  # 限制每类前200篇
                        article_data.append({
                            'Title': article['title'],
                            'Year': article['year'],
                            'Journal': article['journal'],
                            'Author_Count': article['author_count'],
                            'Citations': article['citations'],
                            'References': article['reference_count'],
                            'DE_Keywords': len(article['keywords']),
                            'ID_Keywords_Plus': len(article['keywords_plus'])
                        })
                    
                    if article_data:
                        articles_df = pd.DataFrame(article_data)
                        articles_df.to_excel(writer, sheet_name=f'{category}_Articles', index=False)
        
        print("\n✅ Bibliometric analysis results exported to bibliometric_analysis.xlsx")
    
    def run_bibliometric_analysis(self):
        """运行完整的文献计量学分析"""
        print("🚀 Starting Ecosystem Services Bibliometric Analysis...")
        print("="*80)
        
        # 步骤1: 读取文献
        self.read_literature_files()
        
        if not self.articles:
            print("❌ No valid articles found!")
            return
        
        print(f"\n📊 WOS search already ensures all articles are ecosystem service related")
        print(f"📊 Analyzing {len(self.articles)} articles with keywords")
        
        # 步骤2: CICES分类 - 使用改进的匹配逻辑并确保去重
        print("\n📊 Performing CICES classification with improved matching...")
        classification_results = self.classify_articles_by_cices()
        
        # 计算CICES文献总数（去重）
        total_cices_articles = self.calculate_total_cices_articles(classification_results)
        
        print(f"分类结果: Provisioning={len(classification_results['Provisioning'])}, "
              f"Regulating={len(classification_results['Regulating'])}, "
              f"Cultural={len(classification_results['Cultural'])}")
        print(f"包含CICES分类的文献总数（去重）: {total_cices_articles}")
        
        # 步骤3: 计算总体指标
        print("📊 Calculating overall bibliometric indicators...")
        overall_indicators = self.calculate_bibliometric_indicators(
            self.articles, "Overall Articles"
        )
        
        # 步骤4: 计算各CICES类别指标
        print("📊 Calculating CICES category bibliometric indicators...")
        cices_indicators = {}
        for category in ['Provisioning', 'Regulating', 'Cultural']:
            if classification_results[category]:
                cices_indicators[category] = self.calculate_bibliometric_indicators(
                    classification_results[category], f"{category} Services"
                )
        
        # 步骤5: 生成报告
        self.generate_bibliometric_report(overall_indicators, cices_indicators, total_cices_articles)
        
        # 步骤6: 创建可视化
        print("\n📊 Creating visualizations...")
        self.create_bibliometric_visualizations(overall_indicators, cices_indicators)
        
        # 步骤7: 导出结果
        print("💾 Exporting analysis results...")
        self.export_bibliometric_results(overall_indicators, cices_indicators, classification_results, total_cices_articles)
        
        print("\n" + "="*80)
        print("✅ Bibliometric analysis completed! Generated files:")
        print("  📊 bibliometric_analysis.png")
        print("  📁 bibliometric_analysis.xlsx")
        print("="*80)

# 运行分析
if __name__ == "__main__":
    analyzer = EcosystemServicesBibliometricAnalyzer()
    analyzer.run_bibliometric_analysis()
```

```python
import os
import re
import pandas as pd
from collections import Counter, defaultdict
import hashlib

class EcosystemServicesAIAnalyzer:
    def __init__(self, data_directory="./"):
        self.data_directory = data_directory
        self.all_articles = []
        self.articles = []  # 所有有关键词的文献 - 这是我们要分析的数据集
        
        # 筛选后的AI关键词（不包含传统关键词）
        self.ai_keywords = [
            "machine learning", "supervised learning", "unsupervised learning", 
            "association rules", "dimensionality reduction", 
            "semi supervised learning", "self training", "low density separation models", 
            "graph based algorithms", "generative models", "reinforcement learning", 
            "deep learning", "deep feed forward networks", "convolution neural networks", 
            "siamese neural networks", "transformers", "generative adversarial networks", 
            "graph neural networks", "recurrent neural network", "active learning", 
            "uncertainty sampling", "diversity sampling", "transfer learning", 
            "instance-based transfer learning", "feature-based transfer learning", 
            "parameter-based transfer learning", "continual learning", 
            "regularization-based continual learning", "memory reply", "dynamic architectures", 
            "natural language processing", "machine translation", "question answering", 
            "sentiment analysis", "text generation", "topic modeling", "graphplan", 
            "satplan", "slam algorithms", "fastslam", "graphslam", "motion planning algorithms", 
            "sampling-based algorithms", "rrt", "graph search algorithms", "a-star", 
            "reaction-based algorithms", "dwa", "interpolating curve algorithms", 
            "image recognition", "machine vision", "nerf", "vision transformer", 
            "tacotron", "vits", "wavenet", "samplern", "expert systems", 
            "frame-based systems", "certainty factors", "RNN", "CNN", "LSTM", 
            "Long Short-Term Memory", "ML", "DL", "RL"
        ]
        
        # 传统AI/统计关键词
        self.traditional_ai_keywords = [
            "regression", "classification", "clustering"
        ]

        # CICES分类
        self.cices_categories = {
            'Provisioning': {
                'terrestrial_crops': [
                    'crops', 'cereals', 'root crops', 'soft fruit', 'fodder', 'livestock feed',
                    'fibres', 'materials', 'cultivated plants', 'fungi', 'algae', 'bacteria',
                    'plantation', 'energy source', 'bioenergy'
                ],
                'aquatic_crops': [
                    'aquaculture', 'in-situ aquaculture', 'cultivated aquatic plants'
                ],
                'reared_animals': [
                    'reared animals', 'beef', 'dairy', 'animal products',
                    'fibres from animals', 'mechanical energy', 'biogas'
                ],
                'aquatic_animals': [
                    'reared aquatic animals', 'in-situ aquaculture animals'
                ],
                'wild_plants': [
                    'wild plants', 'terrestrial plants', 'aquatic plants', 'wild fungi', 'wild algae',
                    'wild plant materials', 'wild plant energy'
                ],
                'wild_animals': [
                    'wild animals', 'terrestrial animals', 'aquatic animals',
                    'wild animal nutrition', 'wild animal materials'
                ],
                'genetic_material': [
                    'genetic material', 'seeds', 'spores', 'plant materials',
                    'breeding', 'new strains', 'varieties',
                    'individual genes', 'biological entities',
                    'animal genetic material', 'population establishment'
                ]
            },
            'Regulating': {
                'biochemical_transformation': [
                    'bio-remediation', 'micro-organisms', 'pollution degradation',
                    'filtration', 'sequestration', 'storage', 'accumulation',
                    'waste mediation', 'toxic substances', 'anthropogenic origin',
                    'smell reduction', 'noise attenuation', 'visual screening'
                ],
                'flow_extreme_events': [
                    'erosion control', 'water erosion', 'wind erosion',
                    'hydrological cycle', 'water flow regulation', 'runoff', 'base flows', 'peak flows',
                    'hazard mitigation', 'mass movement', 'landslides', 'flood mitigation', 'surge protection',
                    'wind protection', 'fire protection'
                ],
                'lifecycle_maintenance': [
                    'pollination', 'gamete dispersal', 'seed dispersal',
                    'nursery populations', 'breeding grounds', 'gene pool protection',
                    'refuge habitats', 'feeding grounds'
                ],
                'pest_disease_control': [
                    'pest control', 'invasive species', 'disease control'
                ],
                'soil_quality': [
                    'soil quality', 'weathering processes', 'decomposition', 'fixing processes',
                    'soil structure', 'biological agents', 'soil porosity'
                ],
                'water_conditions': [
                    'water conditions', 'chemical condition', 'macronutrients',
                    'freshwater regulation', 'saltwater regulation'
                ],
                'atmospheric_conditions': [
                    'atmospheric composition', 'oceanic circulation', 'climate change',
                    'temperature regulation', 'humidity', 'ventilation', 'transpiration', 'microclimate'
                ]
            },
            'Cultural': {
                'physical_experiential': [
                    'recreational activities', 'health promotion', 'recuperation', 'enjoyment',
                    'active interactions', 'immersive interactions',
                    'passive interactions', 'observational interactions'
                ],
                'intellectual_representative': [
                    'scientific investigation', 'traditional ecological knowledge', 'genetic diversity',
                    'education', 'training',
                    'culture', 'heritage', 'cultural resonance',
                    'aesthetic experiences'
                ],
                'indirect_interactions': [
                    'entertainment', 'representation outside setting'
                ],
                'spiritual_symbolic': [
                    'symbolic meaning', 'sense of place', 'distinctiveness',
                    'spiritual meaning', 'religious meaning',
                    'contemporary existence', 'conservation importance',
                    'inter-generational existence', 'genetic diversity conservation'
                ]
            }
        }

    def read_literature_files(self):
        """读取所有文献文件"""
        files = []
        for f in os.listdir(self.data_directory):
            if f.endswith('.txt') and 'download' in f and any(year in f for year in 
                ['2015', '2016', '2017', '2018', '2019', '2020', '2021', '2022', '2023', '2024', '2025']):
                files.append(f)
        
        files.sort()
        print(f"📚 找到 {len(files)} 个文献文件")
        
        for file in files:
            file_path = os.path.join(self.data_directory, file)
            try:
                with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
                    content = f.read()
                    self.parse_articles(content, file)
            except Exception as e:
                print(f"❌ 读取文件 {file} 时出错: {e}")
        
        print(f"✅ 总共解析了 {len(self.all_articles)} 篇文献")
        print(f"✅ 有关键词的文献: {len(self.articles)} 篇")

    def parse_articles(self, content, filename):
        """解析文献内容"""
        articles = re.split(r'\nER\s*\n', content)
        
        for article_text in articles:
            if not article_text.strip():
                continue
            
            if not re.search(r'(^|\n)PT\s+', article_text):
                continue
            
            article_data = {
                'filename': filename,
                'keywords': set(),
                'keywords_plus': set(),
                'title': '',
                'year': '',
                'abstract': '',
                'journal': '',
                'authors': []
            }
            
            # 提取ID字段
            id_match = re.search(r'\nID\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if id_match:
                id_content = id_match.group(1)
                keywords_plus = self.extract_keywords_from_field(id_content)
                article_data['keywords_plus'].update(keywords_plus)
            
            # 提取DE字段
            de_match = re.search(r'\nDE\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if de_match:
                de_content = de_match.group(1)
                keywords = self.extract_keywords_from_field(de_content)
                article_data['keywords'].update(keywords)
            
            # 提取标题
            ti_match = re.search(r'\nTI\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if ti_match:
                article_data['title'] = ti_match.group(1).replace('\n', ' ').strip()
            
            # 提取年份
            py_match = re.search(r'\nPY\s+(\d{4})', article_text)
            if py_match:
                article_data['year'] = py_match.group(1)
            
            # 提取摘要
            ab_match = re.search(r'\nAB\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if ab_match:
                article_data['abstract'] = ab_match.group(1).replace('\n', ' ').strip()
            
            # 提取期刊
            so_match = re.search(r'\nSO\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if so_match:
                article_data['journal'] = so_match.group(1).replace('\n', ' ').strip()
            
            # 提取作者
            au_match = re.search(r'\nAU\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if au_match:
                authors_text = au_match.group(1)
                authors = [author.strip() for author in authors_text.split('\n') if author.strip()]
                article_data['authors'] = authors
            
            self.all_articles.append(article_data)
            
            if article_data['keywords'] or article_data['keywords_plus']:
                article_data['all_keywords'] = article_data['keywords'] | article_data['keywords_plus']
                self.articles.append(article_data)

    def extract_keywords_from_field(self, field_content):
        """从ID或DE字段内容中提取关键词"""
        if not field_content:
            return []
            
        field_content = field_content.replace('\n', ' ')
        keywords = re.split(r'[;\n]', field_content)
        
        cleaned_keywords = []
        for keyword in keywords:
            keyword = keyword.strip().lower()
            keyword = re.sub(r'\s+', ' ', keyword)
            keyword = keyword.strip('.,;()[]{}"\'-')
            
            if keyword and len(keyword) > 2:
                cleaned_keywords.append(keyword)
                
        return cleaned_keywords
    
    def generate_article_hash(self, article):
        """生成文献唯一标识符，用于去重"""
        # 使用标题、年份、第一作者生成hash
        identifier_string = f"{article.get('title', '')}{article.get('year', '')}"
        if article.get('authors'):
            identifier_string += article['authors'][0]
        return hashlib.md5(identifier_string.encode()).hexdigest()
    
    def is_complete_word_match(self, article_keyword, cices_keyword):
        """
        检查CICES关键词是否作为完整词或短语出现在文献关键词中
        确保避免部分匹配（如"crop"匹配"microscopy"）
        """
        article_kw = article_keyword.lower().strip()
        cices_kw = cices_keyword.lower().strip()
        
        # 如果是完全相同，直接匹配
        if article_kw == cices_kw:
            return True
        
        # 创建词边界匹配模式
        # 对于单词，使用\b词边界；对于短语，确保前后是空格或字符串边界
        if ' ' in cices_kw:
            # 多词短语：确保短语作为完整单位出现
            pattern = r'(?:^|\s)' + re.escape(cices_kw) + r'(?:\s|$)'
        else:
            # 单词：使用词边界确保完整单词匹配
            pattern = r'\b' + re.escape(cices_kw) + r'\b'
        
        # 双向匹配：检查cices_keyword是否在article_keyword中，或反之
        match_in_article = bool(re.search(pattern, article_kw))
        
        # 反向匹配：检查article_keyword是否作为完整词出现在cices_keyword中
        if ' ' in article_kw:
            reverse_pattern = r'(?:^|\s)' + re.escape(article_kw) + r'(?:\s|$)'
        else:
            reverse_pattern = r'\b' + re.escape(article_kw) + r'\b'
        match_in_cices = bool(re.search(reverse_pattern, cices_kw))
        
        return match_in_article or match_in_cices

    def classify_articles_by_cices(self):
        """根据CICES框架对文献进行分类，使用完整词匹配并确保去重"""
        classification_results = {
            'Provisioning': [],
            'Regulating': [],
            'Cultural': []
        }
        
        # 使用字典追踪已添加的文献（基于hash去重）
        added_articles = {
            'Provisioning': set(),
            'Regulating': set(), 
            'Cultural': set()
        }
        
        # 分类文献
        for article in self.articles:
            article_categories = set()
            article_hash = self.generate_article_hash(article)
            
            # 准备搜索文本 - 与AI检索使用相同的域
            title_abstract = (article.get('title', '') + ' ' + article.get('abstract', '')).lower()
            keywords_text = ' '.join(article['all_keywords']).lower()
            full_text = title_abstract + ' ' + keywords_text
            
            for main_category in ['Provisioning', 'Regulating', 'Cultural']:
                for subcategory, cices_keywords in self.cices_categories[main_category].items():
                    for cices_keyword in cices_keywords:
                        # 在相同的检索域（全文）中使用完整词匹配逻辑
                        if self.is_complete_word_match(full_text, cices_keyword):
                            article_categories.add(main_category)
                            
                            # 只有当文献hash未被添加过才添加到结果中
                            if article_hash not in added_articles[main_category]:
                                classification_results[main_category].append(article)
                                added_articles[main_category].add(article_hash)
                            break
                    if main_category in article_categories:
                        break
                if main_category in article_categories:
                    continue  # 继续检查其他类别，因为一篇文章可能属于多个类别
            
            article['cices_categories'] = list(article_categories)
        
        return classification_results

    def analyze_ai_usage(self):
        """分析AI技术使用情况 - 基于所有有关键词的文献"""
        print("🤖 开始分析AI技术使用情况...")
        
        ai_results = {
            # 1. 筛选后AI关键词文献
            'ai_full_text': [],        # 标题+摘要+ID+DE中包含筛选后AI关键词的文献
            'ai_keywords_only': [],    # ID+DE中包含筛选后AI关键词的文献
            
            # 2. 传统AI关键词文献
            'traditional_full_text': [],    # 标题+摘要+ID+DE中包含传统AI关键词的文献
            'traditional_keywords_only': [], # ID+DE中包含传统AI关键词的文献
            
            # 3. 先进传统AI文献（既包含传统关键词又包含筛选后AI关键词）
            'advanced_traditional_full_text': [],    # 先进的传统AI（全文搜索）
            'advanced_traditional_keywords_only': [], # 先进的传统AI（仅关键词）
            
            # 4. 频次统计（分别统计全文和关键词搜索）
            'ai_keyword_frequency_full_text': Counter(),  # AI关键词频次（全文搜索）
            'ai_keyword_frequency_keywords_only': Counter(),  # AI关键词频次（仅关键词搜索）
            'traditional_keyword_frequency_full_text': Counter(),  # 传统AI关键词频次（全文搜索）
            'traditional_keyword_frequency_keywords_only': Counter()  # 传统AI关键词频次（仅关键词搜索）
        }
        
        # 分析所有有关键词的文献
        for article in self.articles:
            # 准备搜索文本
            title_abstract = (article.get('title', '') + ' ' + article.get('abstract', '')).lower()
            keywords_text = ' '.join(article['all_keywords']).lower()
            full_text = title_abstract + ' ' + keywords_text
            
            # 1. 搜索筛选后的AI关键词
            matched_ai_full = []
            matched_ai_keywords = []
            
            for ai_keyword in self.ai_keywords:
                ai_keyword_lower = ai_keyword.lower()
                
                # 全文搜索
                if ai_keyword_lower in full_text:
                    matched_ai_full.append(ai_keyword)
                    ai_results['ai_keyword_frequency_full_text'][ai_keyword] += 1
                
                # 仅关键词搜索
                if ai_keyword_lower in keywords_text:
                    matched_ai_keywords.append(ai_keyword)
                    ai_results['ai_keyword_frequency_keywords_only'][ai_keyword] += 1
            
            # 记录包含筛选后AI关键词的文献
            if matched_ai_full:
                ai_results['ai_full_text'].append({
                    'article': article,
                    'matched_keywords': matched_ai_full
                })
            
            if matched_ai_keywords:
                ai_results['ai_keywords_only'].append({
                    'article': article,
                    'matched_keywords': matched_ai_keywords
                })
            
            # 2. 搜索传统AI关键词
            matched_traditional_full = []
            matched_traditional_keywords = []
            
            for trad_keyword in self.traditional_ai_keywords:
                trad_keyword_lower = trad_keyword.lower()
                
                # 全文搜索
                if trad_keyword_lower in full_text:
                    matched_traditional_full.append(trad_keyword)
                    ai_results['traditional_keyword_frequency_full_text'][trad_keyword] += 1
                
                # 仅关键词搜索
                if trad_keyword_lower in keywords_text:
                    matched_traditional_keywords.append(trad_keyword)
                    ai_results['traditional_keyword_frequency_keywords_only'][trad_keyword] += 1
            
            # 记录包含传统AI关键词的文献
            if matched_traditional_full:
                traditional_data_full = {
                    'article': article,
                    'matched_traditional': matched_traditional_full,
                    'is_advanced': bool(matched_ai_full),  # 是否同时包含筛选后AI关键词
                    'matched_ai': matched_ai_full
                }
                ai_results['traditional_full_text'].append(traditional_data_full)
                
                # 如果是先进的，也加入先进列表
                if traditional_data_full['is_advanced']:
                    ai_results['advanced_traditional_full_text'].append(traditional_data_full)
            
            if matched_traditional_keywords:
                traditional_data_keywords = {
                    'article': article,
                    'matched_traditional': matched_traditional_keywords,
                    'is_advanced': bool(matched_ai_keywords),  # 是否同时包含筛选后AI关键词
                    'matched_ai': matched_ai_keywords
                }
                ai_results['traditional_keywords_only'].append(traditional_data_keywords)
                
                # 如果是先进的，也加入先进列表
                if traditional_data_keywords['is_advanced']:
                    ai_results['advanced_traditional_keywords_only'].append(traditional_data_keywords)
        
        return ai_results

    def analyze_ai_by_cices(self, ai_results, classification_results):
        """按CICES类别分析AI使用情况"""
        ai_by_cices = {
            'Provisioning': {
                'ai_full_text': [], 'ai_keywords_only': [],
                'traditional_full_text': [], 'traditional_keywords_only': [],
                'advanced_traditional_full_text': [], 'advanced_traditional_keywords_only': []
            },
            'Regulating': {
                'ai_full_text': [], 'ai_keywords_only': [],
                'traditional_full_text': [], 'traditional_keywords_only': [],
                'advanced_traditional_full_text': [], 'advanced_traditional_keywords_only': []
            },
            'Cultural': {
                'ai_full_text': [], 'ai_keywords_only': [],
                'traditional_full_text': [], 'traditional_keywords_only': [],
                'advanced_traditional_full_text': [], 'advanced_traditional_keywords_only': []
            }
        }
        
        # 建立文献到CICES类别的映射
        article_to_categories = {}
        for category in ['Provisioning', 'Regulating', 'Cultural']:
            for article in classification_results[category]:
                article_id = id(article)
                if article_id not in article_to_categories:
                    article_to_categories[article_id] = []
                article_to_categories[article_id].append(category)
        
        # 分类各类AI文献
        for result_type in ['ai_full_text', 'ai_keywords_only', 'traditional_full_text', 
                           'traditional_keywords_only', 'advanced_traditional_full_text', 
                           'advanced_traditional_keywords_only']:
            for article_data in ai_results[result_type]:
                article = article_data['article']
                article_id = id(article)
                for category in article_to_categories.get(article_id, []):
                    ai_by_cices[category][result_type].append(article_data)
        
        return ai_by_cices

    def analyze_yearly_ai_trends(self, ai_results):
        """分析各年份AI技术使用趋势和先进率变化 - 基于所有有关键词的文献"""
        yearly_ai_stats = defaultdict(lambda: {
            'total_articles': 0,
            'ai_full_text': 0,
            'ai_keywords_only': 0,
            'traditional_full_text': 0,
            'traditional_keywords_only': 0,
            'advanced_traditional_full_text': 0,
            'advanced_traditional_keywords_only': 0
        })
        
        # 统计各年份文献总数
        for article in self.articles:
            year = article.get('year', 'Unknown')
            yearly_ai_stats[year]['total_articles'] += 1
        
        # 统计各年份AI文献数
        for article_data in ai_results['ai_full_text']:
            year = article_data['article'].get('year', 'Unknown')
            yearly_ai_stats[year]['ai_full_text'] += 1
            
        for article_data in ai_results['ai_keywords_only']:
            year = article_data['article'].get('year', 'Unknown')
            yearly_ai_stats[year]['ai_keywords_only'] += 1
        
        # 统计各年份传统AI文献数
        for article_data in ai_results['traditional_full_text']:
            year = article_data['article'].get('year', 'Unknown')
            yearly_ai_stats[year]['traditional_full_text'] += 1
            
        for article_data in ai_results['traditional_keywords_only']:
            year = article_data['article'].get('year', 'Unknown')
            yearly_ai_stats[year]['traditional_keywords_only'] += 1
        
        # 统计各年份先进传统AI文献数
        for article_data in ai_results['advanced_traditional_full_text']:
            year = article_data['article'].get('year', 'Unknown')
            yearly_ai_stats[year]['advanced_traditional_full_text'] += 1
            
        for article_data in ai_results['advanced_traditional_keywords_only']:
            year = article_data['article'].get('year', 'Unknown')
            yearly_ai_stats[year]['advanced_traditional_keywords_only'] += 1
        
        return yearly_ai_stats

    def generate_ai_report(self, ai_results, ai_by_cices, main_categories_count):
        """生成AI分析报告 - 基于所有有关键词的文献"""
        print("\n" + "="*80)
        print("🤖 生态系统服务文献AI技术分析报告")
        print("="*80)
        
        # 1. 总体AI使用统计
        print(f"\n📊 一、总体AI技术使用统计")
        print(f"  ├─ 有关键词文献总数: {len(self.articles)}")
        print(f"  ├─ AI文献数（全文搜索）: {len(ai_results['ai_full_text'])}")
        print(f"  ├─ AI文献数（仅关键词）: {len(ai_results['ai_keywords_only'])}")
        print(f"  ├─ AI使用率（全文搜索）: {len(ai_results['ai_full_text'])/len(self.articles)*100:.2f}%")
        print(f"  └─ AI使用率（仅关键词）: {len(ai_results['ai_keywords_only'])/len(self.articles)*100:.2f}%")
        
        # 2. 传统AI关键词先进性分析
        print(f"\n🧠 二、传统AI关键词先进性分析")
        print(f"  📍 全文搜索结果：")
        print(f"    ├─ 传统AI文献数: {len(ai_results['traditional_full_text'])}")
        print(f"    ├─ 先进传统AI文献数: {len(ai_results['advanced_traditional_full_text'])}")
        if len(ai_results['traditional_full_text']) > 0:
            advanced_ratio = len(ai_results['advanced_traditional_full_text'])/len(ai_results['traditional_full_text'])*100
            print(f"    └─ 先进率: {advanced_ratio:.2f}%")
        else:
            print(f"    └─ 先进率: 0.00%")
            
        print(f"  📍 仅关键词搜索结果：")
        print(f"    ├─ 传统AI文献数: {len(ai_results['traditional_keywords_only'])}")
        print(f"    ├─ 先进传统AI文献数: {len(ai_results['advanced_traditional_keywords_only'])}")
        if len(ai_results['traditional_keywords_only']) > 0:
            advanced_ratio_kw = len(ai_results['advanced_traditional_keywords_only'])/len(ai_results['traditional_keywords_only'])*100
            print(f"    └─ 先进率: {advanced_ratio_kw:.2f}%")
        else:
            print(f"    └─ 先进率: 0.00%")
        
        # 3. 各CICES类别的AI使用情况
        print(f"\n🌳 三、各CICES类别的AI技术使用情况")
        print(f"  📍 全文搜索结果：")
        print(f"  {'类别':<12} {'总数':<8} {'AI':<8} {'传统AI':<8} {'先进':<10} {'AI率%':<8} {'先进率%':<8}")
        print("  " + "-" * 70)
        
        for category in ['Provisioning', 'Regulating', 'Cultural']:
            total_docs = main_categories_count.get(category, 0)
            ai_full = len(ai_by_cices[category]['ai_full_text'])
            traditional_full = len(ai_by_cices[category]['traditional_full_text'])
            advanced_full = len(ai_by_cices[category]['advanced_traditional_full_text'])
            ai_rate = (ai_full / total_docs * 100) if total_docs > 0 else 0
            advanced_rate = (advanced_full / traditional_full * 100) if traditional_full > 0 else 0
            
            print(f"  {category:<12} {total_docs:<8} {ai_full:<8} {traditional_full:<8} {advanced_full:<10} {ai_rate:<8.2f} {advanced_rate:<8.2f}")
        
        print(f"\n  📍 仅关键词搜索结果：")
        print(f"  {'类别':<12} {'总数':<8} {'AI':<8} {'传统AI':<8} {'先进':<10} {'AI率%':<8} {'先进率%':<8}")
        print("  " + "-" * 70)
        
        for category in ['Provisioning', 'Regulating', 'Cultural']:
            total_docs = main_categories_count.get(category, 0)
            ai_keywords = len(ai_by_cices[category]['ai_keywords_only'])
            traditional_keywords = len(ai_by_cices[category]['traditional_keywords_only'])
            advanced_keywords = len(ai_by_cices[category]['advanced_traditional_keywords_only'])
            ai_rate = (ai_keywords / total_docs * 100) if total_docs > 0 else 0
            advanced_rate = (advanced_keywords / traditional_keywords * 100) if traditional_keywords > 0 else 0
            
            print(f"  {category:<12} {total_docs:<8} {ai_keywords:<8} {traditional_keywords:<8} {advanced_keywords:<10} {ai_rate:<8.2f} {advanced_rate:<8.2f}")
        
        # 4. 高频AI技术统计
        print(f"\n🔧 四、高频AI技术统计（前20）")
        print(f"  📍 全文搜索结果：")
        print(f"  {'排名':<6} {'AI技术':<45} {'频次':<10} {'使用率%':<10}")
        print("  " + "-" * 75)
        
        for i, (ai_tech, count) in enumerate(ai_results['ai_keyword_frequency_full_text'].most_common(20), 1):
            usage_rate = (count / len(self.articles)) * 100
            print(f"  {i:<6} {ai_tech[:43]:<45} {count:<10} {usage_rate:<10.2f}")
        
        print(f"\n  📍 仅关键词搜索结果：")
        print(f"  {'排名':<6} {'AI技术':<45} {'频次':<10} {'使用率%':<10}")
        print("  " + "-" * 75)
        
        for i, (ai_tech, count) in enumerate(ai_results['ai_keyword_frequency_keywords_only'].most_common(20), 1):
            usage_rate = (count / len(self.articles)) * 100
            print(f"  {i:<6} {ai_tech[:43]:<45} {count:<10} {usage_rate:<10.2f}")
        
        # 5. 传统AI关键词使用统计
        print(f"\n📊 五、传统AI关键词使用统计")
        print(f"  📍 全文搜索结果：")
        print(f"  {'关键词':<15} {'总数':<8} {'先进数':<10} {'先进率%':<10}")
        print("  " + "-" * 50)
        
        for trad_keyword in self.traditional_ai_keywords:
            total_usage = ai_results['traditional_keyword_frequency_full_text'].get(trad_keyword, 0)
            advanced_usage_full = sum(1 for article_data in ai_results['advanced_traditional_full_text'] 
                                     if trad_keyword in article_data['matched_traditional'])
            
            advanced_rate = (advanced_usage_full / total_usage * 100) if total_usage > 0 else 0
            print(f"  {trad_keyword:<15} {total_usage:<8} {advanced_usage_full:<10} {advanced_rate:<10.2f}")
        
        print(f"\n  📍 仅关键词搜索结果：")
        print(f"  {'关键词':<15} {'总数':<8} {'先进数':<10} {'先进率%':<10}")
        print("  " + "-" * 50)
        
        for trad_keyword in self.traditional_ai_keywords:
            total_usage_kw = ai_results['traditional_keyword_frequency_keywords_only'].get(trad_keyword, 0)
            advanced_usage_kw = sum(1 for article_data in ai_results['advanced_traditional_keywords_only'] 
                                   if trad_keyword in article_data['matched_traditional'])
            
            advanced_rate_kw = (advanced_usage_kw / total_usage_kw * 100) if total_usage_kw > 0 else 0
            print(f"  {trad_keyword:<15} {total_usage_kw:<8} {advanced_usage_kw:<10} {advanced_rate_kw:<10.2f}")
        
        # 6. 年度AI先进率变化趋势
        yearly_stats = self.analyze_yearly_ai_trends(ai_results)
        
        print(f"\n📈 六、年度AI先进率变化趋势")
        
        # 筛选有效年份并排序
        valid_years = sorted([year for year in yearly_stats.keys() if year.isdigit() and year != 'Unknown'])
        
        if valid_years:
            print(f"  📍 全文搜索结果：")
            print(f"  {'年份':<6} {'总数':<8} {'AI':<8} {'传统AI':<8} {'先进':<10} {'AI率%':<8} {'先进率%':<10}")
            print("  " + "-" * 75)
            
            for year in valid_years:
                stats = yearly_stats[year]
                total = stats['total_articles']
                ai_full = stats['ai_full_text']
                trad_full = stats['traditional_full_text']
                adv_full = stats['advanced_traditional_full_text']
                
                ai_rate = (ai_full / total * 100) if total > 0 else 0
                advanced_rate = (adv_full / trad_full * 100) if trad_full > 0 else 0
                
                print(f"  {year:<6} {total:<8} {ai_full:<8} {trad_full:<8} {adv_full:<10} {ai_rate:<8.2f} {advanced_rate:<10.2f}")
            
            print(f"\n  📍 仅关键词搜索结果：")
            print(f"  {'年份':<6} {'总数':<8} {'AI':<8} {'传统AI':<8} {'先进':<10} {'AI率%':<8} {'先进率%':<10}")
            print("  " + "-" * 75)
            
            for year in valid_years:
                stats = yearly_stats[year]
                total = stats['total_articles']
                ai_kw = stats['ai_keywords_only']
                trad_kw = stats['traditional_keywords_only']
                adv_kw = stats['advanced_traditional_keywords_only']
                
                ai_rate = (ai_kw / total * 100) if total > 0 else 0
                advanced_rate = (adv_kw / trad_kw * 100) if trad_kw > 0 else 0
                
                print(f"  {year:<6} {total:<8} {ai_kw:<8} {trad_kw:<8} {adv_kw:<10} {ai_rate:<8.2f} {advanced_rate:<10.2f}")
        else:
            print("  无有效年份数据")
        
        return yearly_stats

    def export_ai_results(self, ai_results, ai_by_cices, main_categories_count, yearly_stats, total_cices_articles):
        """导出AI分析结果到Excel - 基于所有有关键词的文献，仅关键词字段检索"""
        with pd.ExcelWriter('./ecosystem_services_ai_analysis.xlsx', engine='openpyxl') as writer:
            
            # 0. CICES分类统计
            cices_summary_data = {
                '指标': [
                    '有关键词文献总数',
                    '包含CICES分类的文献总数（去重）', 
                    'CICES覆盖率(%)',
                    'Provisioning服务文献数',
                    'Regulating服务文献数', 
                    'Cultural服务文献数'
                ],
                '数值': [
                    len(self.articles),
                    total_cices_articles,
                    round((total_cices_articles/len(self.articles)*100), 2),
                    main_categories_count.get('Provisioning', 0),
                    main_categories_count.get('Regulating', 0),
                    main_categories_count.get('Cultural', 0)
                ]
            }
            cices_summary_df = pd.DataFrame(cices_summary_data)
            cices_summary_df.to_excel(writer, sheet_name='CICES分类统计', index=False)
            
            # 1. 总体AI使用统计
            summary_data = {
                '指标': [
                    '有关键词文献总数', 
                    'AI文献数（仅关键词）',
                    'AI使用率（仅关键词）%',
                    '传统AI文献数（仅关键词）',
                    '先进传统AI文献数（仅关键词）',
                    '传统AI先进率（仅关键词）%'
                ],
                '数值': [
                    len(self.articles),
                    len(ai_results['ai_keywords_only']),
                    round(len(ai_results['ai_keywords_only'])/len(self.articles)*100, 2),
                    len(ai_results['traditional_keywords_only']),
                    len(ai_results['advanced_traditional_keywords_only']),
                    round(len(ai_results['advanced_traditional_keywords_only'])/len(ai_results['traditional_keywords_only'])*100, 2) if len(ai_results['traditional_keywords_only']) > 0 else 0
                ]
            }
            summary_df = pd.DataFrame(summary_data)
            summary_df.to_excel(writer, sheet_name='总体统计', index=False)
            
            # 2. CICES类别AI使用统计
            cices_data = []
            for category in ['Provisioning', 'Regulating', 'Cultural']:
                total_docs = main_categories_count.get(category, 0)
                cices_data.append({
                    '类别': category,
                    '文献总数': total_docs,
                    'AI文献数': len(ai_by_cices[category]['ai_keywords_only']),
                    '传统AI文献数': len(ai_by_cices[category]['traditional_keywords_only']),
                    '先进传统AI文献数': len(ai_by_cices[category]['advanced_traditional_keywords_only']),
                    'AI使用率_%': round(len(ai_by_cices[category]['ai_keywords_only'])/total_docs*100, 2) if total_docs > 0 else 0,
                    '先进率_%': round(len(ai_by_cices[category]['advanced_traditional_keywords_only'])/len(ai_by_cices[category]['traditional_keywords_only'])*100, 2) if len(ai_by_cices[category]['traditional_keywords_only']) > 0 else 0
                })
            
            cices_df = pd.DataFrame(cices_data)
            cices_df.to_excel(writer, sheet_name='CICES类别AI使用', index=False)
            
            # 3. AI技术频次统计
            ai_tech_data_keywords = []
            for ai_tech, count in ai_results['ai_keyword_frequency_keywords_only'].most_common():
                ai_tech_data_keywords.append({
                    'AI技术': ai_tech,
                    '频次': count,
                    '使用率_%': round((count/len(self.articles))*100, 2)
                })
            ai_tech_keywords_df = pd.DataFrame(ai_tech_data_keywords)
            ai_tech_keywords_df.to_excel(writer, sheet_name='AI技术频次', index=False)
            
            # 4. 传统AI关键词先进性分析
            traditional_data = []
            for trad_keyword in self.traditional_ai_keywords:
                total_usage_kw = ai_results['traditional_keyword_frequency_keywords_only'].get(trad_keyword, 0)
                advanced_usage_kw = sum(1 for article_data in ai_results['advanced_traditional_keywords_only'] 
                                       if trad_keyword in article_data['matched_traditional'])
                
                traditional_data.append({
                    '传统关键词': trad_keyword,
                    '总使用数': total_usage_kw,
                    '先进使用数': advanced_usage_kw,
                    '先进率_%': round((advanced_usage_kw / total_usage_kw * 100), 2) if total_usage_kw > 0 else 0
                })
            
            traditional_df = pd.DataFrame(traditional_data)
            traditional_df.to_excel(writer, sheet_name='传统AI先进性', index=False)
            
            # 5. 年度AI使用趋势
            valid_years = sorted([year for year in yearly_stats.keys() if year.isdigit() and year != 'Unknown'])
            if valid_years:
                yearly_data = []
                for year in valid_years:
                    stats = yearly_stats[year]
                    yearly_data.append({
                        '年份': year,
                        '文献总数': stats['total_articles'],
                        'AI文献数': stats['ai_keywords_only'],
                        '传统AI文献数': stats['traditional_keywords_only'],
                        '先进传统AI文献数': stats['advanced_traditional_keywords_only'],
                        'AI使用率_%': round((stats['ai_keywords_only'] / stats['total_articles'] * 100), 2) if stats['total_articles'] > 0 else 0,
                        '先进率_%': round((stats['advanced_traditional_keywords_only'] / stats['traditional_keywords_only'] * 100), 2) if stats['traditional_keywords_only'] > 0 else 0
                    })
                
                yearly_df = pd.DataFrame(yearly_data)
                yearly_df.to_excel(writer, sheet_name='年度AI趋势', index=False)
        
        print("\n✅ AI分析结果已导出到 ecosystem_services_ai_analysis.xlsx")

    def run_ai_analysis(self):
        """运行AI专项分析"""
        print("🤖 开始生态系统服务文献AI技术分析...")
        print("="*80)
        
        # 步骤1: 读取文献
        self.read_literature_files()
        
        if not self.all_articles:
            print("❌ 未找到有效文献！")
            return
        
        if not self.articles:
            print("❌ 未找到有关键词的文献！")
            return
        
        print(f"\n📊 WOS检索已确保所有文献都是生态系统服务相关")
        print(f"📊 正在分析 {len(self.articles)} 篇有关键词的文献")
        print(f"📊 检索范围：仅在ID+DE关键词字段中检索")
        
        # 步骤2: CICES分类 - 使用改进的完整词匹配逻辑
        print("📊 正在进行CICES分类...")
        classification_results = self.classify_articles_by_cices()
        
        # 计算CICES文献总数（去重）
        total_cices_articles = self.calculate_total_cices_articles(classification_results)
        
        main_categories_count = {}
        for category in ['Provisioning', 'Regulating', 'Cultural']:
            main_categories_count[category] = len(classification_results[category])
        
        print(f"CICES分类结果: Provisioning={main_categories_count.get('Provisioning', 0)}, "
              f"Regulating={main_categories_count.get('Regulating', 0)}, "
              f"Cultural={main_categories_count.get('Cultural', 0)}")
        print(f"包含CICES分类的文献总数（去重）: {total_cices_articles}")
        
        # 步骤3: AI技术使用分析
        ai_results = self.analyze_ai_usage()
        
        # 步骤4: 按CICES分类分析AI使用
        print("📊 正在按CICES类别分析AI使用情况...")
        ai_by_cices = self.analyze_ai_by_cices(ai_results, classification_results)
        
        # 步骤5: 生成报告（包含年度趋势分析）
        yearly_stats = self.generate_ai_report(ai_results, ai_by_cices, main_categories_count, total_cices_articles)
        
        # 步骤6: 导出结果
        print("💾 正在导出分析结果...")
        self.export_ai_results(ai_results, ai_by_cices, main_categories_count, yearly_stats, total_cices_articles)
        
        print("\n" + "="*80)
        print("✅ AI分析完成！生成的文件：")
        print("  📁 ecosystem_services_ai_analysis.xlsx")
        print("="*80)

# 运行分析
if __name__ == "__main__":
    analyzer = EcosystemServicesAIAnalyzer()
    analyzer.run_ai_analysis()
```

```python
import os
import re
import pandas as pd
from collections import Counter, defaultdict

class EcosystemServicesAIAnalyzer:
    def __init__(self, data_directory="./"):
        self.data_directory = data_directory
        self.all_articles = []
        self.articles = []  # 所有有关键词的文献 - 这是我们要分析的数据集
        
        # 筛选后的AI关键词（不包含传统关键词）
        self.ai_keywords = [
            "machine learning", "supervised learning", "unsupervised learning", 
            "association rules", "dimensionality reduction", 
            "semi supervised learning", "self training", "low density separation models", 
            "graph based algorithms", "generative models", "reinforcement learning", 
            "deep learning", "deep feed forward networks", "convolution neural networks", 
            "siamese neural networks", "transformers", "generative adversarial networks", 
            "graph neural networks", "recurrent neural network", "active learning", 
            "uncertainty sampling", "diversity sampling", "transfer learning", 
            "instance-based transfer learning", "feature-based transfer learning", 
            "parameter-based transfer learning", "continual learning", 
            "regularization-based continual learning", "memory reply", "dynamic architectures", 
            "natural language processing", "machine translation", "question answering", 
            "sentiment analysis", "text generation", "topic modeling", "graphplan", 
            "satplan", "slam algorithms", "fastslam", "graphslam", "motion planning algorithms", 
            "sampling-based algorithms", "rrt", "graph search algorithms", "a-star", 
            "reaction-based algorithms", "dwa", "interpolating curve algorithms", 
            "image recognition", "machine vision", "nerf", "vision transformer", 
            "tacotron", "vits", "wavenet", "samplern", "expert systems", 
            "frame-based systems", "certainty factors", "RNN", "CNN", "LSTM", 
            "Long Short-Term Memory", "ML", "DL", "RL"
        ]
        
        # 传统AI/统计关键词
        self.traditional_ai_keywords = [
            "regression", "classification", "clustering"
        ]

        # CICES分类
        self.cices_categories = {
            'Provisioning': {
                'terrestrial_crops': [
                    'crops', 'cereals', 'root crops', 'soft fruit', 'fodder', 'livestock feed',
                    'fibres', 'materials', 'cultivated plants', 'fungi', 'algae', 'bacteria',
                    'plantation', 'energy source', 'bioenergy'
                ],
                'aquatic_crops': [
                    'aquaculture', 'in-situ aquaculture', 'cultivated aquatic plants'
                ],
                'reared_animals': [
                    'reared animals', 'beef', 'dairy', 'animal products',
                    'fibres from animals', 'mechanical energy', 'biogas'
                ],
                'aquatic_animals': [
                    'reared aquatic animals', 'in-situ aquaculture animals'
                ],
                'wild_plants': [
                    'wild plants', 'terrestrial plants', 'aquatic plants', 'wild fungi', 'wild algae',
                    'wild plant materials', 'wild plant energy'
                ],
                'wild_animals': [
                    'wild animals', 'terrestrial animals', 'aquatic animals',
                    'wild animal nutrition', 'wild animal materials'
                ],
                'genetic_material': [
                    'genetic material', 'seeds', 'spores', 'plant materials',
                    'breeding', 'new strains', 'varieties',
                    'individual genes', 'biological entities',
                    'animal genetic material', 'population establishment'
                ]
            },
            'Regulating': {
                'biochemical_transformation': [
                    'bio-remediation', 'micro-organisms', 'pollution degradation',
                    'filtration', 'sequestration', 'storage', 'accumulation',
                    'waste mediation', 'toxic substances', 'anthropogenic origin',
                    'smell reduction', 'noise attenuation', 'visual screening'
                ],
                'flow_extreme_events': [
                    'erosion control', 'water erosion', 'wind erosion',
                    'hydrological cycle', 'water flow regulation', 'runoff', 'base flows', 'peak flows',
                    'hazard mitigation', 'mass movement', 'landslides', 'flood mitigation', 'surge protection',
                    'wind protection', 'fire protection'
                ],
                'lifecycle_maintenance': [
                    'pollination', 'gamete dispersal', 'seed dispersal',
                    'nursery populations', 'breeding grounds', 'gene pool protection',
                    'refuge habitats', 'feeding grounds'
                ],
                'pest_disease_control': [
                    'pest control', 'invasive species', 'disease control'
                ],
                'soil_quality': [
                    'soil quality', 'weathering processes', 'decomposition', 'fixing processes',
                    'soil structure', 'biological agents', 'soil porosity'
                ],
                'water_conditions': [
                    'water conditions', 'chemical condition', 'macronutrients',
                    'freshwater regulation', 'saltwater regulation'
                ],
                'atmospheric_conditions': [
                    'atmospheric composition', 'oceanic circulation', 'climate change',
                    'temperature regulation', 'humidity', 'ventilation', 'transpiration', 'microclimate'
                ]
            },
            'Cultural': {
                'physical_experiential': [
                    'recreational activities', 'health promotion', 'recuperation', 'enjoyment',
                    'active interactions', 'immersive interactions',
                    'passive interactions', 'observational interactions'
                ],
                'intellectual_representative': [
                    'scientific investigation', 'traditional ecological knowledge', 'genetic diversity',
                    'education', 'training',
                    'culture', 'heritage', 'cultural resonance',
                    'aesthetic experiences'
                ],
                'indirect_interactions': [
                    'entertainment', 'representation outside setting'
                ],
                'spiritual_symbolic': [
                    'symbolic meaning', 'sense of place', 'distinctiveness',
                    'spiritual meaning', 'religious meaning',
                    'contemporary existence', 'conservation importance',
                    'inter-generational existence', 'genetic diversity conservation'
                ]
            }
        }

    def read_literature_files(self):
        """读取所有文献文件"""
        files = []
        for f in os.listdir(self.data_directory):
            if f.endswith('.txt') and 'download' in f and any(year in f for year in 
                ['2015', '2016', '2017', '2018', '2019', '2020', '2021', '2022', '2023', '2024', '2025']):
                files.append(f)
        
        files.sort()
        print(f"📚 找到 {len(files)} 个文献文件")
        
        for file in files:
            file_path = os.path.join(self.data_directory, file)
            try:
                with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
                    content = f.read()
                    self.parse_articles(content, file)
            except Exception as e:
                print(f"❌ 读取文件 {file} 时出错: {e}")
        
        print(f"✅ 总共解析了 {len(self.all_articles)} 篇文章")
        print(f"✅ 有关键词的文章: {len(self.articles)} 篇")

    def parse_articles(self, content, filename):
        """解析文章内容"""
        articles = re.split(r'\nER\s*\n', content)
        
        for article_text in articles:
            if not article_text.strip():
                continue
            
            if not re.search(r'(^|\n)PT\s+', article_text):
                continue
            
            article_data = {
                'filename': filename,
                'keywords': set(),
                'keywords_plus': set(),
                'title': '',
                'year': '',
                'abstract': '',
                'journal': '',
                'authors': []
            }
            
            # 提取ID字段
            id_match = re.search(r'\nID\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if id_match:
                id_content = id_match.group(1)
                keywords_plus = self.extract_keywords_from_field(id_content)
                article_data['keywords_plus'].update(keywords_plus)
            
            # 提取DE字段
            de_match = re.search(r'\nDE\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if de_match:
                de_content = de_match.group(1)
                keywords = self.extract_keywords_from_field(de_content)
                article_data['keywords'].update(keywords)
            
            # 提取标题
            ti_match = re.search(r'\nTI\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if ti_match:
                article_data['title'] = ti_match.group(1).replace('\n', ' ').strip()
            
            # 提取年份
            py_match = re.search(r'\nPY\s+(\d{4})', article_text)
            if py_match:
                article_data['year'] = py_match.group(1)
            
            # 提取摘要
            ab_match = re.search(r'\nAB\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if ab_match:
                article_data['abstract'] = ab_match.group(1).replace('\n', ' ').strip()
            
            # 提取期刊
            so_match = re.search(r'\nSO\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if so_match:
                article_data['journal'] = so_match.group(1).replace('\n', ' ').strip()
            
            # 提取作者
            au_match = re.search(r'\nAU\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if au_match:
                authors_text = au_match.group(1)
                authors = [author.strip() for author in authors_text.split('\n') if author.strip()]
                article_data['authors'] = authors
            
            self.all_articles.append(article_data)
            
            if article_data['keywords'] or article_data['keywords_plus']:
                article_data['all_keywords'] = article_data['keywords'] | article_data['keywords_plus']
                self.articles.append(article_data)

    def extract_keywords_from_field(self, field_content):
        """从ID或DE字段内容中提取关键词"""
        if not field_content:
            return []
            
        field_content = field_content.replace('\n', ' ')
        keywords = re.split(r'[;\n]', field_content)
        
        cleaned_keywords = []
        for keyword in keywords:
            keyword = keyword.strip().lower()
            keyword = re.sub(r'\s+', ' ', keyword)
            keyword = keyword.strip('.,;()[]{}"\'-')
            
            if keyword and len(keyword) > 2:
                cleaned_keywords.append(keyword)
                
        return cleaned_keywords

    def classify_articles_by_cices(self):
        """根据CICES框架对文献进行分类 - 修正：基于所有有关键词的文献"""
        classification_results = {
            'Provisioning': [],
            'Regulating': [],
            'Cultural': []
        }
        
        # 修正：直接对self.articles（所有有关键词的文献）进行分类
        for article in self.articles:
            article_categories = set()
            
            for main_category in ['Provisioning', 'Regulating', 'Cultural']:
                for subcategory, keywords in self.cices_categories[main_category].items():
                    for article_keyword in article['all_keywords']:
                        for cices_keyword in keywords:
                            if (cices_keyword.lower() in article_keyword.lower() or 
                                article_keyword.lower() in cices_keyword.lower()):
                                article_categories.add(main_category)
                                classification_results[main_category].append(article)
                                break
                        if main_category in article_categories:
                            break
                    if main_category in article_categories:
                        break
            
            article['cices_categories'] = list(article_categories)
        
        # 去重
        for category in classification_results:
            seen = set()
            unique_articles = []
            for article in classification_results[category]:
                article_id = id(article)
                if article_id not in seen:
                    seen.add(article_id)
                    unique_articles.append(article)
            classification_results[category] = unique_articles
        
        return classification_results

    def analyze_ai_usage(self):
        """分析AI技术使用情况 - 修正：基于所有有关键词的文献"""
        print("🤖 开始分析AI技术使用情况...")
        
        ai_results = {
            # 1. 筛选后AI关键词文献
            'ai_full_text': [],        # 标题+摘要+ID+DE中包含筛选后AI关键词的文献
            'ai_keywords_only': [],    # ID+DE中包含筛选后AI关键词的文献
            
            # 2. 传统AI关键词文献
            'traditional_full_text': [],    # 标题+摘要+ID+DE中包含传统AI关键词的文献
            'traditional_keywords_only': [], # ID+DE中包含传统AI关键词的文献
            
            # 3. 先进传统AI文献（既包含传统关键词又包含筛选后AI关键词）
            'advanced_traditional_full_text': [],    # 先进的传统AI（全文搜索）
            'advanced_traditional_keywords_only': [], # 先进的传统AI（仅关键词）
            
            # 4. 频次统计（分别统计全文和关键词搜索）
            'ai_keyword_frequency_full_text': Counter(),  # AI关键词频次（全文搜索）
            'ai_keyword_frequency_keywords_only': Counter(),  # AI关键词频次（仅关键词搜索）
            'traditional_keyword_frequency_full_text': Counter(),  # 传统AI关键词频次（全文搜索）
            'traditional_keyword_frequency_keywords_only': Counter()  # 传统AI关键词频次（仅关键词搜索）
        }
        
        # 修正：分析self.articles而不是self.es_articles
        for article in self.articles:
            # 准备搜索文本
            title_abstract = (article.get('title', '') + ' ' + article.get('abstract', '')).lower()
            keywords_text = ' '.join(article['all_keywords']).lower()
            full_text = title_abstract + ' ' + keywords_text
            
            # 1. 搜索筛选后的AI关键词
            matched_ai_full = []
            matched_ai_keywords = []
            
            for ai_keyword in self.ai_keywords:
                ai_keyword_lower = ai_keyword.lower()
                
                # 全文搜索
                if ai_keyword_lower in full_text:
                    matched_ai_full.append(ai_keyword)
                    ai_results['ai_keyword_frequency_full_text'][ai_keyword] += 1
                
                # 仅关键词搜索
                if ai_keyword_lower in keywords_text:
                    matched_ai_keywords.append(ai_keyword)
                    ai_results['ai_keyword_frequency_keywords_only'][ai_keyword] += 1
            
            # 记录包含筛选后AI关键词的文献
            if matched_ai_full:
                ai_results['ai_full_text'].append({
                    'article': article,
                    'matched_keywords': matched_ai_full
                })
            
            if matched_ai_keywords:
                ai_results['ai_keywords_only'].append({
                    'article': article,
                    'matched_keywords': matched_ai_keywords
                })
            
            # 2. 搜索传统AI关键词
            matched_traditional_full = []
            matched_traditional_keywords = []
            
            for trad_keyword in self.traditional_ai_keywords:
                trad_keyword_lower = trad_keyword.lower()
                
                # 全文搜索
                if trad_keyword_lower in full_text:
                    matched_traditional_full.append(trad_keyword)
                    ai_results['traditional_keyword_frequency_full_text'][trad_keyword] += 1
                
                # 仅关键词搜索
                if trad_keyword_lower in keywords_text:
                    matched_traditional_keywords.append(trad_keyword)
                    ai_results['traditional_keyword_frequency_keywords_only'][trad_keyword] += 1
            
            # 记录包含传统AI关键词的文献
            if matched_traditional_full:
                traditional_data_full = {
                    'article': article,
                    'matched_traditional': matched_traditional_full,
                    'is_advanced': bool(matched_ai_full),  # 是否同时包含筛选后AI关键词
                    'matched_ai': matched_ai_full
                }
                ai_results['traditional_full_text'].append(traditional_data_full)
                
                # 如果是先进的，也加入先进列表
                if traditional_data_full['is_advanced']:
                    ai_results['advanced_traditional_full_text'].append(traditional_data_full)
            
            if matched_traditional_keywords:
                traditional_data_keywords = {
                    'article': article,
                    'matched_traditional': matched_traditional_keywords,
                    'is_advanced': bool(matched_ai_keywords),  # 是否同时包含筛选后AI关键词
                    'matched_ai': matched_ai_keywords
                }
                ai_results['traditional_keywords_only'].append(traditional_data_keywords)
                
                # 如果是先进的，也加入先进列表
                if traditional_data_keywords['is_advanced']:
                    ai_results['advanced_traditional_keywords_only'].append(traditional_data_keywords)
        
        return ai_results

    def analyze_ai_by_cices(self, ai_results, classification_results):
        """按CICES类别分析AI使用情况"""
        ai_by_cices = {
            'Provisioning': {
                'ai_full_text': [], 'ai_keywords_only': [],
                'traditional_full_text': [], 'traditional_keywords_only': [],
                'advanced_traditional_full_text': [], 'advanced_traditional_keywords_only': []
            },
            'Regulating': {
                'ai_full_text': [], 'ai_keywords_only': [],
                'traditional_full_text': [], 'traditional_keywords_only': [],
                'advanced_traditional_full_text': [], 'advanced_traditional_keywords_only': []
            },
            'Cultural': {
                'ai_full_text': [], 'ai_keywords_only': [],
                'traditional_full_text': [], 'traditional_keywords_only': [],
                'advanced_traditional_full_text': [], 'advanced_traditional_keywords_only': []
            }
        }
        
        # 建立文章到CICES类别的映射
        article_to_categories = {}
        for category in ['Provisioning', 'Regulating', 'Cultural']:
            for article in classification_results[category]:
                article_id = id(article)
                if article_id not in article_to_categories:
                    article_to_categories[article_id] = []
                article_to_categories[article_id].append(category)
        
        # 分类各类AI文献
        for result_type in ['ai_full_text', 'ai_keywords_only', 'traditional_full_text', 
                           'traditional_keywords_only', 'advanced_traditional_full_text', 
                           'advanced_traditional_keywords_only']:
            for article_data in ai_results[result_type]:
                article = article_data['article']
                article_id = id(article)
                for category in article_to_categories.get(article_id, []):
                    ai_by_cices[category][result_type].append(article_data)
        
        return ai_by_cices

    def analyze_yearly_ai_trends(self, ai_results):
        """分析各年份AI技术使用趋势和先进率变化 - 修正：基于所有有关键词的文献"""
        yearly_ai_stats = defaultdict(lambda: {
            'total_articles': 0,
            'ai_full_text': 0,
            'ai_keywords_only': 0,
            'traditional_full_text': 0,
            'traditional_keywords_only': 0,
            'advanced_traditional_full_text': 0,
            'advanced_traditional_keywords_only': 0
        })
        
        # 统计各年份文献总数 - 修正：使用self.articles
        for article in self.articles:
            year = article.get('year', 'Unknown')
            yearly_ai_stats[year]['total_articles'] += 1
        
        # 统计各年份AI文献数
        for article_data in ai_results['ai_full_text']:
            year = article_data['article'].get('year', 'Unknown')
            yearly_ai_stats[year]['ai_full_text'] += 1
            
        for article_data in ai_results['ai_keywords_only']:
            year = article_data['article'].get('year', 'Unknown')
            yearly_ai_stats[year]['ai_keywords_only'] += 1
        
        # 统计各年份传统AI文献数
        for article_data in ai_results['traditional_full_text']:
            year = article_data['article'].get('year', 'Unknown')
            yearly_ai_stats[year]['traditional_full_text'] += 1
            
        for article_data in ai_results['traditional_keywords_only']:
            year = article_data['article'].get('year', 'Unknown')
            yearly_ai_stats[year]['traditional_keywords_only'] += 1
        
        # 统计各年份先进传统AI文献数
        for article_data in ai_results['advanced_traditional_full_text']:
            year = article_data['article'].get('year', 'Unknown')
            yearly_ai_stats[year]['advanced_traditional_full_text'] += 1
            
        for article_data in ai_results['advanced_traditional_keywords_only']:
            year = article_data['article'].get('year', 'Unknown')
            yearly_ai_stats[year]['advanced_traditional_keywords_only'] += 1
        
        return yearly_ai_stats

    def generate_ai_report(self, ai_results, ai_by_cices, main_categories_count):
        """生成AI分析报告 - 修正：基于所有有关键词的文献"""
        print("\n" + "="*80)
        print("🤖 Ecosystem Services Literature AI Technology Analysis Report")
        print("="*80)
        
        # 1. 总体AI使用统计 - 修正：基于self.articles
        print(f"\n📊 I. Overall AI Technology Usage Statistics")
        print(f"  ├─ Total Articles with Keywords: {len(self.articles)}")
        print(f"  ├─ AI Articles (Full Text Search): {len(ai_results['ai_full_text'])}")
        print(f"  ├─ AI Articles (Keywords Only): {len(ai_results['ai_keywords_only'])}")
        print(f"  ├─ AI Usage Rate (Full Text): {len(ai_results['ai_full_text'])/len(self.articles)*100:.2f}%")
        print(f"  └─ AI Usage Rate (Keywords Only): {len(ai_results['ai_keywords_only'])/len(self.articles)*100:.2f}%")
        
        # 2. 传统AI关键词先进性分析
        print(f"\n🧠 II. Traditional AI Keywords Advancement Analysis")
        print(f"  📍 Full Text Search Results：")
        print(f"    ├─ Traditional AI Articles: {len(ai_results['traditional_full_text'])}")
        print(f"    ├─ Advanced Traditional Articles: {len(ai_results['advanced_traditional_full_text'])}")
        if len(ai_results['traditional_full_text']) > 0:
            advanced_ratio = len(ai_results['advanced_traditional_full_text'])/len(ai_results['traditional_full_text'])*100
            print(f"    └─ Advancement Rate: {advanced_ratio:.2f}%")
        else:
            print(f"    └─ Advancement Rate: 0.00%")
            
        print(f"  📍 Keywords Only Search Results：")
        print(f"    ├─ Traditional AI Articles: {len(ai_results['traditional_keywords_only'])}")
        print(f"    ├─ Advanced Traditional Articles: {len(ai_results['advanced_traditional_keywords_only'])}")
        if len(ai_results['traditional_keywords_only']) > 0:
            advanced_ratio_kw = len(ai_results['advanced_traditional_keywords_only'])/len(ai_results['traditional_keywords_only'])*100
            print(f"    └─ Advancement Rate: {advanced_ratio_kw:.2f}%")
        else:
            print(f"    └─ Advancement Rate: 0.00%")
        
        # 3. 各CICES类别的AI使用情况
        print(f"\n🌳 III. AI Technology Usage by CICES Categories")
        print(f"  📍 Full Text Search Results：")
        print(f"  {'Category':<12} {'Total':<8} {'AI':<8} {'Trad AI':<8} {'Advanced':<10} {'AI %':<8} {'Adv %':<8}")
        print("  " + "-" * 70)
        
        for category in ['Provisioning', 'Regulating', 'Cultural']:
            total_docs = main_categories_count.get(category, 0)
            ai_full = len(ai_by_cices[category]['ai_full_text'])
            traditional_full = len(ai_by_cices[category]['traditional_full_text'])
            advanced_full = len(ai_by_cices[category]['advanced_traditional_full_text'])
            ai_rate = (ai_full / total_docs * 100) if total_docs > 0 else 0
            advanced_rate = (advanced_full / traditional_full * 100) if traditional_full > 0 else 0
            
            print(f"  {category:<12} {total_docs:<8} {ai_full:<8} {traditional_full:<8} {advanced_full:<10} {ai_rate:<8.2f} {advanced_rate:<8.2f}")
        
        print(f"\n  📍 Keywords Only Search Results：")
        print(f"  {'Category':<12} {'Total':<8} {'AI':<8} {'Trad AI':<8} {'Advanced':<10} {'AI %':<8} {'Adv %':<8}")
        print("  " + "-" * 70)
        
        for category in ['Provisioning', 'Regulating', 'Cultural']:
            total_docs = main_categories_count.get(category, 0)
            ai_keywords = len(ai_by_cices[category]['ai_keywords_only'])
            traditional_keywords = len(ai_by_cices[category]['traditional_keywords_only'])
            advanced_keywords = len(ai_by_cices[category]['advanced_traditional_keywords_only'])
            ai_rate = (ai_keywords / total_docs * 100) if total_docs > 0 else 0
            advanced_rate = (advanced_keywords / traditional_keywords * 100) if traditional_keywords > 0 else 0
            
            print(f"  {category:<12} {total_docs:<8} {ai_keywords:<8} {traditional_keywords:<8} {advanced_keywords:<10} {ai_rate:<8.2f} {advanced_rate:<8.2f}")
        
        # 4. 高频AI技术统计
        print(f"\n🔧 IV. High-Frequency AI Technologies (Top 20)")
        print(f"  📍 Full Text Search Results：")
        print(f"  {'Rank':<6} {'AI Technology':<45} {'Frequency':<10} {'Usage %':<10}")
        print("  " + "-" * 75)
        
        for i, (ai_tech, count) in enumerate(ai_results['ai_keyword_frequency_full_text'].most_common(20), 1):
            usage_rate = (count / len(self.articles)) * 100  # 修正：基于self.articles
            print(f"  {i:<6} {ai_tech[:43]:<45} {count:<10} {usage_rate:<10.2f}")
        
        print(f"\n  📍 Keywords Only Search Results：")
        print(f"  {'Rank':<6} {'AI Technology':<45} {'Frequency':<10} {'Usage %':<10}")
        print("  " + "-" * 75)
        
        for i, (ai_tech, count) in enumerate(ai_results['ai_keyword_frequency_keywords_only'].most_common(20), 1):
            usage_rate = (count / len(self.articles)) * 100  # 修正：基于self.articles
            print(f"  {i:<6} {ai_tech[:43]:<45} {count:<10} {usage_rate:<10.2f}")
        
        # 5. 传统AI关键词使用统计
        print(f"\n📊 V. Traditional AI Keywords Usage Statistics")
        print(f"  📍 Full Text Search Results：")
        print(f"  {'Keyword':<15} {'Total':<8} {'Advanced':<10} {'Adv %':<10}")
        print("  " + "-" * 50)
        
        for trad_keyword in self.traditional_ai_keywords:
            total_usage = ai_results['traditional_keyword_frequency_full_text'].get(trad_keyword, 0)
            advanced_usage_full = sum(1 for article_data in ai_results['advanced_traditional_full_text'] 
                                     if trad_keyword in article_data['matched_traditional'])
            
            advanced_rate = (advanced_usage_full / total_usage * 100) if total_usage > 0 else 0
            print(f"  {trad_keyword:<15} {total_usage:<8} {advanced_usage_full:<10} {advanced_rate:<10.2f}")
        
        print(f"\n  📍 Keywords Only Search Results：")
        print(f"  {'Keyword':<15} {'Total':<8} {'Advanced':<10} {'Adv %':<10}")
        print("  " + "-" * 50)
        
        for trad_keyword in self.traditional_ai_keywords:
            total_usage_kw = ai_results['traditional_keyword_frequency_keywords_only'].get(trad_keyword, 0)
            advanced_usage_kw = sum(1 for article_data in ai_results['advanced_traditional_keywords_only'] 
                                   if trad_keyword in article_data['matched_traditional'])
            
            advanced_rate_kw = (advanced_usage_kw / total_usage_kw * 100) if total_usage_kw > 0 else 0
            print(f"  {trad_keyword:<15} {total_usage_kw:<8} {advanced_usage_kw:<10} {advanced_rate_kw:<10.2f}")
        
        # 6. 年度AI先进率变化趋势
        yearly_stats = self.analyze_yearly_ai_trends(ai_results)
        
        print(f"\n📈 VI. Annual AI Advancement Rate Trends")
        
        # 筛选有效年份并排序
        valid_years = sorted([year for year in yearly_stats.keys() if year.isdigit() and year != 'Unknown'])
        
        if valid_years:
            print(f"  📍 Full Text Search Results：")
            print(f"  {'Year':<6} {'Total':<8} {'AI':<8} {'Trad AI':<8} {'Advanced':<10} {'AI %':<8} {'Adv %':<10}")
            print("  " + "-" * 75)
            
            for year in valid_years:
                stats = yearly_stats[year]
                total = stats['total_articles']
                ai_full = stats['ai_full_text']
                trad_full = stats['traditional_full_text']
                adv_full = stats['advanced_traditional_full_text']
                
                ai_rate = (ai_full / total * 100) if total > 0 else 0
                advanced_rate = (adv_full / trad_full * 100) if trad_full > 0 else 0
                
                print(f"  {year:<6} {total:<8} {ai_full:<8} {trad_full:<8} {adv_full:<10} {ai_rate:<8.2f} {advanced_rate:<10.2f}")
            
            print(f"\n  📍 Keywords Only Search Results：")
            print(f"  {'Year':<6} {'Total':<8} {'AI':<8} {'Trad AI':<8} {'Advanced':<10} {'AI %':<8} {'Adv %':<10}")
            print("  " + "-" * 75)
            
            for year in valid_years:
                stats = yearly_stats[year]
                total = stats['total_articles']
                ai_kw = stats['ai_keywords_only']
                trad_kw = stats['traditional_keywords_only']
                adv_kw = stats['advanced_traditional_keywords_only']
                
                ai_rate = (ai_kw / total * 100) if total > 0 else 0
                advanced_rate = (adv_kw / trad_kw * 100) if trad_kw > 0 else 0
                
                print(f"  {year:<6} {total:<8} {ai_kw:<8} {trad_kw:<8} {adv_kw:<10} {ai_rate:<8.2f} {advanced_rate:<10.2f}")
        else:
            print("  No valid year data available")
        
        return yearly_stats

    def export_ai_results(self, ai_results, ai_by_cices, main_categories_count, yearly_stats):
        """导出AI分析结果到Excel - 修正：基于所有有关键词的文献"""
        with pd.ExcelWriter('./ecosystem_services_ai_analysis.xlsx', engine='openpyxl') as writer:
            
            # 1. 总体AI使用统计 - 修正：基于self.articles
            summary_data = {
                'Metric': [
                    'Total Articles with Keywords', 
                    'AI Articles (Full Text Search)', 'AI Articles (Keywords Only)',
                    'AI Usage Rate (Full Text) %', 'AI Usage Rate (Keywords Only) %',
                    'Traditional AI Articles (Full Text)', 'Traditional AI Articles (Keywords Only)',
                    'Advanced Traditional AI Articles (Full Text)', 'Advanced Traditional AI Articles (Keywords Only)',
                    'Traditional AI Advancement Rate (Full Text) %', 'Traditional AI Advancement Rate (Keywords Only) %'
                ],
                'Value': [
                    len(self.articles),  # 修正：使用self.articles
                    len(ai_results['ai_full_text']), len(ai_results['ai_keywords_only']),
                    round(len(ai_results['ai_full_text'])/len(self.articles)*100, 2),
                    round(len(ai_results['ai_keywords_only'])/len(self.articles)*100, 2),
                    len(ai_results['traditional_full_text']), len(ai_results['traditional_keywords_only']),
                    len(ai_results['advanced_traditional_full_text']), len(ai_results['advanced_traditional_keywords_only']),
                    round(len(ai_results['advanced_traditional_full_text'])/len(ai_results['traditional_full_text'])*100, 2) if len(ai_results['traditional_full_text']) > 0 else 0,
                    round(len(ai_results['advanced_traditional_keywords_only'])/len(ai_results['traditional_keywords_only'])*100, 2) if len(ai_results['traditional_keywords_only']) > 0 else 0
                ]
            }
            summary_df = pd.DataFrame(summary_data)
            summary_df.to_excel(writer, sheet_name='Overall_Statistics', index=False)
            
            # 2. CICES类别AI使用统计
            cices_data = []
            for category in ['Provisioning', 'Regulating', 'Cultural']:
                total_docs = main_categories_count.get(category, 0)
                # 全文搜索结果
                cices_data.append({
                    'Category': f"{category}_Full_Text",
                    'Total_Articles': total_docs,
                    'AI_Articles': len(ai_by_cices[category]['ai_full_text']),
                    'Traditional_AI_Articles': len(ai_by_cices[category]['traditional_full_text']),
                    'Advanced_Traditional_Articles': len(ai_by_cices[category]['advanced_traditional_full_text']),
                    'AI_Rate_%': round(len(ai_by_cices[category]['ai_full_text'])/total_docs*100, 2) if total_docs > 0 else 0,
                    'Advanced_Rate_%': round(len(ai_by_cices[category]['advanced_traditional_full_text'])/len(ai_by_cices[category]['traditional_full_text'])*100, 2) if len(ai_by_cices[category]['traditional_full_text']) > 0 else 0
                })
                # 关键词搜索结果
                cices_data.append({
                    'Category': f"{category}_Keywords_Only",
                    'Total_Articles': total_docs,
                    'AI_Articles': len(ai_by_cices[category]['ai_keywords_only']),
                    'Traditional_AI_Articles': len(ai_by_cices[category]['traditional_keywords_only']),
                    'Advanced_Traditional_Articles': len(ai_by_cices[category]['advanced_traditional_keywords_only']),
                    'AI_Rate_%': round(len(ai_by_cices[category]['ai_keywords_only'])/total_docs*100, 2) if total_docs > 0 else 0,
                    'Advanced_Rate_%': round(len(ai_by_cices[category]['advanced_traditional_keywords_only'])/len(ai_by_cices[category]['traditional_keywords_only'])*100, 2) if len(ai_by_cices[category]['traditional_keywords_only']) > 0 else 0
                })
            
            cices_df = pd.DataFrame(cices_data)
            cices_df.to_excel(writer, sheet_name='CICES_AI_Usage', index=False)
            
            # 3. AI技术频次统计
            ai_tech_data_full = []
            for ai_tech, count in ai_results['ai_keyword_frequency_full_text'].most_common():
                ai_tech_data_full.append({
                    'AI_Technology': ai_tech,
                    'Frequency_Full_Text': count,
                    'Usage_Rate_Full_Text_%': round((count/len(self.articles))*100, 2)  # 修正：使用self.articles
                })
            ai_tech_full_df = pd.DataFrame(ai_tech_data_full)
            ai_tech_full_df.to_excel(writer, sheet_name='AI_Tech_Frequency_Full_Text', index=False)
            
            ai_tech_data_keywords = []
            for ai_tech, count in ai_results['ai_keyword_frequency_keywords_only'].most_common():
                ai_tech_data_keywords.append({
                    'AI_Technology': ai_tech,
                    'Frequency_Keywords_Only': count,
                    'Usage_Rate_Keywords_%': round((count/len(self.articles))*100, 2)  # 修正：使用self.articles
                })
            ai_tech_keywords_df = pd.DataFrame(ai_tech_data_keywords)
            ai_tech_keywords_df.to_excel(writer, sheet_name='AI_Tech_Frequency_Keywords', index=False)
            
            # 4. 传统AI关键词先进性分析
            traditional_data = []
            for trad_keyword in self.traditional_ai_keywords:
                # 全文搜索结果
                total_usage_full = ai_results['traditional_keyword_frequency_full_text'].get(trad_keyword, 0)
                advanced_usage_full = sum(1 for article_data in ai_results['advanced_traditional_full_text'] 
                                         if trad_keyword in article_data['matched_traditional'])
                
                # 关键词搜索结果
                total_usage_kw = ai_results['traditional_keyword_frequency_keywords_only'].get(trad_keyword, 0)
                advanced_usage_kw = sum(1 for article_data in ai_results['advanced_traditional_keywords_only'] 
                                       if trad_keyword in article_data['matched_traditional'])
                
                traditional_data.append({
                    'Traditional_Keyword': trad_keyword,
                    'Total_Usage_Full_Text': total_usage_full,
                    'Advanced_Usage_Full_Text': advanced_usage_full,
                    'Advanced_Rate_Full_Text_%': round((advanced_usage_full / total_usage_full * 100), 2) if total_usage_full > 0 else 0,
                    'Total_Usage_Keywords_Only': total_usage_kw,
                    'Advanced_Usage_Keywords_Only': advanced_usage_kw,
                    'Advanced_Rate_Keywords_%': round((advanced_usage_kw / total_usage_kw * 100), 2) if total_usage_kw > 0 else 0
                })
            
            traditional_df = pd.DataFrame(traditional_data)
            traditional_df.to_excel(writer, sheet_name='Traditional_AI_Advancement', index=False)
            
            # 5. 年度AI使用趋势
            valid_years = sorted([year for year in yearly_stats.keys() if year.isdigit() and year != 'Unknown'])
            if valid_years:
                yearly_data = []
                for year in valid_years:
                    stats = yearly_stats[year]
                    yearly_data.append({
                        'Year': year,
                        'Total_Articles': stats['total_articles'],
                        'AI_Full_Text': stats['ai_full_text'],
                        'AI_Keywords_Only': stats['ai_keywords_only'],
                        'Traditional_Full_Text': stats['traditional_full_text'],
                        'Traditional_Keywords_Only': stats['traditional_keywords_only'],
                        'Advanced_Traditional_Full_Text': stats['advanced_traditional_full_text'],
                        'Advanced_Traditional_Keywords_Only': stats['advanced_traditional_keywords_only'],
                        'AI_Rate_Full_Text_%': round((stats['ai_full_text'] / stats['total_articles'] * 100), 2) if stats['total_articles'] > 0 else 0,
                        'AI_Rate_Keywords_%': round((stats['ai_keywords_only'] / stats['total_articles'] * 100), 2) if stats['total_articles'] > 0 else 0,
                        'Advanced_Rate_Full_Text_%': round((stats['advanced_traditional_full_text'] / stats['traditional_full_text'] * 100), 2) if stats['traditional_full_text'] > 0 else 0,
                        'Advanced_Rate_Keywords_%': round((stats['advanced_traditional_keywords_only'] / stats['traditional_keywords_only'] * 100), 2) if stats['traditional_keywords_only'] > 0 else 0
                    })
                
                yearly_df = pd.DataFrame(yearly_data)
                yearly_df.to_excel(writer, sheet_name='Annual_AI_Trends', index=False)
        
        print("\n✅ AI analysis results exported to ecosystem_services_ai_analysis.xlsx")

    def run_ai_analysis(self):
        """运行AI专项分析 - 修正：移除不必要的筛选步骤"""
        print("🤖 Starting Ecosystem Services Literature AI Technology Analysis...")
        print("="*80)
        
        # 步骤1: 读取文献
        self.read_literature_files()
        
        if not self.all_articles:
            print("❌ No valid articles found!")
            return
        
        if not self.articles:
            print("❌ No articles with keywords found!")
            return
        
        print(f"\n📊 WOS search already ensures all articles are ecosystem service related")
        print(f"📊 Analyzing {len(self.articles)} articles with keywords")
        
        # 步骤2: CICES分类 - 直接对所有有关键词的文献进行分类
        print("📊 Performing CICES classification...")
        classification_results = self.classify_articles_by_cices()
        
        main_categories_count = {}
        for category in ['Provisioning', 'Regulating', 'Cultural']:
            main_categories_count[category] = len(classification_results[category])
        
        # 步骤3: AI技术使用分析
        ai_results = self.analyze_ai_usage()
        
        # 步骤4: 按CICES分类分析AI使用
        print("📊 Analyzing AI usage by CICES categories...")
        ai_by_cices = self.analyze_ai_by_cices(ai_results, classification_results)
        
        # 步骤5: 生成报告（包含年度趋势分析）
        yearly_stats = self.generate_ai_report(ai_results, ai_by_cices, main_categories_count)
        
        # 步骤6: 导出结果
        print("💾 Exporting analysis results...")
        self.export_ai_results(ai_results, ai_by_cices, main_categories_count, yearly_stats)
        
        print("\n" + "="*80)
        print("✅ AI analysis completed! Generated files:")
        print("  📁 ecosystem_services_ai_analysis.xlsx")
        print("="*80)

# 运行分析
if __name__ == "__main__":
    analyzer = EcosystemServicesAIAnalyzer()
    analyzer.run_ai_analysis()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Set Times New Roman font and larger font sizes
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['font.size'] = 14  # Base font size increased

class StackedAIChart:
    def __init__(self):
        # Color scheme based on SCI paper colors
        self.colors = {
            'advanced_ai': '#CE2227',    # Deep red R:206 G:34 B:39
            'traditional': '#F47B5D',    # Orange-red R:244 G:123 B:93  
            'other_ai': '#E9A29E',       # Light red R:233 G:162 B:158
            'reference': '#6096B8'       # Blue reference line R:96 G:150 B:184
        }
        
        # Data 1: Keywords search (ID+DE only) - Corrected calculations
        self.id_de_data = {
            'years': [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025],
            'advanced_ai': [7, 8, 14, 15, 17, 28, 32, 35, 25, 38, 23],  # Advanced Traditional
            'traditional': [67, 76, 77, 97, 107, 127, 156, 154, 159, 150, 117],  # Traditional - Advanced  
            'other_ai': [182, 248, 234, 268, 333, 349, 469, 489, 468, 535, 395],  # AI Literature 
            'total_data_science': [256, 332, 325, 380, 457, 504, 657, 678, 652, 723, 535]  # AI Literature + Traditional AI
        }
        
        # Data 2: Full text search (Title+Abstract+ID+DE) - Corrected calculations  
        self.full_text_data = {
            'years': [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025],
            'advanced_ai': [113, 126, 150, 158, 193, 224, 303, 286, 313, 365, 301],  # Advanced Traditional
            'traditional': [70, 66, 80, 111, 114, 135, 158, 203, 209, 232, 152],  # Traditional - Advanced
            'other_ai': [1452, 1690, 1714, 1983, 2277, 2584, 3183, 3203, 3101, 3509, 2628],  # AI Literature
            'total_data_science': [1635, 1882, 1944, 2252, 2584, 2943, 3644, 3692, 3623, 4106, 3081]  # AI Literature + Traditional AI
        }
    
    def create_stacked_charts(self):
        """Create double stacked bar charts"""
        fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(16, 14))
        
        # Chart 1: Keywords search only (ID+DE)
        years = self.id_de_data['years']
        x_pos = np.arange(len(years))
        
        # Stack data
        advanced = self.id_de_data['advanced_ai']
        traditional = self.id_de_data['traditional']
        other = self.id_de_data['other_ai']
        total = self.id_de_data['total_data_science']
        
        # Create stacked bar chart
        p1 = ax1.bar(x_pos, advanced, color=self.colors['advanced_ai'], 
                     label='Both AI and Tradition', alpha=0.9)
        p2 = ax1.bar(x_pos, traditional, bottom=advanced, 
                     color=self.colors['traditional'], label='Only Tradition', alpha=0.9)
        p3 = ax1.bar(x_pos, other, bottom=np.array(advanced) + np.array(traditional),
                     color=self.colors['other_ai'], label='Only AI', alpha=0.9)
        
        # Add Data Science Methods reference line (折线图)
        ax1.plot(x_pos, total, color=self.colors['reference'], linewidth=4, 
                marker='o', markersize=8, label='Total Literature', alpha=0.8, zorder=10)
        
        # Set first chart
        ax1.set_title('Annual AI Application Trends (Keywords Search Only)', fontsize=18, fontweight='bold', pad=25)
        ax1.set_xlabel('Year', fontsize=16, fontweight='bold')
        ax1.set_ylabel('Number of Publications', fontsize=16, fontweight='bold')
        ax1.set_xticks(x_pos)
        ax1.set_xticklabels(years, rotation=45, fontsize=14)
        ax1.legend(loc='upper left', fontsize=14, frameon=True, fancybox=True, shadow=True)
        ax1.grid(True, alpha=0.3, axis='y')
        ax1.set_facecolor('#fafafa')
        ax1.tick_params(axis='both', which='major', labelsize=14)
        
        # Add value labels to 折线图 (关键修改：将fontsize从12改为16，可根据需求调整)
        for i, (year, tot) in enumerate(zip(years, total)):
            ax1.annotate(f'{tot}', (i, tot), textcoords="offset points", 
                       xytext=(0,10), ha='center', fontsize=16, fontweight='bold',  # fontsize加大
                       color=self.colors['reference'])
        
        # Chart 2: Full text search (Title+Abstract+ID+DE)
        years2 = self.full_text_data['years']
        x_pos2 = np.arange(len(years2))
        
        advanced2 = self.full_text_data['advanced_ai']
        traditional2 = self.full_text_data['traditional']
        other2 = self.full_text_data['other_ai']
        total2 = self.full_text_data['total_data_science']
        
        # Create second stacked bar chart
        p4 = ax2.bar(x_pos2, advanced2, color=self.colors['advanced_ai'], 
                     label='Both AI and Tradition', alpha=0.9)
        p5 = ax2.bar(x_pos2, traditional2, bottom=advanced2, 
                     color=self.colors['traditional'], label='Only Tradition', alpha=0.9)
        p6 = ax2.bar(x_pos2, other2, bottom=np.array(advanced2) + np.array(traditional2),
                     color=self.colors['other_ai'], label='Only AI', alpha=0.9)
        
        # Add Data Science Methods reference line (折线图)
        ax2.plot(x_pos2, total2, color=self.colors['reference'], linewidth=4, 
                marker='o', markersize=8, label='Total Literature', alpha=0.8, zorder=10)
        
        # Set second chart
        ax2.set_title('Annual AI Application Trends (Full Text Search)', fontsize=18, fontweight='bold', pad=25)
        ax2.set_xlabel('Year', fontsize=16, fontweight='bold')
        ax2.set_ylabel('Number of Publications', fontsize=16, fontweight='bold')
        ax2.set_xticks(x_pos2)
        ax2.set_xticklabels(years2, rotation=45, fontsize=14)
        ax2.legend(loc='upper left', fontsize=14, frameon=True, fancybox=True, shadow=True)
        ax2.grid(True, alpha=0.3, axis='y')
        ax2.set_facecolor('#fafafa')
        ax2.tick_params(axis='both', which='major', labelsize=14)
        
        # Add value labels to 折线图 (关键修改：将fontsize从12改为16)
        for i, (year, tot) in enumerate(zip(years2, total2)):
            ax2.annotate(f'{tot}', (i, tot), textcoords="offset points", 
                       xytext=(0,10), ha='center', fontsize=16, fontweight='bold',  # fontsize加大
                       color=self.colors['reference'])
        
        plt.tight_layout()
        plt.savefig('./ai_stacked_chart.png', dpi=300, bbox_inches='tight', 
                   facecolor='white', edgecolor='none')
        plt.show()
        
        print("Stacked bar chart generated: ai_stacked_chart.png")
    
    def create_percentage_stacked_chart(self):
        """Create percentage stacked charts"""
        fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(16, 14))
        
        # Chart 1: Keywords search - percentage stacked
        years = self.id_de_data['years']
        x_pos = np.arange(len(years))
        
        advanced = np.array(self.id_de_data['advanced_ai'])
        traditional = np.array(self.id_de_data['traditional'])
        other = np.array(self.id_de_data['other_ai'])
        total = np.array(self.id_de_data['total_data_science'])
        
        # Calculate percentages
        adv_pct = (advanced / total) * 100
        trad_pct = (traditional / total) * 100
        other_pct = (other / total) * 100
        
        # Create percentage stacked chart
        ax1.bar(x_pos, adv_pct, color=self.colors['advanced_ai'], 
               label='Both AI and Tradition', alpha=0.9)
        ax1.bar(x_pos, trad_pct, bottom=adv_pct, 
               color=self.colors['traditional'], label='Only Tradition', alpha=0.9)
        ax1.bar(x_pos, other_pct, bottom=adv_pct + trad_pct,
               color=self.colors['other_ai'], label='Only AI', alpha=0.9)
        
        ax1.set_title('AI Application Type Distribution (Keywords Search)', fontsize=18, fontweight='bold', pad=25)
        ax1.set_xlabel('Year', fontsize=16, fontweight='bold')
        ax1.set_ylabel('Percentage (%)', fontsize=16, fontweight='bold')
        ax1.set_xticks(x_pos)
        ax1.set_xticklabels(years, rotation=45, fontsize=14)
        ax1.legend(loc='upper right', fontsize=14)
        ax1.set_ylim(0, 100)
        ax1.grid(True, alpha=0.3, axis='y')
        ax1.set_facecolor('#fafafa')
        ax1.tick_params(axis='both', which='major', labelsize=14)
        
        # Chart 2: Full text search - percentage stacked
        years2 = self.full_text_data['years']
        x_pos2 = np.arange(len(years2))
        
        advanced2 = np.array(self.full_text_data['advanced_ai'])
        traditional2 = np.array(self.full_text_data['traditional'])
        other2 = np.array(self.full_text_data['other_ai'])
        total2 = np.array(self.full_text_data['total_data_science'])
        
        # Calculate percentages
        adv_pct2 = (advanced2 / total2) * 100
        trad_pct2 = (traditional2 / total2) * 100
        other_pct2 = (other2 / total2) * 100
        
        # Create second percentage stacked chart
        ax2.bar(x_pos2, adv_pct2, color=self.colors['advanced_ai'], 
               label='Both AI and Tradition', alpha=0.9)
        ax2.bar(x_pos2, trad_pct2, bottom=adv_pct2, 
               color=self.colors['traditional'], label='Only Tradition', alpha=0.9)
        ax2.bar(x_pos2, other_pct2, bottom=adv_pct2 + trad_pct2,
               color=self.colors['other_ai'], label='Only AI', alpha=0.9)
        
        ax2.set_title('AI Application Type Distribution (Full Text Search)', fontsize=18, fontweight='bold', pad=25)
        ax2.set_xlabel('Year', fontsize=16, fontweight='bold')
        ax2.set_ylabel('Percentage (%)', fontsize=16, fontweight='bold')
        ax2.set_xticks(x_pos2)
        ax2.set_xticklabels(years2, rotation=45, fontsize=14)
        ax2.legend(loc='upper right', fontsize=14)
        ax2.set_ylim(0, 100)
        ax2.grid(True, alpha=0.3, axis='y')
        ax2.set_facecolor('#fafafa')
        ax2.tick_params(axis='both', which='major', labelsize=14)
        
        plt.tight_layout()
        plt.savefig('./ai_percentage_stacked_chart.png', dpi=300, bbox_inches='tight', 
                   facecolor='white', edgecolor='none')
        plt.show()
        
        print("Percentage stacked chart generated: ai_percentage_stacked_chart.png")
    
    def create_combined_chart(self):
        """Create combined stacked chart comparison"""
        fig = plt.figure(figsize=(18, 12))
        
        # Create grid layout
        gs = fig.add_gridspec(2, 2, height_ratios=[1, 1], width_ratios=[1, 1], 
                             hspace=0.3, wspace=0.2)
        
        # Left side: Absolute number stacked chart
        ax1 = fig.add_subplot(gs[:, 0])
        
        # Data preparation
        years = self.id_de_data['years']
        x_pos = np.arange(len(years))
        width = 0.35
        
        advanced1 = self.id_de_data['advanced_ai']
        traditional1 = self.id_de_data['traditional']
        other1 = self.id_de_data['other_ai']
        
        advanced2 = self.full_text_data['advanced_ai']
        traditional2 = self.full_text_data['traditional']
        other2 = self.full_text_data['other_ai']
        
        # Left bars (Keywords search)
        ax1.bar(x_pos - width/2, advanced1, width, color=self.colors['advanced_ai'], 
               label='Both AI and Tradition (Keywords)', alpha=0.9)
        ax1.bar(x_pos - width/2, traditional1, width, bottom=advanced1,
               color=self.colors['traditional'], label='Only Tradition (Keywords)', alpha=0.9)
        ax1.bar(x_pos - width/2, other1, width, 
               bottom=np.array(advanced1) + np.array(traditional1),
               color=self.colors['other_ai'], label='Only AI (Keywords)', alpha=0.9)
        
        # Right bars (Full text search)
        ax1.bar(x_pos + width/2, advanced2, width, color=self.colors['advanced_ai'], 
               label='Both AI and Tradition (Full Text)', alpha=0.7, hatch='//')
        ax1.bar(x_pos + width/2, traditional2, width, bottom=advanced2,
               color=self.colors['traditional'], label='Only Tradition (Full Text)', alpha=0.7, hatch='//')
        ax1.bar(x_pos + width/2, other2, width,
               bottom=np.array(advanced2) + np.array(traditional2),
               color=self.colors['other_ai'], label='Only AI (Full Text)', alpha=0.7, hatch='//')
        
        ax1.set_title('AI Application Trends Comparison', fontsize=18, fontweight='bold', pad=25)
        ax1.set_xlabel('Year', fontsize=16, fontweight='bold')
        ax1.set_ylabel('Number of Publications', fontsize=16, fontweight='bold')
        ax1.set_xticks(x_pos)
        ax1.set_xticklabels(years, rotation=45, fontsize=14)
        ax1.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=12)
        ax1.grid(True, alpha=0.3, axis='y')
        ax1.set_facecolor('#fafafa')
        ax1.tick_params(axis='both', which='major', labelsize=14)
        
        # Upper right: Keywords search data distribution
        ax2 = fig.add_subplot(gs[0, 1])
        sizes1 = [sum(advanced1), sum(traditional1), sum(other1)]
        labels1 = ['Both AI and Tradition', 'Only Tradition', 'Only AI']
        colors1 = [self.colors['advanced_ai'], self.colors['traditional'], self.colors['other_ai']]
        
        ax2.pie(sizes1, labels=labels1, colors=colors1, autopct='%1.1f%%', 
               startangle=90, alpha=0.9, textprops={'fontsize': 12})
        ax2.set_title('Overall Distribution (Keywords Search)', fontsize=14, fontweight='bold')
        
        # Lower right: Full text search data distribution
        ax3 = fig.add_subplot(gs[1, 1])
        sizes2 = [sum(advanced2), sum(traditional2), sum(other2)]
        
        ax3.pie(sizes2, labels=labels1, colors=colors1, autopct='%1.1f%%', 
               startangle=90, alpha=0.9, textprops={'fontsize': 12})
        ax3.set_title('Overall Distribution (Full Text Search)', fontsize=14, fontweight='bold')
        
        # Add color scheme note
        fig.text(0.02, 0.02, 'Color scheme: SCI paper color palette', 
                fontsize=12, style='italic', alpha=0.7)
        
        plt.savefig('./ai_combined_stacked_chart.png', dpi=300, bbox_inches='tight', 
                   facecolor='white', edgecolor='none')
        plt.show()
        
        print("Combined stacked chart generated: ai_combined_stacked_chart.png")
    
    def run_all_charts(self):
        """Generate stacked charts"""
        print("Generating AI application trend stacked charts...")
        
        # Generate basic stacked charts only
        self.create_stacked_charts()
        
        print("\nStacked charts generated successfully!")
        print("Generated files:")
        print("  - ai_stacked_chart.png (Stacked bar charts)")

# Run chart generation
if __name__ == "__main__":
    chart_generator = StackedAIChart()
    chart_generator.run_all_charts()
```

```python
import os
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter, defaultdict

# 设置字体
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['axes.unicode_minus'] = False

class CICESAIAnalyzer:
    def __init__(self, data_directory="./"):
        self.data_directory = data_directory
        self.all_articles = []
        self.articles = []  # 所有有关键词的文献
        
        # AI关键词列表
        self.ai_keywords = [
            "machine learning", "supervised learning", "unsupervised learning", 
            "association rules", "dimensionality reduction", 
            "semi supervised learning", "self training", "low density separation models", 
            "graph based algorithms", "generative models", "reinforcement learning", 
            "deep learning", "deep feed forward networks", "convolution neural networks", 
            "siamese neural networks", "transformers", "generative adversarial networks", 
            "graph neural networks", "recurrent neural network", "active learning", 
            "uncertainty sampling", "diversity sampling", "transfer learning", 
            "instance-based transfer learning", "feature-based transfer learning", 
            "parameter-based transfer learning", "continual learning", 
            "regularization-based continual learning", "memory reply", "dynamic architectures", 
            "natural language processing", "machine translation", "question answering", 
            "sentiment analysis", "text generation", "topic modeling", "graphplan", 
            "satplan", "slam algorithms", "fastslam", "graphslam", "motion planning algorithms", 
            "sampling-based algorithms", "rrt", "graph search algorithms", "a-star", 
            "reaction-based algorithms", "dwa", "interpolating curve algorithms", 
            "image recognition", "machine vision", "nerf", "vision transformer", 
            "tacotron", "vits", "wavenet", "samplern", "expert systems", 
            "frame-based systems", "certainty factors", "RNN", "CNN", "LSTM", 
            "Long Short-Term Memory", "ML", "DL", "RL"
        ]
        
        # CICES分类关键词
        self.cices_categories = {
            'Provisioning': {
                'terrestrial_crops': [
                    'crops', 'cereals', 'root crops', 'soft fruit', 'fodder', 'livestock feed',
                    'fibres', 'materials', 'cultivated plants', 'fungi', 'algae', 'bacteria',
                    'plantation', 'energy source', 'bioenergy'
                ],
                'aquatic_crops': [
                    'aquaculture', 'in-situ aquaculture', 'cultivated aquatic plants'
                ],
                'reared_animals': [
                    'reared animals', 'beef', 'dairy', 'animal products',
                    'fibres from animals', 'mechanical energy', 'biogas'
                ],
                'aquatic_animals': [
                    'reared aquatic animals', 'in-situ aquaculture animals'
                ],
                'wild_plants': [
                    'wild plants', 'terrestrial plants', 'aquatic plants', 'wild fungi', 'wild algae',
                    'wild plant materials', 'wild plant energy'
                ],
                'wild_animals': [
                    'wild animals', 'terrestrial animals', 'aquatic animals',
                    'wild animal nutrition', 'wild animal materials'
                ],
                'genetic_material': [
                    'genetic material', 'seeds', 'spores', 'plant materials',
                    'breeding', 'new strains', 'varieties',
                    'individual genes', 'biological entities',
                    'animal genetic material', 'population establishment'
                ]
            },
            'Regulating': {
                'biochemical_transformation': [
                    'bio-remediation', 'micro-organisms', 'pollution degradation',
                    'filtration', 'sequestration', 'storage', 'accumulation',
                    'waste mediation', 'toxic substances', 'anthropogenic origin',
                    'smell reduction', 'noise attenuation', 'visual screening'
                ],
                'flow_extreme_events': [
                    'erosion control', 'water erosion', 'wind erosion',
                    'hydrological cycle', 'water flow regulation', 'runoff', 'base flows', 'peak flows',
                    'hazard mitigation', 'mass movement', 'landslides', 'flood mitigation', 'surge protection',
                    'wind protection', 'fire protection'
                ],
                'lifecycle_maintenance': [
                    'pollination', 'gamete dispersal', 'seed dispersal',
                    'nursery populations', 'breeding grounds', 'gene pool protection',
                    'refuge habitats', 'feeding grounds'
                ],
                'pest_disease_control': [
                    'pest control', 'invasive species', 'disease control'
                ],
                'soil_quality': [
                    'soil quality', 'weathering processes', 'decomposition', 'fixing processes',
                    'soil structure', 'biological agents', 'soil porosity'
                ],
                'water_conditions': [
                    'water conditions', 'chemical condition', 'macronutrients',
                    'freshwater regulation', 'saltwater regulation'
                ],
                'atmospheric_conditions': [
                    'atmospheric composition', 'oceanic circulation', 'climate change',
                    'temperature regulation', 'humidity', 'ventilation', 'transpiration', 'microclimate'
                ]
            },
            'Cultural': {
                'physical_experiential': [
                    'recreational activities', 'health promotion', 'recuperation', 'enjoyment',
                    'active interactions', 'immersive interactions',
                    'passive interactions', 'observational interactions'
                ],
                'intellectual_representative': [
                    'scientific investigation', 'traditional ecological knowledge', 'genetic diversity',
                    'education', 'training',
                    'culture', 'heritage', 'cultural resonance',
                    'aesthetic experiences'
                ],
                'indirect_interactions': [
                    'entertainment', 'representation outside setting'
                ],
                'spiritual_symbolic': [
                    'symbolic meaning', 'sense of place', 'distinctiveness',
                    'spiritual meaning', 'religious meaning',
                    'contemporary existence', 'conservation importance',
                    'inter-generational existence', 'genetic diversity conservation'
                ]
            }
        }
        
        # 构建CICES关键词列表
        self.all_cices_keywords = []
        self.cices_keyword_to_category = {}
        for main_category, subcategories in self.cices_categories.items():
            for subcategory, keywords in subcategories.items():
                for keyword in keywords:
                    self.all_cices_keywords.append(keyword.lower())
                    self.cices_keyword_to_category[keyword.lower()] = main_category

    def read_literature_files(self):
        """读取所有文献文件"""
        files = []
        for f in os.listdir(self.data_directory):
            if f.endswith('.txt') and 'download' in f and any(year in f for year in 
                ['2015', '2016', '2017', '2018', '2019', '2020', '2021', '2022', '2023', '2024', '2025']):
                files.append(f)
        
        files.sort()
        print(f"📚 找到 {len(files)} 个文献文件")
        
        for file in files:
            file_path = os.path.join(self.data_directory, file)
            try:
                with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
                    content = f.read()
                    self.parse_articles(content, file)
            except Exception as e:
                print(f"❌ 读取文件 {file} 时出错: {e}")
        
        print(f"✅ 总共解析了 {len(self.all_articles)} 篇文章")
        print(f"✅ 有关键词的文章: {len(self.articles)} 篇")

    def parse_articles(self, content, filename):
        """解析文章内容"""
        articles = re.split(r'\nER\s*\n', content)
        
        for article_text in articles:
            if not article_text.strip():
                continue
            
            if not re.search(r'(^|\n)PT\s+', article_text):
                continue
            
            article_data = {
                'filename': filename,
                'keywords': set(),
                'keywords_plus': set(),
                'title': '',
                'year': '',
                'abstract': '',
                'journal': '',
                'authors': []
            }
            
            # 提取ID字段
            id_match = re.search(r'\nID\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if id_match:
                id_content = id_match.group(1)
                keywords_plus = self.extract_keywords_from_field(id_content)
                article_data['keywords_plus'].update(keywords_plus)
            
            # 提取DE字段
            de_match = re.search(r'\nDE\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if de_match:
                de_content = de_match.group(1)
                keywords = self.extract_keywords_from_field(de_content)
                article_data['keywords'].update(keywords)
            
            # 提取标题
            ti_match = re.search(r'\nTI\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if ti_match:
                article_data['title'] = ti_match.group(1).replace('\n', ' ').strip()
            
            # 提取年份
            py_match = re.search(r'\nPY\s+(\d{4})', article_text)
            if py_match:
                article_data['year'] = py_match.group(1)
            
            # 提取摘要
            ab_match = re.search(r'\nAB\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if ab_match:
                article_data['abstract'] = ab_match.group(1).replace('\n', ' ').strip()
            
            # 提取期刊
            so_match = re.search(r'\nSO\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if so_match:
                article_data['journal'] = so_match.group(1).replace('\n', ' ').strip()
            
            # 提取作者
            au_match = re.search(r'\nAU\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if au_match:
                authors_text = au_match.group(1)
                authors = [author.strip() for author in authors_text.split('\n') if author.strip()]
                article_data['authors'] = authors
            
            self.all_articles.append(article_data)
            
            if article_data['keywords'] or article_data['keywords_plus']:
                article_data['all_keywords'] = article_data['keywords'] | article_data['keywords_plus']
                self.articles.append(article_data)

    def extract_keywords_from_field(self, field_content):
        """从ID或DE字段内容中提取关键词"""
        if not field_content:
            return []
            
        field_content = field_content.replace('\n', ' ')
        keywords = re.split(r'[;\n]', field_content)
        
        cleaned_keywords = []
        for keyword in keywords:
            keyword = keyword.strip().lower()
            keyword = re.sub(r'\s+', ' ', keyword)
            keyword = keyword.strip('.,;()[]{}"\'-')
            
            if keyword and len(keyword) > 2:
                cleaned_keywords.append(keyword)
                
        return cleaned_keywords

    def identify_cices_ai_articles(self):
        """识别同时包含CICES关键词和AI关键词的文献"""
        cices_ai_articles = []
        
        for article in self.articles:
            keywords_text = ' '.join(article['all_keywords']).lower()
            
            # 检查是否包含CICES关键词
            matched_cices = []
            for cices_keyword in self.all_cices_keywords:
                if cices_keyword in keywords_text:
                    matched_cices.append(cices_keyword)
            
            # 检查是否包含AI关键词
            matched_ai = []
            for ai_keyword in self.ai_keywords:
                if ai_keyword.lower() in keywords_text:
                    matched_ai.append(ai_keyword)
            
            # 只保留同时包含CICES和AI关键词的文献
            if matched_cices and matched_ai:
                article['matched_cices'] = matched_cices
                article['matched_ai'] = matched_ai
                cices_ai_articles.append(article)
        
        print(f"🎯 找到同时包含CICES和AI关键词的文献: {len(cices_ai_articles)} 篇")
        return cices_ai_articles

    def analyze_cices_ai_combinations(self, cices_ai_articles):
        """分析CICES关键词与AI关键词的组合"""
        # 统计共现矩阵
        cices_ai_cooccurrence = defaultdict(lambda: defaultdict(int))
        cices_counts = Counter()
        ai_counts = Counter()
        
        # 统计所有组合
        for article in cices_ai_articles:
            for cices_kw in article['matched_cices']:
                cices_counts[cices_kw] += 1
                for ai_kw in article['matched_ai']:
                    cices_ai_cooccurrence[cices_kw][ai_kw] += 1
            
            for ai_kw in article['matched_ai']:
                ai_counts[ai_kw] += 1
        
        return cices_ai_cooccurrence, cices_counts, ai_counts

    def calculate_association_metrics(self, cices_ai_cooccurrence, cices_counts, ai_counts, total_articles):
        """计算关联度指标"""
        # CICES→AI分析
        cices_to_ai_results = []
        for cices_kw in cices_counts:
            cices_category = self.cices_keyword_to_category.get(cices_kw, 'Unknown')
            ai_combinations = []
            
            for ai_kw, cooccurrence in cices_ai_cooccurrence[cices_kw].items():
                if cooccurrence > 0:
                    # 基于CICES-AI共现文献的计算
                    cices_freq_in_combination = cices_counts[cices_kw]  # CICES关键词在组合中的频次
                    ai_freq_in_combination = ai_counts[ai_kw]  # AI关键词在组合中的频次
                    
                    # 支持度：共现文献数/总的CICES-AI组合文献数
                    support = cooccurrence / total_articles
                    
                    # CICES→AI置信度：共现次数/CICES关键词在组合中的总频次
                    confidence_cices_to_ai = cooccurrence / cices_freq_in_combination
                    
                    # CICES→AI提升度
                    expected_ai_prob = ai_freq_in_combination / sum(ai_counts.values())
                    lift_cices_to_ai = confidence_cices_to_ai / expected_ai_prob if expected_ai_prob > 0 else 0
                    
                    ai_combinations.append({
                        'AI_Keyword': ai_kw,
                        'Cooccurrence': cooccurrence,
                        'Confidence_CICES_to_AI': round(confidence_cices_to_ai, 4),
                        'Lift_CICES_to_AI': round(lift_cices_to_ai, 4),
                        'Support': round(support, 4)
                    })
            
            # 按共现次数降序排序
            ai_combinations.sort(key=lambda x: x['Cooccurrence'], reverse=True)
            
            if ai_combinations:  # 只保留有AI组合的CICES关键词
                cices_to_ai_results.append({
                    'CICES_Keyword': cices_kw,
                    'CICES_Category': cices_category,
                    'CICES_Frequency': cices_freq_in_combination,
                    'AI_Combinations': ai_combinations
                })
        
        # AI→CICES分析
        ai_to_cices_results = []
        for ai_kw in ai_counts:
            cices_combinations = []
            
            for cices_kw in cices_counts:
                cooccurrence = cices_ai_cooccurrence[cices_kw][ai_kw]
                if cooccurrence > 0:
                    cices_category = self.cices_keyword_to_category.get(cices_kw, 'Unknown')
                    
                    # 基于CICES-AI共现文献的计算
                    ai_freq_in_combination = ai_counts[ai_kw]
                    cices_freq_in_combination = cices_counts[cices_kw]
                    
                    # 支持度
                    support = cooccurrence / total_articles
                    
                    # AI→CICES置信度：共现次数/AI关键词在组合中的总频次
                    confidence_ai_to_cices = cooccurrence / ai_freq_in_combination
                    
                    # AI→CICES提升度
                    expected_cices_prob = cices_freq_in_combination / sum(cices_counts.values())
                    lift_ai_to_cices = confidence_ai_to_cices / expected_cices_prob if expected_cices_prob > 0 else 0
                    
                    cices_combinations.append({
                        'CICES_Keyword': cices_kw,
                        'CICES_Category': cices_category,
                        'Cooccurrence': cooccurrence,
                        'Confidence_AI_to_CICES': round(confidence_ai_to_cices, 4),
                        'Lift_AI_to_CICES': round(lift_ai_to_cices, 4),
                        'Support': round(support, 4)
                    })
            
            # 按共现次数降序排序
            cices_combinations.sort(key=lambda x: x['Cooccurrence'], reverse=True)
            
            if cices_combinations:  # 只保留有CICES组合的AI关键词
                ai_to_cices_results.append({
                    'AI_Keyword': ai_kw,
                    'AI_Frequency': ai_counts[ai_kw],
                    'CICES_Combinations': cices_combinations
                })
        
        return cices_to_ai_results, ai_to_cices_results

    def generate_detailed_report(self, cices_to_ai_results, ai_to_cices_results, cices_ai_articles):
        """生成详细的分析报告"""
        print("\n" + "="*80)
        print("🔗 CICES与AI关键词组合详细分析报告")
        print("="*80)
        
        # 总体统计
        print(f"\n📊 一、总体统计")
        print(f"  ├─ 有关键词文献总数: {len(self.articles)} 篇")
        print(f"  ├─ CICES-AI组合文献数: {len(cices_ai_articles)} 篇")
        print(f"  ├─ 组合渗透率: {len(cices_ai_articles)/len(self.articles)*100:.2f}%")
        print(f"  ├─ 参与组合的CICES关键词数: {len(cices_to_ai_results)} 个")
        print(f"  └─ 参与组合的AI关键词数: {len(ai_to_cices_results)} 个")
        
        # CICES→AI分析
        print(f"\n🌱 二、CICES关键词适用的AI技术分析")
        print(f"{'='*80}")
        
        # 按CICES类别分组
        cices_by_category = defaultdict(list)
        for result in cices_to_ai_results:
            cices_by_category[result['CICES_Category']].append(result)
        
        for category in ['Provisioning', 'Regulating', 'Cultural']:
            if category in cices_by_category:
                print(f"\n📍 {category}类别:")
                print("-" * 60)
                
                # 按CICES关键词频次排序
                category_results = sorted(cices_by_category[category], 
                                        key=lambda x: x['CICES_Frequency'], reverse=True)
                
                for result in category_results:
                    cices_kw = result['CICES_Keyword']
                    cices_freq = result['CICES_Frequency']
                    print(f"\n🔹 {cices_kw} (频次: {cices_freq})")
                    print(f"  关联的AI技术:")
                    
                    for ai_combo in result['AI_Combinations']:
                        ai_kw = ai_combo['AI_Keyword']
                        cooccur = ai_combo['Cooccurrence']
                        conf = ai_combo['Confidence_CICES_to_AI']
                        lift = ai_combo['Lift_CICES_to_AI']
                        print(f"    • {ai_kw}: 共现{cooccur}次, 置信度{conf:.3f}, 提升度{lift:.3f}")
        
        # AI→CICES分析
        print(f"\n🤖 三、AI技术适用的CICES领域分析")
        print(f"{'='*80}")
        
        # 按AI技术类型分组（简单按关键词分组）
        ai_results_sorted = sorted(ai_to_cices_results, key=lambda x: x['AI_Frequency'], reverse=True)
        
        for result in ai_results_sorted:
            ai_kw = result['AI_Keyword']
            ai_freq = result['AI_Frequency']
            print(f"\n🔹 {ai_kw} (频次: {ai_freq})")
            print(f"  适用的CICES领域:")
            
            for cices_combo in result['CICES_Combinations']:
                cices_kw = cices_combo['CICES_Keyword']
                category = cices_combo['CICES_Category']
                cooccur = cices_combo['Cooccurrence']
                conf = cices_combo['Confidence_AI_to_CICES']
                lift = cices_combo['Lift_AI_to_CICES']
                print(f"    • {cices_kw} ({category}): 共现{cooccur}次, 置信度{conf:.3f}, 提升度{lift:.3f}")

    def create_visualizations(self, cices_to_ai_results, ai_to_cices_results, cices_ai_articles):
        """创建可视化图表"""
        fig = plt.figure(figsize=(20, 12))
        
        # 1. CICES类别AI技术分布
        plt.subplot(2, 3, 1)
        category_ai_counts = defaultdict(int)
        for result in cices_to_ai_results:
            category = result['CICES_Category']
            category_ai_counts[category] += len(result['AI_Combinations'])
        
        categories = list(category_ai_counts.keys())
        counts = list(category_ai_counts.values())
        colors = ['#FF6B6B', '#4ECDC4', '#45B7D1']
        
        plt.bar(categories, counts, color=colors, alpha=0.8)
        plt.xlabel('CICES Category', fontsize=12)
        plt.ylabel('Number of AI Technology Combinations', fontsize=12)
        plt.title('AI Technology Distribution by CICES Category', fontsize=14, fontweight='bold')
        
        # 2. 顶级AI技术使用频次
        plt.subplot(2, 3, 2)
        ai_freq_data = [(result['AI_Keyword'], result['AI_Frequency']) 
                        for result in ai_to_cices_results]
        ai_freq_data.sort(key=lambda x: x[1], reverse=True)
        
        top_ai = ai_freq_data[:10]
        ai_names = [ai[0][:15] + '...' if len(ai[0]) > 15 else ai[0] for ai, freq in top_ai]
        ai_freqs = [freq for ai, freq in top_ai]
        
        plt.barh(range(len(ai_names)), ai_freqs, color='#FF9800', alpha=0.8)
        plt.xlabel('Frequency', fontsize=12)
        plt.title('Top 10 AI Technologies by Frequency', fontsize=14, fontweight='bold')
        plt.yticks(range(len(ai_names)), ai_names)
        
        # 3. 顶级CICES关键词使用频次
        plt.subplot(2, 3, 3)
        cices_freq_data = [(result['CICES_Keyword'], result['CICES_Frequency']) 
                          for result in cices_to_ai_results]
        cices_freq_data.sort(key=lambda x: x[1], reverse=True)
        
        top_cices = cices_freq_data[:10]
        cices_names = [cices[:20] + '...' if len(cices) > 20 else cices for cices, freq in top_cices]
        cices_freqs = [freq for cices, freq in top_cices]
        
        plt.barh(range(len(cices_names)), cices_freqs, color='#4CAF50', alpha=0.8)
        plt.xlabel('Frequency', fontsize=12)
        plt.title('Top 10 CICES Keywords by Frequency', fontsize=14, fontweight='bold')
        plt.yticks(range(len(cices_names)), cices_names)
        
        # 4. 置信度分布（CICES→AI）
        plt.subplot(2, 3, 4)
        all_cices_confidences = []
        for result in cices_to_ai_results:
            for ai_combo in result['AI_Combinations']:
                all_cices_confidences.append(ai_combo['Confidence_CICES_to_AI'])
        
        plt.hist(all_cices_confidences, bins=20, alpha=0.7, color='#9C27B0', edgecolor='black')
        plt.xlabel('CICES→AI Confidence', fontsize=12)
        plt.ylabel('Frequency', fontsize=12)
        plt.title('Distribution of CICES→AI Confidence', fontsize=14, fontweight='bold')
        
        # 5. 置信度分布（AI→CICES）
        plt.subplot(2, 3, 5)
        all_ai_confidences = []
        for result in ai_to_cices_results:
            for cices_combo in result['CICES_Combinations']:
                all_ai_confidences.append(cices_combo['Confidence_AI_to_CICES'])
        
        plt.hist(all_ai_confidences, bins=20, alpha=0.7, color='#607D8B', edgecolor='black')
        plt.xlabel('AI→CICES Confidence', fontsize=12)
        plt.ylabel('Frequency', fontsize=12)
        plt.title('Distribution of AI→CICES Confidence', fontsize=14, fontweight='bold')
        
        # 6. 共现次数热力图（顶级组合）
        plt.subplot(2, 3, 6)
        
        # 构建热力图数据
        top_ai_names = [ai for ai, freq in ai_freq_data[:10]]
        top_cices_names = [cices for cices, freq in cices_freq_data[:10]]
        
        heatmap_data = np.zeros((len(top_cices_names), len(top_ai_names)))
        
        for i, cices_kw in enumerate(top_cices_names):
            for result in cices_to_ai_results:
                if result['CICES_Keyword'] == cices_kw:
                    for ai_combo in result['AI_Combinations']:
                        ai_kw = ai_combo['AI_Keyword']
                        if ai_kw in top_ai_names:
                            j = top_ai_names.index(ai_kw)
                            heatmap_data[i, j] = ai_combo['Cooccurrence']
        
        im = plt.imshow(heatmap_data, cmap='YlOrRd', aspect='auto')
        plt.colorbar(im)
        plt.xlabel('AI Technologies', fontsize=12)
        plt.ylabel('CICES Keywords', fontsize=12)
        plt.title('Co-occurrence Heatmap (Top 10)', fontsize=14, fontweight='bold')
        
        # 设置刻度标签
        plt.xticks(range(len(top_ai_names)), [ai[:10] + '...' if len(ai) > 10 else ai for ai in top_ai_names], rotation=45)
        plt.yticks(range(len(top_cices_names)), [cices[:15] + '...' if len(cices) > 15 else cices for cices in top_cices_names])
        
        plt.tight_layout()
        plt.savefig('./cices_ai_analysis.png', dpi=300, bbox_inches='tight')
        plt.show()

    def export_results(self, cices_to_ai_results, ai_to_cices_results, cices_ai_articles):
        """导出分析结果到Excel"""
        with pd.ExcelWriter('./cices_ai_analysis.xlsx', engine='openpyxl') as writer:
            
            # 1. 总体统计
            summary_data = {
                'Metric': [
                    '有关键词文献总数',
                    'CICES-AI组合文献数',
                    '组合渗透率(%)',
                    '参与组合的CICES关键词数',
                    '参与组合的AI关键词数'
                ],
                'Value': [
                    len(self.articles),
                    len(cices_ai_articles),
                    round(len(cices_ai_articles)/len(self.articles)*100, 2),
                    len(cices_to_ai_results),
                    len(ai_to_cices_results)
                ]
            }
            summary_df = pd.DataFrame(summary_data)
            summary_df.to_excel(writer, sheet_name='总体统计', index=False)
            
            # 2. CICES→AI详细分析
            cices_ai_data = []
            for result in cices_to_ai_results:
                cices_kw = result['CICES_Keyword']
                category = result['CICES_Category']
                cices_freq = result['CICES_Frequency']
                
                for ai_combo in result['AI_Combinations']:
                    cices_ai_data.append({
                        'CICES_Keyword': cices_kw,
                        'CICES_Category': category,
                        'CICES_Frequency': cices_freq,
                        'AI_Keyword': ai_combo['AI_Keyword'],
                        'Cooccurrence': ai_combo['Cooccurrence'],
                        'Confidence_CICES→AI': ai_combo['Confidence_CICES_to_AI'],
                        'Lift_CICES→AI': ai_combo['Lift_CICES_to_AI'],
                        'Support': ai_combo['Support']
                    })
            
            if cices_ai_data:
                cices_ai_df = pd.DataFrame(cices_ai_data)
                cices_ai_df.to_excel(writer, sheet_name='CICES→AI分析', index=False)
            
            # 3. AI→CICES详细分析
            ai_cices_data = []
            for result in ai_to_cices_results:
                ai_kw = result['AI_Keyword']
                ai_freq = result['AI_Frequency']
                
                for cices_combo in result['CICES_Combinations']:
                    ai_cices_data.append({
                        'AI_Keyword': ai_kw,
                        'AI_Frequency': ai_freq,
                        'CICES_Keyword': cices_combo['CICES_Keyword'],
                        'CICES_Category': cices_combo['CICES_Category'],
                        'Cooccurrence': cices_combo['Cooccurrence'],
                        'Confidence_AI→CICES': cices_combo['Confidence_AI_to_CICES'],
                        'Lift_AI→CICES': cices_combo['Lift_AI_to_CICES'],
                        'Support': cices_combo['Support']
                    })
            
            if ai_cices_data:
                ai_cices_df = pd.DataFrame(ai_cices_data)
                ai_cices_df.to_excel(writer, sheet_name='AI→CICES分析', index=False)
            
            # 4. 按CICES类别分组的汇总
            category_summary = []
            cices_by_category = defaultdict(list)
            for result in cices_to_ai_results:
                cices_by_category[result['CICES_Category']].append(result)
            
            for category, results in cices_by_category.items():
                total_cices = len(results)
                total_ai_combinations = sum(len(r['AI_Combinations']) for r in results)
                avg_combinations_per_cices = total_ai_combinations / total_cices if total_cices > 0 else 0
                
                category_summary.append({
                    'CICES_Category': category,
                    'CICES_Keywords_Count': total_cices,
                    'Total_AI_Combinations': total_ai_combinations,
                    'Avg_AI_per_CICES': round(avg_combinations_per_cices, 2)
                })
            
            if category_summary:
                category_df = pd.DataFrame(category_summary)
                category_df.to_excel(writer, sheet_name='CICES类别汇总', index=False)
        
        print("\n✅ 分析结果已导出到 cices_ai_analysis.xlsx")

    def run_analysis(self):
        """运行完整的CICES-AI分析"""
        print("🔗 开始CICES与AI关键词组合分析...")
        print("="*80)
        
        # 步骤1: 读取文献
        self.read_literature_files()
        
        if not self.articles:
            print("❌ 未找到有效的文章数据！")
            return
        
        # 步骤2: 识别CICES-AI组合文献
        cices_ai_articles = self.identify_cices_ai_articles()
        
        if not cices_ai_articles:
            print("❌ 未找到CICES-AI组合文献！")
            return
        
        # 步骤3: 分析组合关系
        print("📊 正在分析CICES-AI组合关系...")
        cices_ai_cooccurrence, cices_counts, ai_counts = self.analyze_cices_ai_combinations(cices_ai_articles)
        
        # 步骤4: 计算关联度指标
        print("📊 正在计算关联度指标...")
        cices_to_ai_results, ai_to_cices_results = self.calculate_association_metrics(
            cices_ai_cooccurrence, cices_counts, ai_counts, len(cices_ai_articles)
        )
        
        # 步骤5: 生成详细报告
        self.generate_detailed_report(cices_to_ai_results, ai_to_cices_results, cices_ai_articles)
        
        # 步骤6: 创建可视化
        print("\n📊 正在创建可视化图表...")
        self.create_visualizations(cices_to_ai_results, ai_to_cices_results, cices_ai_articles)
        
        # 步骤7: 导出结果
        print("💾 正在导出分析结果...")
        self.export_results(cices_to_ai_results, ai_to_cices_results, cices_ai_articles)
        
        print("\n" + "="*80)
        print("✅ CICES-AI组合分析完成！生成的文件:")
        print("  📊 cices_ai_analysis.png")
        print("  📁 cices_ai_analysis.xlsx")
        print("="*80)

# 运行分析
if __name__ == "__main__":
    analyzer = CICESAIAnalyzer()
    analyzer.run_analysis()
```

```python
import os
import re
import pandas as pd
from collections import Counter

class KeywordFrequencyAnalyzer:
    def __init__(self, data_directory="./"):
        self.data_directory = data_directory
        self.all_articles = []
        self.articles = []  # 所有有关键词的文献
        
        # AI关键词列表（不包括传统的regression, classification, clustering）
        self.ai_keywords = [
            "machine learning", "supervised learning", "unsupervised learning", 
            "association rules", "dimensionality reduction", 
            "semi supervised learning", "self training", "low density separation models", 
            "graph based algorithms", "generative models", "reinforcement learning", 
            "deep learning", "deep feed forward networks", "convolution neural networks", 
            "siamese neural networks", "transformers", "generative adversarial networks", 
            "graph neural networks", "recurrent neural network", "active learning", 
            "uncertainty sampling", "diversity sampling", "transfer learning", 
            "instance-based transfer learning", "feature-based transfer learning", 
            "parameter-based transfer learning", "continual learning", 
            "regularization-based continual learning", "memory reply", "dynamic architectures", 
            "natural language processing", "machine translation", "question answering", 
            "sentiment analysis", "text generation", "topic modeling", "graphplan", 
            "satplan", "slam algorithms", "fastslam", "graphslam", "motion planning algorithms", 
            "sampling-based algorithms", "rrt", "graph search algorithms", "a-star", 
            "reaction-based algorithms", "dwa", "interpolating curve algorithms", 
            "image recognition", "machine vision", "nerf", "vision transformer", 
            "tacotron", "vits", "wavenet", "samplern", "expert systems", 
            "frame-based systems", "certainty factors", "RNN", "CNN", "LSTM", 
            "Long Short-Term Memory", "ML", "DL", "RL"
        ]
        
        # CICES分类关键词
        self.cices_categories = {
            'Provisioning': {
                'terrestrial_crops': [
                    'crops', 'cereals', 'root crops', 'soft fruit', 'fodder', 'livestock feed',
                    'fibres', 'materials', 'cultivated plants', 'fungi', 'algae', 'bacteria',
                    'plantation', 'energy source', 'bioenergy'
                ],
                'aquatic_crops': [
                    'aquaculture', 'in-situ aquaculture', 'cultivated aquatic plants'
                ],
                'reared_animals': [
                    'reared animals', 'beef', 'dairy', 'animal products',
                    'fibres from animals', 'mechanical energy', 'biogas'
                ],
                'aquatic_animals': [
                    'reared aquatic animals', 'in-situ aquaculture animals'
                ],
                'wild_plants': [
                    'wild plants', 'terrestrial plants', 'aquatic plants', 'wild fungi', 'wild algae',
                    'wild plant materials', 'wild plant energy'
                ],
                'wild_animals': [
                    'wild animals', 'terrestrial animals', 'aquatic animals',
                    'wild animal nutrition', 'wild animal materials'
                ],
                'genetic_material': [
                    'genetic material', 'seeds', 'spores', 'plant materials',
                    'breeding', 'new strains', 'varieties',
                    'individual genes', 'biological entities',
                    'animal genetic material', 'population establishment'
                ]
            },
            'Regulating': {
                'biochemical_transformation': [
                    'bio-remediation', 'micro-organisms', 'pollution degradation',
                    'filtration', 'sequestration', 'storage', 'accumulation',
                    'waste mediation', 'toxic substances', 'anthropogenic origin',
                    'smell reduction', 'noise attenuation', 'visual screening'
                ],
                'flow_extreme_events': [
                    'erosion control', 'water erosion', 'wind erosion',
                    'hydrological cycle', 'water flow regulation', 'runoff', 'base flows', 'peak flows',
                    'hazard mitigation', 'mass movement', 'landslides', 'flood mitigation', 'surge protection',
                    'wind protection', 'fire protection'
                ],
                'lifecycle_maintenance': [
                    'pollination', 'gamete dispersal', 'seed dispersal',
                    'nursery populations', 'breeding grounds', 'gene pool protection',
                    'refuge habitats', 'feeding grounds'
                ],
                'pest_disease_control': [
                    'pest control', 'invasive species', 'disease control'
                ],
                'soil_quality': [
                    'soil quality', 'weathering processes', 'decomposition', 'fixing processes',
                    'soil structure', 'biological agents', 'soil porosity'
                ],
                'water_conditions': [
                    'water conditions', 'chemical condition', 'macronutrients',
                    'freshwater regulation', 'saltwater regulation'
                ],
                'atmospheric_conditions': [
                    'atmospheric composition', 'oceanic circulation', 'climate change',
                    'temperature regulation', 'humidity', 'ventilation', 'transpiration', 'microclimate'
                ]
            },
            'Cultural': {
                'physical_experiential': [
                    'recreational activities', 'health promotion', 'recuperation', 'enjoyment',
                    'active interactions', 'immersive interactions',
                    'passive interactions', 'observational interactions'
                ],
                'intellectual_representative': [
                    'scientific investigation', 'traditional ecological knowledge', 'genetic diversity',
                    'education', 'training',
                    'culture', 'heritage', 'cultural resonance',
                    'aesthetic experiences'
                ],
                'indirect_interactions': [
                    'entertainment', 'representation outside setting'
                ],
                'spiritual_symbolic': [
                    'symbolic meaning', 'sense of place', 'distinctiveness',
                    'spiritual meaning', 'religious meaning',
                    'contemporary existence', 'conservation importance',
                    'inter-generational existence', 'genetic diversity conservation'
                ]
            }
        }
        
        # 构建CICES关键词到类别的映射
        self.cices_keyword_to_category = {}
        self.all_cices_keywords = []
        for main_category, subcategories in self.cices_categories.items():
            for subcategory, keywords in subcategories.items():
                for keyword in keywords:
                    keyword_lower = keyword.lower()
                    self.cices_keyword_to_category[keyword_lower] = main_category
                    self.all_cices_keywords.append(keyword_lower)

    def read_literature_files(self):
        """读取所有文献文件"""
        files = []
        for f in os.listdir(self.data_directory):
            if f.endswith('.txt') and 'download' in f and any(year in f for year in 
                ['2015', '2016', '2017', '2018', '2019', '2020', '2021', '2022', '2023', '2024', '2025']):
                files.append(f)
        
        files.sort()
        print(f"找到 {len(files)} 个文献文件")
        
        for file in files:
            file_path = os.path.join(self.data_directory, file)
            try:
                with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
                    content = f.read()
                    self.parse_articles(content, file)
            except Exception as e:
                print(f"读取文件 {file} 时出错: {e}")
        
        print(f"总共解析了 {len(self.all_articles)} 篇文章")
        print(f"有关键词的文章: {len(self.articles)} 篇")

    def parse_articles(self, content, filename):
        """解析文章内容"""
        articles = re.split(r'\nER\s*\n', content)
        
        for article_text in articles:
            if not article_text.strip():
                continue
            
            if not re.search(r'(^|\n)PT\s+', article_text):
                continue
            
            article_data = {
                'filename': filename,
                'keywords': set(),
                'keywords_plus': set()
            }
            
            # 提取ID字段
            id_match = re.search(r'\nID\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if id_match:
                id_content = id_match.group(1)
                keywords_plus = self.extract_keywords_from_field(id_content)
                article_data['keywords_plus'].update(keywords_plus)
            
            # 提取DE字段
            de_match = re.search(r'\nDE\s+(.+?)(?=\n[A-Z]{1,2}\s|\nER|\Z)', article_text, re.DOTALL)
            if de_match:
                de_content = de_match.group(1)
                keywords = self.extract_keywords_from_field(de_content)
                article_data['keywords'].update(keywords)
            
            self.all_articles.append(article_data)
            
            if article_data['keywords'] or article_data['keywords_plus']:
                article_data['all_keywords'] = article_data['keywords'] | article_data['keywords_plus']
                self.articles.append(article_data)

    def extract_keywords_from_field(self, field_content):
        """从ID或DE字段内容中提取关键词"""
        if not field_content:
            return []
            
        field_content = field_content.replace('\n', ' ')
        keywords = re.split(r'[;\n]', field_content)
        
        cleaned_keywords = []
        for keyword in keywords:
            keyword = keyword.strip().lower()
            keyword = re.sub(r'\s+', ' ', keyword)
            keyword = keyword.strip('.,;()[]{}"\'-')
            
            if keyword and len(keyword) > 2:
                cleaned_keywords.append(keyword)
                
        return cleaned_keywords

    def analyze_keyword_frequencies(self):
        """统计CICES关键词和AI技术的频率"""
        cices_counter = Counter()
        ai_counter = Counter()
        
        print("正在统计关键词频率...")
        
        for article in self.articles:
            keywords_text = ' '.join(article['all_keywords']).lower()
            
            # 统计CICES关键词频率
            for cices_keyword in self.all_cices_keywords:
                if cices_keyword in keywords_text:
                    cices_counter[cices_keyword] += 1
            
            # 统计AI关键词频率
            for ai_keyword in self.ai_keywords:
                if ai_keyword.lower() in keywords_text:
                    ai_counter[ai_keyword] += 1
        
        return cices_counter, ai_counter

    def generate_frequency_report(self, cices_counter, ai_counter):
        """生成频率统计报告"""
        print("\n" + "="*80)
        print("CICES关键词与AI技术频率统计报告")
        print("="*80)
        
        # 总体统计
        print(f"\n一、总体统计")
        print(f"  - 有关键词文献总数: {len(self.articles)} 篇")
        print(f"  - 出现的CICES关键词数: {len(cices_counter)} 个")
        print(f"  - 出现的AI技术数: {len(ai_counter)} 个")
        
        # CICES关键词频率统计（按类别分组）
        print(f"\n二、CICES关键词频率统计（按类别分组）")
        print("="*60)
        
        # 按类别分组CICES关键词
        cices_by_category = {'Provisioning': [], 'Regulating': [], 'Cultural': []}
        
        for keyword, freq in cices_counter.items():
            category = self.cices_keyword_to_category.get(keyword, 'Unknown')
            if category in cices_by_category:
                cices_by_category[category].append((keyword, freq))
        
        # 输出各类别的关键词
        for category in ['Provisioning', 'Regulating', 'Cultural']:
            if cices_by_category[category]:
                # 按频率降序排序
                cices_by_category[category].sort(key=lambda x: x[1], reverse=True)
                
                print(f"\n{category}类别:")
                print("-" * 50)
                print(f"{'排名':<4} {'关键词':<35} {'频次':<8} {'使用率(%)':<10}")
                print("-" * 50)
                
                for i, (keyword, freq) in enumerate(cices_by_category[category], 1):
                    usage_rate = (freq / len(self.articles)) * 100
                    print(f"{i:<4} {keyword:<35} {freq:<8} {usage_rate:<10.2f}")
        
        # AI技术频率统计
        print(f"\n三、AI技术频率统计")
        print("="*60)
        print(f"{'排名':<4} {'AI技术':<40} {'频次':<8} {'使用率(%)':<10}")
        print("-" * 60)
        
        ai_sorted = ai_counter.most_common()
        for i, (ai_tech, freq) in enumerate(ai_sorted, 1):
            usage_rate = (freq / len(self.articles)) * 100
            print(f"{i:<4} {ai_tech:<40} {freq:<8} {usage_rate:<10.2f}")

    def export_frequency_results(self, cices_counter, ai_counter):
        """导出频率统计结果到Excel"""
        with pd.ExcelWriter('./keyword_frequency_analysis.xlsx', engine='openpyxl') as writer:
            
            # 1. 总体统计
            summary_data = {
                'Metric': [
                    '有关键词文献总数',
                    '出现的CICES关键词数',
                    '出现的AI技术数',
                    'CICES关键词总出现次数',
                    'AI技术总出现次数'
                ],
                'Value': [
                    len(self.articles),
                    len(cices_counter),
                    len(ai_counter),
                    sum(cices_counter.values()),
                    sum(ai_counter.values())
                ]
            }
            summary_df = pd.DataFrame(summary_data)
            summary_df.to_excel(writer, sheet_name='总体统计', index=False)
            
            # 2. CICES关键词频率统计
            cices_data = []
            for keyword, freq in cices_counter.items():
                category = self.cices_keyword_to_category.get(keyword, 'Unknown')
                usage_rate = (freq / len(self.articles)) * 100
                cices_data.append({
                    'CICES_Keyword': keyword,
                    'Category': category,
                    'Frequency': freq,
                    'Usage_Rate_%': round(usage_rate, 2)
                })
            
            # 按频率降序排序
            cices_data.sort(key=lambda x: x['Frequency'], reverse=True)
            cices_df = pd.DataFrame(cices_data)
            cices_df.to_excel(writer, sheet_name='CICES关键词频率', index=False)
            
            # 3. AI技术频率统计
            ai_data = []
            for ai_tech, freq in ai_counter.items():
                usage_rate = (freq / len(self.articles)) * 100
                ai_data.append({
                    'AI_Technology': ai_tech,
                    'Frequency': freq,
                    'Usage_Rate_%': round(usage_rate, 2)
                })
            
            # 按频率降序排序
            ai_data.sort(key=lambda x: x['Frequency'], reverse=True)
            ai_df = pd.DataFrame(ai_data)
            ai_df.to_excel(writer, sheet_name='AI技术频率', index=False)
            
            # 4. 按CICES类别汇总
            category_summary = []
            cices_by_category = {'Provisioning': [], 'Regulating': [], 'Cultural': []}
            
            for keyword, freq in cices_counter.items():
                category = self.cices_keyword_to_category.get(keyword, 'Unknown')
                if category in cices_by_category:
                    cices_by_category[category].append((keyword, freq))
            
            for category, keywords_freqs in cices_by_category.items():
                if keywords_freqs:
                    total_freq = sum(freq for _, freq in keywords_freqs)
                    keyword_count = len(keywords_freqs)
                    avg_freq = total_freq / keyword_count if keyword_count > 0 else 0
                    
                    category_summary.append({
                        'Category': category,
                        'Keywords_Count': keyword_count,
                        'Total_Frequency': total_freq,
                        'Average_Frequency': round(avg_freq, 2)
                    })
            
            if category_summary:
                category_df = pd.DataFrame(category_summary)
                category_df.to_excel(writer, sheet_name='CICES类别汇总', index=False)
        
        print("\n分析结果已导出到 keyword_frequency_analysis.xlsx")

    def run_analysis(self):
        """运行完整的频率分析"""
        print("开始CICES关键词与AI技术频率统计分析...")
        print("="*60)
        
        # 步骤1: 读取文献
        self.read_literature_files()
        
        if not self.articles:
            print("未找到有效的文章数据！")
            return
        
        # 步骤2: 统计关键词频率
        cices_counter, ai_counter = self.analyze_keyword_frequencies()
        
        # 步骤3: 生成报告
        self.generate_frequency_report(cices_counter, ai_counter)
        
        # 步骤4: 导出结果
        print("\n正在导出分析结果...")
        self.export_frequency_results(cices_counter, ai_counter)
        
        print("\n" + "="*60)
        print("频率统计分析完成！生成的文件:")
        print("  keyword_frequency_analysis.xlsx")
        print("="*60)

# 运行分析
if __name__ == "__main__":
    analyzer = KeywordFrequencyAnalyzer()
    analyzer.run_analysis()
```
